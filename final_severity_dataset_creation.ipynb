```python

# ============================================================
# METHOD 2: POTATO LEAF SEGMENTATION V5
# Shadow-Aware Lesion-Guided GrabCut
#
# Main improvements over V4:
#   1. Better shadow-component aggregation
#   2. Original-shadow evidence retained after morphology
#   3. Thicker boundary support region
#   4. Deep-shadow fallback using distance transform
#   5. Safer shadow removal
#   6. Lesion-guided GrabCut retained
#   7. Contour cleanup
#   8. Detailed debug statistics
#   9. Severity calculated from lesion ∩ leaf
# ============================================================


# ============================================================
# STEP 0 — GOOGLE DRIVE + ZIP EXTRACTION
# ============================================================

from google.colab import drive
drive.mount('/content/drive')

from pathlib import Path
import zipfile
import os

ZIP_PATH = Path("/content/drive/MyDrive/Potato_Severity_Dataset.zip")
EXTRACT_DIR = Path("/content/Potato_Severity_Dataset")

if not ZIP_PATH.exists():
    raise FileNotFoundError(
        f"ZIP file not found:\n{ZIP_PATH}"
    )

print("ZIP found:", ZIP_PATH)
print(
    f"ZIP size: "
    f"{ZIP_PATH.stat().st_size / (1024**3):.2f} GB"
)

EXTRACT_DIR.mkdir(parents=True, exist_ok=True)

print("\nExtracting dataset...")

with zipfile.ZipFile(ZIP_PATH, "r") as z:
    z.extractall(EXTRACT_DIR)

print("Extraction completed.")
print("Dataset root:", EXTRACT_DIR)


# ============================================================
# STEP 1 — IMPORTS
# ============================================================

import cv2
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from pathlib import Path
import math
import warnings

warnings.filterwarnings("ignore")


# ============================================================
# STEP 2 — PATH CONFIGURATION
# ============================================================

ROOT = Path("/content/Potato_Severity_Dataset")

VALID_IMAGES_DIR = ROOT / "valid" / "images"
VALID_MASKS_DIR  = ROOT / "valid" / "lesion_masks"

OUTPUT_DIR = ROOT / "leaf_segmentation_v5"

MASK_DIR      = OUTPUT_DIR / "leaf_masks"
OVERLAY_DIR   = OUTPUT_DIR / "overlays"
COMBINED_DIR  = OUTPUT_DIR / "combined"
DEBUG_DIR     = OUTPUT_DIR / "debug"
BEFORE_DIR    = DEBUG_DIR / "before_shadow"
SHADOW_DIR    = DEBUG_DIR / "shadow"
FINAL_DIR     = DEBUG_DIR / "final"

for d in [
    OUTPUT_DIR,
    MASK_DIR,
    OVERLAY_DIR,
    COMBINED_DIR,
    DEBUG_DIR,
    BEFORE_DIR,
    SHADOW_DIR,
    FINAL_DIR,
]:
    d.mkdir(parents=True, exist_ok=True)


# ============================================================
# STEP 3 — GENERAL PARAMETERS
# ============================================================

# --------------------------
# Image / GrabCut
# --------------------------

BORDER_SIZE = 8
GRABCUT_ITERATIONS = 7

MIN_COMPONENT_AREA_RATIO = 0.003

# --------------------------
# Component scoring
# --------------------------

W_LESION = 0.45
W_COLOR  = 0.25
W_CENTER = 0.15
W_BORDER = 0.15

# --------------------------
# QC
# --------------------------

MIN_LEAF_AREA_RATIO = 0.03
MAX_LEAF_AREA_RATIO = 0.90

# ============================================================
# SHADOW PARAMETERS
# ============================================================

# These preserve your validated V4 thresholds.
SHADOW_V_MAX = 90
SHADOW_S_MAX = 85
SHADOW_CHROMA_MAX = 28

# Fragmentation repair
SHADOW_CLOSE_KERNEL = 7
SHADOW_CLOSE_ITERATIONS = 2

# Component gate
SHADOW_MIN_AREA = 40

# Original evidence required after morphology
SHADOW_MIN_ORIGINAL_RATIO = 0.35

# Original V4 value retained
SHADOW_MIN_RATIO = 0.55

# Boundary support
SHADOW_BOUNDARY_TOUCH_RATIO = 0.02
SHADOW_BOUNDARY_DILATION_KERNEL = 7

# Deep shadow
SHADOW_MIN_DEPTH = 3.0

# Safety
MAX_SHADOW_REMOVAL_RATIO = 0.25

# ============================================================
# CONTOUR PARAMETERS
# ============================================================

MIN_FINAL_COMPONENT_RATIO = 0.003

CONTOUR_CLOSE_KERNEL = 5
CONTOUR_OPEN_KERNEL = 3

# Remove tiny isolated structures
MIN_COMPONENT_KEEP_RATIO = 0.003

# ============================================================
# PILOT
# ============================================================

# Set None -> process all
PILOT_N = None

IMAGE_EXTS = {
    ".jpg",
    ".jpeg",
    ".png",
    ".bmp",
    ".webp"
}


# ============================================================
# STEP 4 — IMAGE I/O
# ============================================================

def read_rgb(path):
    """
    Read image as RGB uint8.
    """
    img = cv2.imread(
        str(path),
        cv2.IMREAD_COLOR
    )

    if img is None:
        raise FileNotFoundError(
            f"Could not read image:\n{path}"
        )

    return cv2.cvtColor(
        img,
        cv2.COLOR_BGR2RGB
    )


def read_binary_mask(path, shape):
    """
    Read lesion mask and resize using nearest-neighbor.
    """
    mask = cv2.imread(
        str(path),
        cv2.IMREAD_GRAYSCALE
    )

    if mask is None:
        raise FileNotFoundError(
            f"Could not read mask:\n{path}"
        )

    if mask.shape != shape:
        mask = cv2.resize(
            mask,
            (shape[1], shape[0]),
            interpolation=cv2.INTER_NEAREST
        )

    return (
        (mask > 127)
        .astype(np.uint8)
        * 255
    )


def infer_class(path):
    text = str(path).lower()

    if "healthy" in text:
        return "Potato Healthy"

    if (
        "late_blight" in text
        or "late-blight" in text
        or "lateblight" in text
    ):
        return "Potato Late Blight"

    return "Unknown"


def has_lesion(mask):
    if mask is None:
        return False

    return bool(
        np.any(mask > 0)
    )


# ============================================================
# STEP 5 — COLOR CANDIDATE
# ============================================================

def green_candidate(rgb):

    hsv = cv2.cvtColor(
        rgb,
        cv2.COLOR_RGB2HSV
    )

    lab = cv2.cvtColor(
        rgb,
        cv2.COLOR_RGB2LAB
    )

    h, s, v = cv2.split(hsv)
    L, a, b = cv2.split(lab)

    condition_1 = (
        (h >= 28) &
        (h <= 92) &
        (s >= 50) &
        (v >= 35)
    )

    condition_2 = (
        (a <= 140) &
        (s >= 45) &
        (v >= 35)
    )

    candidate = (
        condition_1 |
        condition_2
    )

    mask = (
        candidate
        .astype(np.uint8)
        * 255
    )

    open_kernel = cv2.getStructuringElement(
        cv2.MORPH_ELLIPSE,
        (3, 3)
    )

    close_kernel = cv2.getStructuringElement(
        cv2.MORPH_ELLIPSE,
        (5, 5)
    )

    mask = cv2.morphologyEx(
        mask,
        cv2.MORPH_OPEN,
        open_kernel,
        iterations=1
    )

    mask = cv2.morphologyEx(
        mask,
        cv2.MORPH_CLOSE,
        close_kernel,
        iterations=1
    )

    return mask


# ============================================================
# STEP 6 — GENERIC LEAF CANDIDATE
# ============================================================

def generic_leaf_candidate(rgb):

    h_img, w_img = rgb.shape[:2]

    lab = cv2.cvtColor(
        rgb,
        cv2.COLOR_RGB2LAB
    )

    hsv = cv2.cvtColor(
        rgb,
        cv2.COLOR_RGB2HSV
    )

    L, A, B = cv2.split(lab)
    H, S, V = cv2.split(hsv)

    # --------------------------------------------------------
    # Border
    # --------------------------------------------------------

    b = BORDER_SIZE

    border_mask = np.zeros(
        (h_img, w_img),
        dtype=bool
    )

    border_mask[:b, :] = True
    border_mask[-b:, :] = True
    border_mask[:, :b] = True
    border_mask[:, -b:] = True

    border_lab = lab[
        border_mask
    ].astype(np.float32)

    median_lab = np.median(
        border_lab,
        axis=0
    )

    std_lab = (
        np.std(
            border_lab,
            axis=0
        ) + 5.0
    )

    diff_L = (
        L.astype(np.float32)
        - median_lab[0]
    ) / std_lab[0]

    diff_A = (
        A.astype(np.float32)
        - median_lab[1]
    ) / std_lab[1]

    diff_B = (
        B.astype(np.float32)
        - median_lab[2]
    ) / std_lab[2]

    color_distance = np.sqrt(
        diff_L**2 +
        diff_A**2 +
        diff_B**2
    )

    # --------------------------------------------------------
    # Existing V4 evidence
    # --------------------------------------------------------

    dark_difference = (
        median_lab[0]
        - L.astype(np.float32)
    )

    dark_leaf = (
        dark_difference > 10
    )

    saturated = (
        S >= 35
    )

    strong_difference = (
        color_distance > 2.0
    )

    candidate = (
        strong_difference &
        (
            dark_leaf |
            saturated
        )
    )

    # Existing green candidate
    green = green_candidate(rgb)

    candidate = (
        candidate |
        (green > 0)
    )

    # Remove image border
    candidate[:b, :] = False
    candidate[-b:, :] = False
    candidate[:, :b] = False
    candidate[:, -b:] = False

    mask = (
        candidate
        .astype(np.uint8)
        * 255
    )

    # Morphology
    open_kernel = cv2.getStructuringElement(
        cv2.MORPH_ELLIPSE,
        (3, 3)
    )

    close_kernel = cv2.getStructuringElement(
        cv2.MORPH_ELLIPSE,
        (5, 5)
    )

    mask = cv2.morphologyEx(
        mask,
        cv2.MORPH_OPEN,
        open_kernel,
        iterations=1
    )

    mask = cv2.morphologyEx(
        mask,
        cv2.MORPH_CLOSE,
        close_kernel,
        iterations=2
    )

    return mask


# ============================================================
# STEP 7 — CENTER SCORE
# ============================================================

def center_score(component):

    ys, xs = np.where(
        component > 0
    )

    if len(xs) == 0:
        return 0.0

    h, w = component.shape

    cx = xs.mean()
    cy = ys.mean()

    center_x = (
        (w - 1) / 2.0
    )

    center_y = (
        (h - 1) / 2.0
    )

    dx = (
        (cx - center_x)
        / (w / 2.0)
    )

    dy = (
        (cy - center_y)
        / (h / 2.0)
    )

    distance = np.sqrt(
        dx**2 + dy**2
    )

    return float(
        1.0 - min(
            distance,
            1.0
        )
    )


# ============================================================
# STEP 8 — BORDER CONTACT
# ============================================================

def border_contact_ratio(
    component,
    border_size=BORDER_SIZE
):

    h, w = component.shape

    border = np.zeros(
        (h, w),
        dtype=bool
    )

    border[:border_size, :] = True
    border[-border_size:, :] = True
    border[:, :border_size] = True
    border[:, -border_size:] = True

    comp = component > 0

    area = comp.sum()

    if area == 0:
        return 1.0

    contact = (
        np.logical_and(
            comp,
            border
        ).sum()
    )

    return float(
        contact / area
    )


# ============================================================
# STEP 9 — SELECT BEST LEAF COMPONENT
# ============================================================

def select_leaf_component(
    binary_mask,
    lesion_mask,
    generic_mask
):

    binary = (
        binary_mask > 0
    ).astype(np.uint8)

    n_labels, labels, stats, _ = (
        cv2.connectedComponentsWithStats(
            binary,
            connectivity=8
        )
    )

    if n_labels <= 1:
        return binary_mask.copy()

    lesion_bool = (
        lesion_mask > 0
    )

    generic_bool = (
        generic_mask > 0
    )

    total_pixels = binary.size

    lesion_total = lesion_bool.sum()

    best_label = None
    best_score = -999999.0

    for label in range(
        1,
        n_labels
    ):

        area = stats[
            label,
            cv2.CC_STAT_AREA
        ]

        if (
            area <
            total_pixels *
            MIN_COMPONENT_AREA_RATIO
        ):
            continue

        component = (
            labels == label
        )

        # ----------------------------------------------------
        # Lesion coverage
        # ----------------------------------------------------

        if lesion_total > 0:

            lesion_inside = (
                np.logical_and(
                    component,
                    lesion_bool
                ).sum()
            )

            lesion_coverage = (
                lesion_inside /
                lesion_total
            )

        else:
            lesion_coverage = 0.0

        # ----------------------------------------------------
        # Generic candidate fraction
        # ----------------------------------------------------

        generic_inside = (
            np.logical_and(
                component,
                generic_bool
            ).sum()
        )

        generic_fraction = (
            generic_inside /
            area
        )

        # ----------------------------------------------------
        # Geometry
        # ----------------------------------------------------

        c_score = center_score(
            component.astype(
                np.uint8
            )
        )

        b_score = border_contact_ratio(
            component.astype(
                np.uint8
            )
        )

        # ----------------------------------------------------
        # Final score
        # ----------------------------------------------------

        if lesion_total > 0:

            score = (
                W_LESION *
                lesion_coverage
                +
                W_COLOR *
                generic_fraction
                +
                W_CENTER *
                c_score
                -
                W_BORDER *
                b_score
            )

        else:

            score = (
                0.60 *
                generic_fraction
                +
                0.25 *
                c_score
                -
                0.15 *
                b_score
            )

        if score > best_score:

            best_score = score
            best_label = label

    # fallback
    if best_label is None:

        areas = stats[
            1:,
            cv2.CC_STAT_AREA
        ]

        best_label = (
            1 +
            np.argmax(areas)
        )

    return (
        labels == best_label
    ).astype(np.uint8) * 255


# ============================================================
# STEP 10 — SHADOW DETECTION
# ============================================================

def detect_shadow(rgb, leaf_mask, lesion_mask=None):

    h, w = rgb.shape[:2]

    leaf = (
        leaf_mask > 0
    )

    if lesion_mask is None:

        lesion = np.zeros(
            (h, w),
            dtype=bool
        )

    else:

        lesion = (
            lesion_mask > 0
        )

    hsv = cv2.cvtColor(
        rgb,
        cv2.COLOR_RGB2HSV
    )

    lab = cv2.cvtColor(
        rgb,
        cv2.COLOR_RGB2LAB
    )

    H, S, V = cv2.split(hsv)

    L, A, B = cv2.split(lab)

    chroma = np.sqrt(
        (
            A.astype(np.float32)
            - 128.0
        ) ** 2
        +
        (
            B.astype(np.float32)
            - 128.0
        ) ** 2
    )

    # --------------------------------------------------------
    # Existing validated threshold
    # --------------------------------------------------------

    dark = (
        V <= SHADOW_V_MAX
    )

    low_sat = (
        S <= SHADOW_S_MAX
    )

    low_chroma = (
        chroma <= SHADOW_CHROMA_MAX
    )

    shadow_like = (
        dark &
        low_sat &
        low_chroma
    )

    # Only current leaf
    shadow_like &= leaf

    # Never classify known lesion as shadow
    shadow_like &= ~lesion

    return shadow_like.astype(
        np.uint8
    ) * 255


# ============================================================
# STEP 11 — SHADOW COMPONENT REFINEMENT
# ============================================================

def remove_conservative_shadow(
    rgb,
    leaf_mask,
    lesion_mask=None,
    debug=False
):

    refined = leaf_mask.copy()

    h, w = rgb.shape[:2]

    leaf = (
        refined > 0
    )

    if lesion_mask is None:

        lesion = np.zeros(
            (h, w),
            dtype=bool
        )

    else:

        lesion = (
            lesion_mask > 0
        )

    original_leaf_area = int(
        leaf.sum()
    )

    if original_leaf_area == 0:
        return refined, None

    # ========================================================
    # ORIGINAL SHADOW DETECTION
    # ========================================================

    shadow_original_u8 = detect_shadow(
        rgb,
        refined,
        lesion_mask
    )

    shadow_original = (
        shadow_original_u8 > 0
    )

    original_shadow_pixels = int(
        shadow_original.sum()
    )

    if original_shadow_pixels == 0:

        debug_info = {
            "original_shadow_pixels": 0,
            "closed_shadow_pixels": 0,
            "components_total": 0,
            "components_removed": 0,
            "removed_pixels": 0,
        }

        return refined, debug_info

    # ========================================================
    # FIX 1 — MERGE FRAGMENTED SHADOW
    # ========================================================

    close_kernel = cv2.getStructuringElement(
        cv2.MORPH_ELLIPSE,
        (
            SHADOW_CLOSE_KERNEL,
            SHADOW_CLOSE_KERNEL
        )
    )

    shadow_closed_u8 = cv2.morphologyEx(
        shadow_original_u8,
        cv2.MORPH_CLOSE,
        close_kernel,
        iterations=SHADOW_CLOSE_ITERATIONS
    )

    shadow_closed = (
        shadow_closed_u8 > 0
    )

    # Keep it inside leaf
    shadow_closed &= leaf

    # Never cross lesion
    shadow_closed &= ~lesion

    # ========================================================
    # FIX 2 — THICKER BOUNDARY REGION
    # ========================================================

    boundary = cv2.morphologyEx(
        refined,
        cv2.MORPH_GRADIENT,
        np.ones(
            (3, 3),
            np.uint8
        )
    ) > 0

    boundary = cv2.dilate(
        boundary.astype(np.uint8),
        cv2.getStructuringElement(
            cv2.MORPH_ELLIPSE,
            (
                SHADOW_BOUNDARY_DILATION_KERNEL,
                SHADOW_BOUNDARY_DILATION_KERNEL
            )
        )
    ) > 0

    # ========================================================
    # CONNECTED COMPONENTS
    # ========================================================

    n_labels, labels, stats, _ = (
        cv2.connectedComponentsWithStats(
            shadow_closed.astype(
                np.uint8
            ),
            connectivity=8
        )
    )

    shadow_remove = np.zeros(
        (h, w),
        dtype=np.uint8
    )

    total_remove = 0
    components_removed = 0

    component_debug = []

    # ========================================================
    # COMPONENT LOOP
    # ========================================================

    for label in range(
        1,
        n_labels
    ):

        component = (
            labels == label
        )

        area = int(
            stats[
                label,
                cv2.CC_STAT_AREA
            ]
        )

        if area < SHADOW_MIN_AREA:
            continue

        # ----------------------------------------------------
        # How much of component was REALLY shadow originally?
        # ----------------------------------------------------

        original_shadow_pixels_component = int(
            np.logical_and(
                component,
                shadow_original
            ).sum()
        )

        original_shadow_ratio = (
            original_shadow_pixels_component
            / area
        )

        # ----------------------------------------------------
        # Boundary contact
        # ----------------------------------------------------

        boundary_pixels = int(
            np.logical_and(
                component,
                boundary
            ).sum()
        )

        boundary_ratio = (
            boundary_pixels /
            area
        )

        # ----------------------------------------------------
        # Distance/depth
        # ----------------------------------------------------

        component_u8 = (
            component.astype(
                np.uint8
            )
        )

        distance_map = cv2.distanceTransform(
            component_u8,
            cv2.DIST_L2,
            5
        )

        max_depth = float(
            distance_map.max()
        )

        # ----------------------------------------------------
        # Candidate decision
        # ----------------------------------------------------

        evidence_ok = (
            original_shadow_ratio
            >= SHADOW_MIN_ORIGINAL_RATIO
        )

        strong_shadow_ok = (
            original_shadow_ratio
            >= SHADOW_MIN_RATIO
        )

        boundary_ok = (
            boundary_ratio
            >= SHADOW_BOUNDARY_TOUCH_RATIO
        )

        deep_shadow_ok = (
            max_depth
            >= SHADOW_MIN_DEPTH
            and
            strong_shadow_ok
        )

        # ----------------------------------------------------
        # Main gate
        #
        # Either:
        #   1. touches expanded boundary
        # OR
        #   2. is deep enough inside current leaf
        #
        # But original shadow evidence is always required.
        # ----------------------------------------------------

        keep_as_shadow = (
            evidence_ok
            and
            (
                boundary_ok
                or
                deep_shadow_ok
            )
        )

        if keep_as_shadow:

            # ------------------------------------------------
            # Global safety limit
            # ------------------------------------------------

            if (
                total_remove + area
                <=
                original_leaf_area *
                MAX_SHADOW_REMOVAL_RATIO
            ):

                shadow_remove[
                    component
                ] = 255

                total_remove += area

                components_removed += 1

        if debug:

            component_debug.append({
                "label": label,
                "area": area,
                "original_shadow_pixels":
                    original_shadow_pixels_component,
                "original_shadow_ratio":
                    original_shadow_ratio,
                "boundary_ratio":
                    boundary_ratio,
                "max_depth":
                    max_depth,
                "evidence_ok":
                    evidence_ok,
                "boundary_ok":
                    boundary_ok,
                "deep_shadow_ok":
                    deep_shadow_ok,
                "removed":
                    bool(keep_as_shadow)
            })

    # ========================================================
    # APPLY REMOVAL
    # ========================================================

    refined[
        shadow_remove > 0
    ] = 0

    # Preserve lesion for local consistency
    refined[
        lesion
    ] = 255

    debug_info = {

        "original_shadow_pixels":
            original_shadow_pixels,

        "closed_shadow_pixels":
            int(shadow_closed.sum()),

        "shadow_ratio_before":
            (
                original_shadow_pixels /
                original_leaf_area
            ),

        "removed_pixels":
            total_remove,

        "removed_ratio":
            (
                total_remove /
                original_leaf_area
            ),

        "components_total":
            n_labels - 1,

        "components_removed":
            components_removed,

        "component_debug":
            component_debug
    }

    return refined, debug_info


# ============================================================
# STEP 12 — GRABCUT SEGMENTATION
# ============================================================

def segment_leaf(rgb, lesion_mask=None):

    h, w = rgb.shape[:2]

    bgr = cv2.cvtColor(
        rgb,
        cv2.COLOR_RGB2BGR
    )

    if lesion_mask is None:

        lesion_mask = np.zeros(
            (h, w),
            dtype=np.uint8
        )

    lesion_exists = has_lesion(
        lesion_mask
    )

    # --------------------------------------------------------
    # Generic candidate
    # --------------------------------------------------------

    generic = generic_leaf_candidate(
        rgb
    )

    generic_bool = (
        generic > 0
    )

    # --------------------------------------------------------
    # Detect potential shadow BEFORE GrabCut
    # --------------------------------------------------------

    initial_shadow = detect_shadow(
        rgb,
        generic,
        lesion_mask
    )

    shadow_bool = (
        initial_shadow > 0
    )

    # --------------------------------------------------------
    # GrabCut mask
    # --------------------------------------------------------

    gc_mask = np.full(
        (h, w),
        cv2.GC_PR_BGD,
        dtype=np.uint8
    )

    b = BORDER_SIZE

    # Definite background = image border
    gc_mask[:b, :] = cv2.GC_BGD
    gc_mask[-b:, :] = cv2.GC_BGD
    gc_mask[:, :b] = cv2.GC_BGD
    gc_mask[:, -b:] = cv2.GC_BGD

    # --------------------------------------------------------
    # Generic candidate -> probable foreground
    # --------------------------------------------------------

    generic_bool[:b, :] = False
    generic_bool[-b:, :] = False
    generic_bool[:, :b] = False
    generic_bool[:, -b:] = False

    gc_mask[
        generic_bool
    ] = cv2.GC_PR_FGD

    # --------------------------------------------------------
    # Shadow -> probable background
    #
    # IMPORTANT:
    # We don't immediately call all shadow pixels definite BG.
    # They are probable background.
    # --------------------------------------------------------

    shadow_bool[:b, :] = False
    shadow_bool[-b:, :] = False
    shadow_bool[:, :b] = False
    shadow_bool[:, -b:] = False

    gc_mask[
        shadow_bool
    ] = cv2.GC_PR_BGD

    # --------------------------------------------------------
    # Lesion -> definite foreground
    # --------------------------------------------------------

    if lesion_exists:

        lesion_bool = (
            lesion_mask > 0
        )

        lesion_bool[:b, :] = False
        lesion_bool[-b:, :] = False
        lesion_bool[:, :b] = False
        lesion_bool[:, -b:] = False

        gc_mask[
            lesion_bool
        ] = cv2.GC_FGD

        # ----------------------------------------------------
        # Lesion neighbourhood = probable foreground
        # ----------------------------------------------------

        lesion_dilate = cv2.dilate(
            lesion_mask,
            cv2.getStructuringElement(
                cv2.MORPH_ELLIPSE,
                (21, 21)
            ),
            iterations=1
        ) > 0

        lesion_dilate[:b, :] = False
        lesion_dilate[-b:, :] = False
        lesion_dilate[:, :b] = False
        lesion_dilate[:, -b:] = False

        gc_mask[
            lesion_dilate &
            (~lesion_bool)
        ] = cv2.GC_PR_FGD

    # --------------------------------------------------------
    # Restore definite border background
    # --------------------------------------------------------

    gc_mask[:b, :] = cv2.GC_BGD
    gc_mask[-b:, :] = cv2.GC_BGD
    gc_mask[:, :b] = cv2.GC_BGD
    gc_mask[:, -b:] = cv2.GC_BGD

    gc_mask[:, :b] = cv2.GC_BGD
    gc_mask[:, -b:] = cv2.GC_BGD

    # --------------------------------------------------------
    # Check foreground quantity
    # --------------------------------------------------------

    fg_pixels = np.sum(
        (gc_mask == cv2.GC_FGD) |
        (gc_mask == cv2.GC_PR_FGD)
    )

    # --------------------------------------------------------
    # Fallback if too little foreground evidence
    # --------------------------------------------------------

    if fg_pixels < (
        h * w * 0.005
    ):

        lab = cv2.cvtColor(
            rgb,
            cv2.COLOR_RGB2LAB
        )

        L = lab[:, :, 0]

        center = np.zeros(
            (h, w),
            dtype=bool
        )

        x1 = int(w * 0.10)
        x2 = int(w * 0.90)

        y1 = int(h * 0.10)
        y2 = int(h * 0.90)

        center[
            y1:y2,
            x1:x2
        ] = True

        threshold = np.percentile(
            L[center],
            45
        )

        fallback = (
            L <= threshold
        ) & center

        fallback[:b, :] = False
        fallback[-b:, :] = False
        fallback[:, :b] = False
        fallback[:, -b:] = False

        gc_mask[
            fallback
        ] = cv2.GC_PR_FGD

    # --------------------------------------------------------
    # GrabCut
    # --------------------------------------------------------

    bg_model = np.zeros(
        (1, 65),
        np.float64
    )

    fg_model = np.zeros(
        (1, 65),
        np.float64
    )

    cv2.grabCut(
        bgr,
        gc_mask,
        None,
        bg_model,
        fg_model,
        GRABCUT_ITERATIONS,
        cv2.GC_INIT_WITH_MASK
    )

    # --------------------------------------------------------
    # Binary foreground
    # --------------------------------------------------------

    leaf_candidate = (
        np.isin(
            gc_mask,
            [
                cv2.GC_FGD,
                cv2.GC_PR_FGD
            ]
        ).astype(np.uint8)
        * 255
    )

    # --------------------------------------------------------
    # Basic cleanup
    # --------------------------------------------------------

    close_kernel = cv2.getStructuringElement(
        cv2.MORPH_ELLIPSE,
        (5, 5)
    )

    open_kernel = cv2.getStructuringElement(
        cv2.MORPH_ELLIPSE,
        (3, 3)
    )

    leaf_candidate = cv2.morphologyEx(
        leaf_candidate,
        cv2.MORPH_CLOSE,
        close_kernel,
        iterations=1
    )

    leaf_candidate = cv2.morphologyEx(
        leaf_candidate,
        cv2.MORPH_OPEN,
        open_kernel,
        iterations=1
    )

    # --------------------------------------------------------
    # Select main leaf component
    # --------------------------------------------------------

    leaf = select_leaf_component(
        leaf_candidate,
        lesion_mask,
        generic
    )

    # IMPORTANT:
    # We do NOT force lesion into the leaf here.
    # Lesion is used for guidance/QC.
    # This avoids artificially expanding leaf mask.
    #
    # If required for robustness, lesion-connected pixels
    # can later be considered with a controlled rule.

    return (
        leaf,
        generic,
        initial_shadow,
        gc_mask
    )


# ============================================================
# STEP 13 — FINAL CONTOUR CLEANUP
# ============================================================

def refine_leaf_contour(
    leaf_mask,
    lesion_mask=None
):

    refined = leaf_mask.copy()

    h, w = refined.shape

    # --------------------------------------------------------
    # Morphological cleanup
    # --------------------------------------------------------

    close_kernel = cv2.getStructuringElement(
        cv2.MORPH_ELLIPSE,
        (
            CONTOUR_CLOSE_KERNEL,
            CONTOUR_CLOSE_KERNEL
        )
    )

    open_kernel = cv2.getStructuringElement(
        cv2.MORPH_ELLIPSE,
        (
            CONTOUR_OPEN_KERNEL,
            CONTOUR_OPEN_KERNEL
        )
    )

    refined = cv2.morphologyEx(
        refined,
        cv2.MORPH_CLOSE,
        close_kernel,
        iterations=1
    )

    refined = cv2.morphologyEx(
        refined,
        cv2.MORPH_OPEN,
        open_kernel,
        iterations=1
    )

    # --------------------------------------------------------
    # Keep strongest component
    # --------------------------------------------------------

    binary = (
        refined > 0
    ).astype(np.uint8)

    n_labels, labels, stats, _ = (
        cv2.connectedComponentsWithStats(
            binary,
            connectivity=8
        )
    )

    if n_labels <= 1:
        return refined

    total_pixels = h * w

    best_label = None
    best_score = -1e9

    lesion_bool = (
        lesion_mask > 0
        if lesion_mask is not None
        else np.zeros(
            (h, w),
            dtype=bool
        )
    )

    lesion_total = lesion_bool.sum()

    for label in range(
        1,
        n_labels
    ):

        area = int(
            stats[
                label,
                cv2.CC_STAT_AREA
            ]
        )

        if (
            area <
            total_pixels *
            MIN_COMPONENT_KEEP_RATIO
        ):
            continue

        comp = (
            labels == label
        )

        # center
        c_score = center_score(
            comp.astype(
                np.uint8
            )
        )

        # lesion support
        if lesion_total > 0:

            lesion_inside = (
                np.logical_and(
                    comp,
                    lesion_bool
                ).sum()
            )

            lesion_support = (
                lesion_inside /
                lesion_total
            )

        else:

            lesion_support = 0.0

        # Slight preference for larger object
        area_score = (
            area /
            float(total_pixels)
        )

        score = (
            0.30 * c_score
            +
            0.45 * lesion_support
            +
            0.25 * min(
                area_score * 5.0,
                1.0
            )
        )

        if score > best_score:

            best_score = score
            best_label = label

    if best_label is not None:

        refined = (
            labels == best_label
        ).astype(np.uint8) * 255

    return refined


# ============================================================
# STEP 14 — OVERLAY
# ============================================================

def make_overlay(
    rgb,
    leaf_mask,
    lesion_mask=None
):

    overlay = rgb.copy()

    # Leaf boundary
    boundary = cv2.morphologyEx(
        leaf_mask,
        cv2.MORPH_GRADIENT,
        np.ones(
            (3, 3),
            np.uint8
        )
    )

    overlay[
        boundary > 0
    ] = [0, 255, 0]

    # Lesion
    if lesion_mask is not None:

        lesion = (
            lesion_mask > 0
        )

        lesion_overlay = (
            overlay.copy()
        )

        lesion_overlay[
            lesion
        ] = [255, 0, 0]

        overlay = cv2.addWeighted(
            overlay,
            0.70,
            lesion_overlay,
            0.30,
            0
        )

        # redraw leaf boundary
        overlay[
            boundary > 0
        ] = [0, 255, 0]

    return overlay


# ============================================================
# STEP 15 — QC + SEVERITY
# ============================================================

def calculate_qc(
    rgb,
    lesion_mask,
    leaf_mask
):

    h, w = rgb.shape[:2]

    total = h * w

    lesion = (
        lesion_mask > 0
    )

    leaf = (
        leaf_mask > 0
    )

    lesion_pixels = int(
        lesion.sum()
    )

    leaf_pixels = int(
        leaf.sum()
    )

    # --------------------------------------------------------
    # Lesion outside leaf
    # --------------------------------------------------------

    lesion_outside = int(
        np.logical_and(
            lesion,
            ~leaf
        ).sum()
    )

    # --------------------------------------------------------
    # Lesion inside leaf
    # --------------------------------------------------------

    lesion_inside = int(
        np.logical_and(
            lesion,
            leaf
        ).sum()
    )

    # --------------------------------------------------------
    # Ratios
    # --------------------------------------------------------

    leaf_ratio = (
        leaf_pixels / total
        if total > 0
        else np.nan
    )

    lesion_ratio = (
        lesion_pixels / total
        if total > 0
        else np.nan
    )

    # --------------------------------------------------------
    # Severity
    #
    # IMPORTANT:
    # Only lesion pixels inside leaf count.
    # --------------------------------------------------------

    severity = (
        100.0 *
        lesion_inside /
        leaf_pixels
        if leaf_pixels > 0
        else np.nan
    )

    # --------------------------------------------------------
    # Border
    # --------------------------------------------------------

    b = BORDER_SIZE

    border = np.zeros(
        (h, w),
        dtype=bool
    )

    border[:b, :] = True
    border[-b:, :] = True
    border[:, :b] = True
    border[:, -b:] = True

    border_touch = bool(
        np.logical_and(
            leaf,
            border
        ).sum() > 0
    )

    # --------------------------------------------------------
    # QC flags
    # --------------------------------------------------------

    flags = []

    if leaf_pixels == 0:
        flags.append(
            "empty_leaf_mask"
        )

    if (
        leaf_ratio <
        MIN_LEAF_AREA_RATIO
    ):
        flags.append(
            "leaf_too_small"
        )

    if (
        leaf_ratio >
        MAX_LEAF_AREA_RATIO
    ):
        flags.append(
            "leaf_too_large"
        )

    if lesion_outside > 0:
        flags.append(
            "lesion_outside_leaf"
        )

    if border_touch:
        flags.append(
            "leaf_touches_border"
        )

    status = (
        "OK"
        if len(flags) == 0
        else "CHECK"
    )

    return {

        "leaf_pixels":
            leaf_pixels,

        "lesion_pixels":
            lesion_pixels,

        "lesion_inside_leaf":
            lesion_inside,

        "leaf_area_ratio":
            leaf_ratio,

        "lesion_area_ratio":
            lesion_ratio,

        "lesion_outside_leaf":
            lesion_outside,

        "severity_percent":
            severity,

        "leaf_touches_border":
            border_touch,

        "qc_status":
            status,

        "qc_flags":
            ";".join(flags)
    }


# ============================================================
# STEP 16 — BUILD IMAGE/MASK RECORDS
# ============================================================

def build_records(
    image_dir,
    mask_dir
):

    records = []

    for image_path in sorted(
        image_dir.rglob("*")
    ):

        if not image_path.is_file():
            continue

        if (
            image_path.suffix.lower()
            not in IMAGE_EXTS
        ):
            continue

        possible_masks = [

            mask_dir /
            f"{image_path.stem}.png",

            mask_dir /
            f"{image_path.stem}.jpg",

            mask_dir /
            f"{image_path.stem}.jpeg",

            mask_dir /
            f"{image_path.stem}.bmp",
        ]

        mask_path = None

        for p in possible_masks:

            if p.exists():

                mask_path = p
                break

        if mask_path is None:
            continue

        records.append({

            "image_path":
                str(image_path),

            "mask_path":
                str(mask_path),

            "filename":
                image_path.name,

            "class_name":
                infer_class(
                    image_path
                )
        })

    return pd.DataFrame(
        records
    )


validation_df = build_records(
    VALID_IMAGES_DIR,
    VALID_MASKS_DIR
)

if validation_df.empty:

    raise ValueError(
        "No image-mask pairs found."
    )

print(
    "=============================================="
)

print(
    "POTATO LEAF SEGMENTATION V5"
)

print(
    "Shadow-Aware Lesion-Guided GrabCut"
)

print(
    "=============================================="
)

print(
    f"\nFound image-mask pairs: "
    f"{len(validation_df)}"
)

print(
    "\nClass distribution:"
)

print(
    validation_df[
        "class_name"
    ].value_counts(
        dropna=False
    )
)


# ============================================================
# STEP 17 — FULL DATASET / SPLIT CONFIGURATION
# ============================================================

PILOT_N = None

SPLITS = ["train", "valid", "test"]
all_results = []

# Keep every split visually/data-wise separate so identical filenames
# across train/valid/test cannot overwrite each other.
SPLIT_DIRS = {}
for split in SPLITS:
    split_root = OUTPUT_DIR / split
    split_dirs = {
        "root": split_root,
        "mask": split_root / "leaf_masks",
        "overlay": split_root / "overlays",
        "combined": split_root / "combined",
        "debug": split_root / "debug",
        "before": split_root / "debug" / "before_shadow",
        "shadow": split_root / "debug" / "shadow",
        "final": split_root / "debug" / "final",
    }
    for d in split_dirs.values():
        if isinstance(d, Path):
            d.mkdir(parents=True, exist_ok=True)
    SPLIT_DIRS[split] = split_dirs


# ============================================================
# STEP 18 — PROCESS ONE SPLIT
# ============================================================

def process_split(split, images_dir, masks_dir):

    df = build_records(images_dir, masks_dir)
    print("\n" + "=" * 70)
    print(f"SPLIT: {split}")
    print(f"Image-mask pairs: {len(df)}")
    print("=" * 70)

    if df.empty:
        return []

    if PILOT_N is not None:
        selected = []

        for class_name, group in df.groupby("class_name", sort=False):
            proportion = len(group) / len(df)
            n_take = max(1, round(PILOT_N * proportion))
            n_take = min(n_take, len(group))
            selected.append(
                group.sample(n=n_take, random_state=42)
            )

        df = pd.concat(selected, ignore_index=True)

        if len(df) > PILOT_N:
            df = (
                df.sample(n=PILOT_N, random_state=42)
                .reset_index(drop=True)
            )

    print(f"Processing: {len(df)} images")

    split_dirs = SPLIT_DIRS[split]
    results = []

    for idx, row in df.iterrows():

        image_path = Path(row["image_path"])
        mask_path = Path(row["mask_path"])

        print("\n" + "-" * 70)
        print(f"[{idx + 1}/{len(df)}] {image_path.name}")

        try:

            # ----------------------------------------------------
            # Read
            # ----------------------------------------------------
            rgb = read_rgb(image_path)

            lesion_mask = read_binary_mask(
                mask_path,
                rgb.shape[:2]
            )

            # ----------------------------------------------------
            # GrabCut
            # ----------------------------------------------------
            (
                leaf_before_shadow,
                generic_candidate,
                initial_shadow,
                grabcut_debug
            ) = segment_leaf(
                rgb,
                lesion_mask
            )

            # ----------------------------------------------------
            # Shadow refinement
            # ----------------------------------------------------
            (
                leaf_after_shadow,
                shadow_debug
            ) = remove_conservative_shadow(
                rgb,
                leaf_before_shadow,
                lesion_mask,
                debug=True
            )

            # ----------------------------------------------------
            # Contour refinement
            # ----------------------------------------------------
            leaf_final = refine_leaf_contour(
                leaf_after_shadow,
                lesion_mask
            )

            # ----------------------------------------------------
            # QC + severity
            # ----------------------------------------------------
            qc = calculate_qc(
                rgb,
                lesion_mask,
                leaf_final
            )

            # ----------------------------------------------------
            # Overlay
            # ----------------------------------------------------
            overlay = make_overlay(
                rgb,
                leaf_final,
                lesion_mask
            )

            # ----------------------------------------------------
            # Save final leaf mask
            # ----------------------------------------------------
            leaf_mask_path = (
                split_dirs["mask"] /
                f"{image_path.stem}_leaf.png"
            )
            cv2.imwrite(
                str(leaf_mask_path),
                leaf_final
            )

            # ----------------------------------------------------
            # Save overlay
            # ----------------------------------------------------
            overlay_path = (
                split_dirs["overlay"] /
                f"{image_path.stem}_overlay.jpg"
            )
            cv2.imwrite(
                str(overlay_path),
                cv2.cvtColor(
                    overlay,
                    cv2.COLOR_RGB2BGR
                )
            )

            # ----------------------------------------------------
            # Save before-shadow mask
            # ----------------------------------------------------
            before_path = (
                split_dirs["before"] /
                f"{image_path.stem}_before.png"
            )
            cv2.imwrite(
                str(before_path),
                leaf_before_shadow
            )

            # ----------------------------------------------------
            # Save detected shadow mask
            # ----------------------------------------------------
            shadow_path = (
                split_dirs["shadow"] /
                f"{image_path.stem}_shadow.png"
            )
            cv2.imwrite(
                str(shadow_path),
                initial_shadow
            )

            # ----------------------------------------------------
            # Save final debug mask
            # ----------------------------------------------------
            final_debug_path = (
                split_dirs["final"] /
                f"{image_path.stem}_final.png"
            )
            cv2.imwrite(
                str(final_debug_path),
                leaf_final
            )

            # ----------------------------------------------------
            # Save combined visual panel
            # ----------------------------------------------------
            fig, axes = plt.subplots(
                1,
                5,
                figsize=(22, 5)
            )

            axes[0].imshow(rgb)
            axes[0].set_title("Original")

            axes[1].imshow(
                leaf_before_shadow,
                cmap="gray"
            )
            axes[1].set_title("Before Shadow")

            axes[2].imshow(
                initial_shadow,
                cmap="gray"
            )
            axes[2].set_title("Detected Shadow")

            axes[3].imshow(
                leaf_final,
                cmap="gray"
            )
            axes[3].set_title("Final Leaf")

            axes[4].imshow(overlay)
            axes[4].set_title(
                f"Severity={qc['severity_percent']:.2f}%"
            )

            for ax in axes:
                ax.axis("off")

            plt.tight_layout()

            combined_path = (
                split_dirs["combined"] /
                f"{image_path.stem}_combined.jpg"
            )

            plt.savefig(
                combined_path,
                dpi=140,
                bbox_inches="tight"
            )
            plt.close(fig)

            # ----------------------------------------------------
            # Record
            # ----------------------------------------------------
            record = {
                "split": split,
                "filename": image_path.name,
                "class_name": row["class_name"],
                "image_path": str(image_path),
                "lesion_mask_path": str(mask_path),
                "leaf_mask_path": str(leaf_mask_path),
                "overlay_path": str(overlay_path),
                "combined_path": str(combined_path),
                "severity_percent": qc["severity_percent"],
                "leaf_pixels": qc["leaf_pixels"],
                "lesion_pixels": qc["lesion_pixels"],
                "lesion_inside_leaf": qc["lesion_inside_leaf"],
                "lesion_outside_leaf": qc["lesion_outside_leaf"],
                "leaf_area_ratio": qc["leaf_area_ratio"],
                "shadow_pixels_detected": (
                    shadow_debug["original_shadow_pixels"]
                    if shadow_debug is not None else 0
                ),
                "shadow_pixels_after_close": (
                    shadow_debug["closed_shadow_pixels"]
                    if shadow_debug is not None else 0
                ),
                "shadow_pixels_removed": (
                    shadow_debug["removed_pixels"]
                    if shadow_debug is not None else 0
                ),
                "shadow_removed_ratio": (
                    shadow_debug["removed_ratio"]
                    if shadow_debug is not None else 0.0
                ),
                "shadow_components_total": (
                    shadow_debug["components_total"]
                    if shadow_debug is not None else 0
                ),
                "shadow_components_removed": (
                    shadow_debug["components_removed"]
                    if shadow_debug is not None else 0
                ),
                "qc_status": qc["qc_status"],
                "qc_flags": qc["qc_flags"]
            }

            results.append(record)

            print(
                f"Leaf area: {qc['leaf_area_ratio'] * 100:.2f}%"
            )
            print(
                f"Shadow detected: "
                f"{record['shadow_pixels_detected']} px"
            )
            print(
                f"Shadow removed: "
                f"{record['shadow_pixels_removed']} px"
            )
            print(
                f"Severity: {qc['severity_percent']:.2f}%"
            )
            print(f"QC: {qc['qc_status']}")

            if qc["qc_flags"]:
                print("Flags:", qc["qc_flags"])

        except Exception as e:

            print("ERROR:", repr(e))

            results.append({
                "split": split,
                "filename": image_path.name,
                "class_name": row["class_name"],
                "image_path": str(image_path),
                "lesion_mask_path": str(mask_path),
                "qc_status": "ERROR",
                "qc_flags": str(e)
            })

    return results


# ============================================================
# STEP 19 — RUN ALL SPLITS
# ============================================================

for split in SPLITS:

    images_dir = ROOT / split / "images"
    masks_dir = ROOT / split / "lesion_masks"

    if not images_dir.exists():
        print(f"Skipping {split} (folder not found): {images_dir}")
        continue

    split_results = process_split(
        split,
        images_dir,
        masks_dir
    )

    all_results.extend(split_results)


# ============================================================
# STEP 20 — RESULTS DATAFRAME + MASTER CSV
# ============================================================

results_df = pd.DataFrame(all_results)

CSV_PATH = (
    OUTPUT_DIR /
    "leaf_segmentation_v5_ALL_SPLITS.csv"
)

results_df.to_csv(
    CSV_PATH,
    index=False
)

print("\n" + "=" * 70)
print("V5 FULL DATASET PROCESSING COMPLETED")
print("=" * 70)
print("Results saved to:")
print(CSV_PATH)
print("Result shape:", results_df.shape)


# ============================================================
# STEP 21 — SUMMARY STATISTICS
# ============================================================

if not results_df.empty:

    numeric_cols = [
        "severity_percent",
        "leaf_area_ratio",
        "shadow_pixels_detected",
        "shadow_pixels_removed",
        "shadow_removed_ratio",
    ]

    available_numeric = [
        c for c in numeric_cols
        if c in results_df.columns
    ]

    print("\nOverall summary statistics:")
    display(
        results_df[available_numeric].describe()
    )

    print("\nSplit-wise summary:")
    split_summary = (
        results_df
        .groupby("split")
        .agg(
            images=("filename", "count"),
            qc_check=("qc_status", lambda s: int((s == "CHECK").sum())),
            qc_error=("qc_status", lambda s: int((s == "ERROR").sum())),
            mean_severity=("severity_percent", "mean"),
            mean_leaf_area=("leaf_area_ratio", "mean"),
            mean_shadow_removed_ratio=("shadow_removed_ratio", "mean"),
        )
        .reset_index()
    )

    display(split_summary)

    print("\nQC distribution by split:")
    print(
        results_df
        .groupby("split")["qc_status"]
        .value_counts()
    )

    print("\nCHECK-flagged images:")
    check_df = results_df[
        results_df.qc_status == "CHECK"
    ][["filename", "split", "qc_flags"]]

    display(check_df)

    print(
        f"\nCHECK count: {len(check_df)} / {len(results_df)} "
        f"({100.0 * len(check_df) / len(results_df):.2f}%)"
    )


# ============================================================
# STEP 22 — DISPLAY SAMPLE RESULTS
# ============================================================

display_cols = [
    "split",
    "filename",
    "class_name",
    "severity_percent",
    "leaf_area_ratio",
    "shadow_pixels_detected",
    "shadow_pixels_removed",
    "shadow_removed_ratio",
    "lesion_outside_leaf",
    "qc_status",
    "qc_flags"
]

display_cols = [
    c for c in display_cols
    if c in results_df.columns
]

if not results_df.empty:
    print("\nFirst results:")
    display(
        results_df[display_cols].head(20)
    )


# ============================================================
# STEP 23 — RANDOM VISUAL CHECK
# ============================================================

def show_random_results(n=6, seed=42):

    if results_df.empty:
        print("No results.")
        return

    n = min(n, len(results_df))

    sample_df = results_df.sample(
        n=n,
        random_state=seed
    )

    cols = 3
    rows = math.ceil(n / cols)

    fig, axes = plt.subplots(
        rows,
        cols,
        figsize=(18, 6 * rows)
    )

    axes = np.array(axes).reshape(-1)

    for ax, (_, row) in zip(
        axes,
        sample_df.iterrows()
    ):

        combined_path = Path(
            row.get("combined_path", "")
        )

        if combined_path.exists():
            img = cv2.imread(str(combined_path))
            img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
            ax.imshow(img)

        ax.set_title(
            f"{row['split']} | {row['filename']}"
        )
        ax.axis("off")

    for ax in axes[len(sample_df):]:
        ax.axis("off")

    plt.tight_layout()
    plt.show()


show_random_results(n=6, seed=42)


# ============================================================
# STEP 24 — SHADOW DEBUG REPORT
# ============================================================

if not results_df.empty:

    shadow_cols = [
        "split",
        "filename",
        "shadow_pixels_detected",
        "shadow_pixels_after_close",
        "shadow_pixels_removed",
        "shadow_removed_ratio",
        "shadow_components_total",
        "shadow_components_removed"
    ]

    shadow_cols = [
        c for c in shadow_cols
        if c in results_df.columns
    ]

    print("\nShadow refinement report:")
    display(
        results_df[shadow_cols]
        .sort_values(
            "shadow_removed_ratio",
            ascending=False
        )
        .head(20)
    )


# ============================================================
# STEP 25 — EXPORT PATH SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("OUTPUT DIRECTORIES")
print("=" * 70)
print("Master CSV:", CSV_PATH)

for split in SPLITS:
    if split not in SPLIT_DIRS:
        continue
    print(f"\n[{split}]")
    print("Final leaf masks :", SPLIT_DIRS[split]["mask"])
    print("Overlays         :", SPLIT_DIRS[split]["overlay"])
    print("Combined panels  :", SPLIT_DIRS[split]["combined"])
    print("Before-shadow    :", SPLIT_DIRS[split]["before"])
    print("Shadow masks     :", SPLIT_DIRS[split]["shadow"])
    print("Final debug      :", SPLIT_DIRS[split]["final"])

```

```python
# ============================================================
# POTATO SEVERITY DATASET - FINAL INTEGRATED PIPELINE
#
# INPUT:
#   ROOT/
#       train/images
#       train/lesion_masks
#       valid/images
#       valid/lesion_masks
#       test/images
#       test/lesion_masks
#
# OUTPUT:
#   Potato_Severity_Dataset_FINAL/
#       train/images
#       train/lesion_masks
#       train/leaf_masks
#       valid/images
#       valid/lesion_masks
#       valid/leaf_masks
#       test/images
#       test/lesion_masks
#       test/leaf_masks
#       severity_dataset/potato_severity_FINAL.csv
#
# IMPORTANT:
#   CSV paths are dataset-root-relative.
#   No /content/, /content/drive/, C:\, etc.
# ============================================================


# ============================================================
# STEP 0 — IMPORTS
# ============================================================

from pathlib import Path
import shutil
import cv2
import numpy as np
import pandas as pd
import math
import warnings

warnings.filterwarnings("ignore")


# ============================================================
# STEP 1 — RUNTIME PATH CONFIGURATION
# ============================================================

# This is ONLY the location of the currently mounted/source
# dataset in the current environment.
#
# Change this ONE variable when moving to another environment.
#
# The CSV itself will NOT contain this path.

ROOT = Path("/content/Potato_Severity_Dataset")

FINAL_ROOT = Path("/content/Potato_Severity_Dataset_FINAL")

CSV_PATH = (
    FINAL_ROOT
    / "severity_dataset"
    / "potato_severity_FINAL.csv"
)

SPLITS = ["train", "valid", "test"]

TARGET_CLASSES = {
    "Potato Healthy",
    "Potato Late Blight"
}

IMAGE_EXTS = {
    ".jpg",
    ".jpeg",
    ".png",
    ".bmp",
    ".webp"
}

MASK_EXTS = {
    ".png",
    ".jpg",
    ".jpeg",
    ".bmp"
}


# ============================================================
# STEP 2 — CHECK SOURCE STRUCTURE
# ============================================================

if not ROOT.exists():
    raise FileNotFoundError(
        f"Source dataset root not found:\n{ROOT}"
    )

print("=" * 80)
print("SOURCE DATASET")
print("=" * 80)
print(ROOT)

for split in SPLITS:

    image_dir = ROOT / split / "images"
    mask_dir = ROOT / split / "lesion_masks"

    if not image_dir.exists():
        raise FileNotFoundError(
            f"Missing image directory:\n{image_dir}"
        )

    if not mask_dir.exists():
        raise FileNotFoundError(
            f"Missing lesion-mask directory:\n{mask_dir}"
        )

    print(
        f"{split:>5} | "
        f"images={len(list(image_dir.iterdir())):4d} | "
        f"lesion_masks={len(list(mask_dir.iterdir())):4d}"
    )


# ============================================================
# STEP 3 — CREATE FINAL DATASET DIRECTORIES
# ============================================================

for split in SPLITS:

    for subdir in [
        "images",
        "lesion_masks",
        "leaf_masks",
    ]:

        (
            FINAL_ROOT
            / split
            / subdir
        ).mkdir(
            parents=True,
            exist_ok=True
        )

(
    FINAL_ROOT
    / "severity_dataset"
).mkdir(
    parents=True,
    exist_ok=True
)


# ============================================================
# STEP 4 — CLASS DETECTION
# ============================================================

def infer_class(path):

    name = path.name.lower()

    # IMPORTANT:
    # Early and Late Blight are checked separately.
    #
    # "early_blight" must NEVER become Late Blight.

    if (
        "early_blight" in name
        or "early-blight" in name
        or "earlyblight" in name
    ):
        return "Potato Early Blight"

    if (
        "late_blight" in name
        or "late-blight" in name
        or "lateblight" in name
    ):
        return "Potato Late Blight"

    if "healthy" in name:
        return "Potato Healthy"

    return "Unknown"


# ============================================================
# STEP 5 — EXACT IMAGE ↔ MASK PAIRING
# ============================================================

def build_records(
    image_dir,
    mask_dir,
    split
):

    records = []

    images = sorted(
        p
        for p in image_dir.iterdir()
        if (
            p.is_file()
            and p.suffix.lower() in IMAGE_EXTS
        )
    )

    # --------------------------------------------------------
    # Build mask index by STEM
    #
    # Example:
    #
    # xxx.jpg
    # xxx.png
    #
    # Both have stem "xxx"
    # --------------------------------------------------------

    mask_index = {}

    for mask_path in sorted(mask_dir.iterdir()):

        if (
            not mask_path.is_file()
            or mask_path.suffix.lower() not in MASK_EXTS
        ):
            continue

        stem = mask_path.stem

        if stem in mask_index:

            raise RuntimeError(
                "Duplicate lesion-mask stem detected:\n"
                f"{stem}\n"
                f"{mask_index[stem]}\n"
                f"{mask_path}"
            )

        mask_index[stem] = mask_path

    # --------------------------------------------------------
    # Pair image and mask by exact STEM
    # --------------------------------------------------------

    for image_path in images:

        mask_path = mask_index.get(
            image_path.stem
        )

        if mask_path is None:

            records.append({

                "split": split,

                "image_path":
                    image_path,

                "mask_path":
                    None,

                "filename":
                    image_path.name,

                "class_name":
                    infer_class(image_path),

                "pair_status":
                    "MISSING_LESION_MASK"
            })

            continue

        records.append({

            "split": split,

            "image_path":
                image_path,

            "mask_path":
                mask_path,

            "filename":
                image_path.name,

            "class_name":
                infer_class(image_path),

            "pair_status":
                "PAIRED"
        })

    return records


# ============================================================
# STEP 6 — READ BINARY MASK
# ============================================================

def read_binary_mask(
    path,
    shape
):

    if path is None:
        raise FileNotFoundError(
            "Lesion mask path is None."
        )

    mask = cv2.imread(
        str(path),
        cv2.IMREAD_GRAYSCALE
    )

    if mask is None:
        raise FileNotFoundError(
            f"Could not read lesion mask:\n{path}"
        )

    if mask.shape != shape:

        mask = cv2.resize(
            mask,
            (
                shape[1],
                shape[0]
            ),
            interpolation=cv2.INTER_NEAREST
        )

    return (
        (mask > 127)
        .astype(np.uint8)
        * 255
    )


# ============================================================
# STEP 7 — READ RGB
# ============================================================

def read_rgb(path):

    image = cv2.imread(
        str(path),
        cv2.IMREAD_COLOR
    )

    if image is None:

        raise FileNotFoundError(
            f"Could not read image:\n{path}"
        )

    return cv2.cvtColor(
        image,
        cv2.COLOR_BGR2RGB
    )


# ============================================================
# STEP 8 — BUILD ALL SOURCE RECORDS
# ============================================================

source_records = []

for split in SPLITS:

    records = build_records(
        ROOT / split / "images",
        ROOT / split / "lesion_masks",
        split
    )

    source_records.extend(
        records
    )


source_df = pd.DataFrame(
    source_records
)


print("\n" + "=" * 80)
print("SOURCE PAIRING SUMMARY")
print("=" * 80)

print(
    source_df[
        ["split", "pair_status"]
    ]
    .value_counts()
    .sort_index()
)


# ============================================================
# STEP 9 — REMOVE NON-TARGET CLASSES
# ============================================================

non_target_df = source_df[
    ~source_df["class_name"].isin(
        TARGET_CLASSES
    )
].copy()

target_df = source_df[
    source_df["class_name"].isin(
        TARGET_CLASSES
    )
].copy()


print("\n" + "=" * 80)
print("CLASS FILTER")
print("=" * 80)

print(
    "All source records:",
    len(source_df)
)

print(
    "Target records:",
    len(target_df)
)

print(
    "Non-target records:",
    len(non_target_df)
)

if not non_target_df.empty:

    print("\nExcluded records:")

    print(
        non_target_df[
            [
                "split",
                "filename",
                "class_name",
                "pair_status"
            ]
        ].to_string(
            index=False
        )
    )


# ============================================================
# STEP 10 — ONLY PAIRED TARGET RECORDS ENTER SEGMENTATION
# ============================================================

processable_df = target_df[
    target_df["pair_status"] == "PAIRED"
].copy()


missing_pair_df = target_df[
    target_df["pair_status"] != "PAIRED"
].copy()


print("\n" + "=" * 80)
print("PROCESSABLE DATA")
print("=" * 80)

print(
    "Target records:",
    len(target_df)
)

print(
    "Processable:",
    len(processable_df)
)

print(
    "Missing lesion mask:",
    len(missing_pair_df)
)


# ============================================================
# STEP 11 — FINAL RECORD PROCESSING
#
# IMPORTANT:
# The following functions must already exist from your V5 code:
#
#   segment_leaf()
#   remove_conservative_shadow()
#   refine_leaf_contour()
#   calculate_qc()
#
# Your V5 segmentation logic is NOT changed.
# ============================================================

all_results = []
error_records = []

for split in SPLITS:

    split_df = processable_df[
        processable_df["split"] == split
    ].copy()

    print("\n")
    print("=" * 80)
    print(
        f"PROCESSING SPLIT: {split.upper()}"
    )
    print(
        f"Records: {len(split_df)}"
    )
    print("=" * 80)

    final_image_dir = (
        FINAL_ROOT
        / split
        / "images"
    )

    final_lesion_dir = (
        FINAL_ROOT
        / split
        / "lesion_masks"
    )

    final_leaf_dir = (
        FINAL_ROOT
        / split
        / "leaf_masks"
    )

    for idx, row in enumerate(
        split_df.itertuples(
            index=False
        ),
        start=1
    ):

        image_path = Path(
            row.image_path
        )

        mask_path = Path(
            row.mask_path
        )

        print(
            f"[{idx:4d}/{len(split_df):4d}] "
            f"{image_path.name}"
        )

        try:

            # ==================================================
            # 1. READ IMAGE
            # ==================================================

            rgb = read_rgb(
                image_path
            )

            # ==================================================
            # 2. READ LESION MASK
            # ==================================================

            lesion_mask = read_binary_mask(
                mask_path,
                rgb.shape[:2]
            )

            # ==================================================
            # 3. RUN V5
            # ==================================================

            (
                leaf_before_shadow,
                generic_candidate,
                initial_shadow,
                grabcut_debug
            ) = segment_leaf(
                rgb,
                lesion_mask
            )

            # ==================================================
            # 4. SHADOW REFINEMENT
            # ==================================================

            (
                leaf_after_shadow,
                shadow_debug
            ) = remove_conservative_shadow(
                rgb,
                leaf_before_shadow,
                lesion_mask,
                debug=True
            )

            # ==================================================
            # 5. CONTOUR REFINEMENT
            # ==================================================

            leaf_final = refine_leaf_contour(
                leaf_after_shadow,
                lesion_mask
            )

            # ==================================================
            # 6. CALCULATE SEVERITY
            # ==================================================

            qc = calculate_qc(
                rgb,
                lesion_mask,
                leaf_final
            )

            # --------------------------------------------------
            # Do NOT convert failed severity to 0.
            # --------------------------------------------------

            severity = qc[
                "severity_percent"
            ]

            # --------------------------------------------------
            # Failed leaf segmentation
            # --------------------------------------------------

            if (
                qc["leaf_pixels"] <= 0
                or pd.isna(severity)
            ):

                raise RuntimeError(
                    "Invalid leaf segmentation: "
                    "leaf_pixels <= 0 or severity is NaN."
                )

            # ==================================================
            # 7. OUTPUT FILENAMES
            # ==================================================

            leaf_filename = (
                f"{image_path.stem}_leaf.png"
            )

            final_image_path = (
                final_image_dir
                / image_path.name
            )

            final_lesion_path = (
                final_lesion_dir
                / mask_path.name
            )

            final_leaf_path = (
                final_leaf_dir
                / leaf_filename
            )

            # ==================================================
            # 8. COPY SOURCE IMAGE
            # ==================================================

            shutil.copy2(
                image_path,
                final_image_path
            )

            # ==================================================
            # 9. COPY SOURCE LESION MASK
            # ==================================================

            shutil.copy2(
                mask_path,
                final_lesion_path
            )

            # ==================================================
            # 10. SAVE FINAL LEAF MASK
            # ==================================================

            success = cv2.imwrite(
                str(final_leaf_path),
                leaf_final
            )

            if not success:
                raise RuntimeError(
                    f"Failed to save leaf mask:\n"
                    f"{final_leaf_path}"
                )

            # ==================================================
            # 11. BUILD PORTABLE RELATIVE PATHS
            #
            # IMPORTANT:
            # These are relative to FINAL_ROOT.
            # ==================================================

            image_rel = (
                final_image_path
                .relative_to(FINAL_ROOT)
                .as_posix()
            )

            lesion_rel = (
                final_lesion_path
                .relative_to(FINAL_ROOT)
                .as_posix()
            )

            leaf_rel = (
                final_leaf_path
                .relative_to(FINAL_ROOT)
                .as_posix()
            )

            # ==================================================
            # 12. RECORD
            # ==================================================

            record = {

                "split":
                    split,

                "filename":
                    image_path.name,

                "class_name":
                    row.class_name,

                "image_path":
                    image_rel,

                "lesion_mask_path":
                    lesion_rel,

                "leaf_mask_path":
                    leaf_rel,

                "leaf_pixels":
                    qc["leaf_pixels"],

                "lesion_pixels":
                    qc["lesion_pixels"],

                "lesion_inside_leaf":
                    qc["lesion_inside_leaf"],

                "lesion_outside_leaf":
                    qc["lesion_outside_leaf"],

                "severity_percent":
                    qc["severity_percent"],

                "leaf_area_ratio":
                    qc["leaf_area_ratio"],

                "qc_status":
                    qc["qc_status"],

                "qc_flags":
                    qc["qc_flags"],

                "shadow_pixels_detected":
                    (
                        shadow_debug[
                            "original_shadow_pixels"
                        ]
                        if shadow_debug is not None
                        else 0
                    ),

                "shadow_pixels_after_close":
                    (
                        shadow_debug[
                            "closed_shadow_pixels"
                        ]
                        if shadow_debug is not None
                        else 0
                    ),

                "shadow_pixels_removed":
                    (
                        shadow_debug[
                            "removed_pixels"
                        ]
                        if shadow_debug is not None
                        else 0
                    ),

                "shadow_removed_ratio":
                    (
                        shadow_debug[
                            "removed_ratio"
                        ]
                        if shadow_debug is not None
                        else 0.0
                    ),

                "shadow_components_total":
                    (
                        shadow_debug[
                            "components_total"
                        ]
                        if shadow_debug is not None
                        else 0
                    ),

                "shadow_components_removed":
                    (
                        shadow_debug[
                            "components_removed"
                        ]
                        if shadow_debug is not None
                        else 0
                    ),
            }

            all_results.append(
                record
            )

        except Exception as e:

            error_records.append({

                "split":
                    split,

                "filename":
                    image_path.name,

                "class_name":
                    row.class_name,

                "source_image":
                    str(image_path),

                "source_lesion_mask":
                    str(mask_path),

                "error":
                    repr(e)
            })

            print(
                "   ERROR:",
                repr(e)
            )


# ============================================================
# STEP 12 — CANONICAL DATAFRAME
# ============================================================

final_df = pd.DataFrame(
    all_results
)

error_df = pd.DataFrame(
    error_records
)


# ============================================================
# STEP 13 — SAVE OPTIONAL QC ERROR REPORT
# ============================================================

QC_DIR = (
    FINAL_ROOT
    / "qc"
)

QC_DIR.mkdir(
    parents=True,
    exist_ok=True
)

if not missing_pair_df.empty:

    missing_pair_df_export = (
        missing_pair_df.copy()
    )

    missing_pair_df_export[
        "image_path"
    ] = missing_pair_df_export[
        "image_path"
    ].astype(str)

    missing_pair_df_export[
        "mask_path"
    ] = missing_pair_df_export[
        "mask_path"
    ].apply(
        lambda x:
        str(x)
        if x is not None
        else ""
    )

    missing_pair_df_export.to_csv(
        QC_DIR
        / "missing_lesion_masks.csv",
        index=False
    )


if not error_df.empty:

    error_df.to_csv(
        QC_DIR
        / "processing_errors.csv",
        index=False
    )


if not non_target_df.empty:

    non_target_df[
        [
            "split",
            "filename",
            "class_name",
            "pair_status"
        ]
    ].to_csv(
        QC_DIR
        / "excluded_non_target.csv",
        index=False
    )


# ============================================================
# STEP 14 — CANONICAL COLUMN ORDER
# ============================================================

CANONICAL_COLUMNS = [

    "split",
    "filename",
    "class_name",

    "image_path",
    "lesion_mask_path",
    "leaf_mask_path",

    "leaf_pixels",
    "lesion_pixels",
    "lesion_inside_leaf",
    "lesion_outside_leaf",

    "severity_percent",
    "leaf_area_ratio",

    "qc_status",
    "qc_flags",

    "shadow_pixels_detected",
    "shadow_pixels_after_close",
    "shadow_pixels_removed",
    "shadow_removed_ratio",

    "shadow_components_total",
    "shadow_components_removed",
]


if final_df.empty:

    raise RuntimeError(
        "No valid target records were processed."
    )


final_df = final_df[
    CANONICAL_COLUMNS
]


# ============================================================
# STEP 15 — SAVE CANONICAL CSV
# ============================================================

final_df.to_csv(
    CSV_PATH,
    index=False
)

print("\n")
print("=" * 80)
print("CANONICAL CSV CREATED")
print("=" * 80)

print(
    CSV_PATH
)

print(
    "Rows:",
    len(final_df)
)


# ============================================================
# STEP 16 — AUTOMATIC INTEGRITY CHECKS
# ============================================================

print("\n")
print("=" * 80)
print("AUTOMATIC DATASET INTEGRITY CHECKS")
print("=" * 80)


# ------------------------------------------------------------
# Basic
# ------------------------------------------------------------

assert len(final_df) > 0

assert (
    final_df["split"]
    .isin(SPLITS)
    .all()
)

assert (
    final_df["class_name"]
    .isin(TARGET_CLASSES)
    .all()
)

# ------------------------------------------------------------
# Unique record identity
# ------------------------------------------------------------

duplicate_records = final_df.duplicated(
    subset=[
        "split",
        "filename"
    ]
).sum()

print(
    "Duplicate (split, filename):",
    duplicate_records
)

assert duplicate_records == 0


# ------------------------------------------------------------
# Missing values
# ------------------------------------------------------------

print(
    "Missing image_path:",
    final_df["image_path"].isna().sum()
)

print(
    "Missing lesion_mask_path:",
    final_df["lesion_mask_path"].isna().sum()
)

print(
    "Missing leaf_mask_path:",
    final_df["leaf_mask_path"].isna().sum()
)

print(
    "Missing severity:",
    final_df["severity_percent"].isna().sum()
)

assert final_df[
    "image_path"
].notna().all()

assert final_df[
    "lesion_mask_path"
].notna().all()

assert final_df[
    "leaf_mask_path"
].notna().all()

assert final_df[
    "severity_percent"
].notna().all()


# ------------------------------------------------------------
# Severity range
# ------------------------------------------------------------

severity_negative = (
    final_df["severity_percent"] < 0
).sum()

severity_over_100 = (
    final_df["severity_percent"] > 100
).sum()

print(
    "Severity < 0:",
    severity_negative
)

print(
    "Severity > 100:",
    severity_over_100
)

assert severity_negative == 0
assert severity_over_100 == 0


# ------------------------------------------------------------
# Absolute-path detection
# ------------------------------------------------------------

path_columns = [
    "image_path",
    "lesion_mask_path",
    "leaf_mask_path"
]

for col in path_columns:

    values = (
        final_df[col]
        .astype(str)
    )

    absolute_unix = (
        values.str.startswith("/")
    )

    absolute_windows = (
        values.str.match(
            r"^[A-Za-z]:[\\/]"
        )
    )

    bad = (
        absolute_unix
        |
        absolute_windows
    )

    count_bad = int(
        bad.sum()
    )

    print(
        f"Absolute paths in {col}:",
        count_bad
    )

    assert count_bad == 0


# ------------------------------------------------------------
# No machine-specific strings
# ------------------------------------------------------------

for col in path_columns:

    values = (
        final_df[col]
        .astype(str)
        .str.lower()
    )

    bad_content = (
        values.str.contains(
            "/content/",
            regex=False
        )
        |
        values.str.contains(
            "/content/drive/",
            regex=False
        )
        |
        values.str.contains(
            "mydrive",
            regex=False
        )
    )

    assert int(
        bad_content.sum()
    ) == 0


# ------------------------------------------------------------
# File existence
# ------------------------------------------------------------

missing_files = []

for idx, row in final_df.iterrows():

    for col in path_columns:

        rel = Path(
            row[col]
        )

        full_path = (
            FINAL_ROOT
            / rel
        )

        if not full_path.exists():

            missing_files.append({
                "row": idx,
                "column": col,
                "path": str(rel)
            })


print(
    "Missing final files:",
    len(missing_files)
)

assert len(missing_files) == 0


# ------------------------------------------------------------
# Verify severity formula
# ------------------------------------------------------------

computed_severity = np.where(

    final_df["leaf_pixels"] > 0,

    100.0
    *
    final_df["lesion_inside_leaf"]
    /
    final_df["leaf_pixels"],

    np.nan
)

severity_difference = np.nanmax(
    np.abs(
        computed_severity
        -
        final_df[
            "severity_percent"
        ].to_numpy()
    )
)

print(
    "Maximum severity recalculation difference:",
    severity_difference
)

assert (
    np.isfinite(
        severity_difference
    )
    and
    severity_difference < 1e-9
)


# ============================================================
# STEP 17 — FINAL SUMMARY
# ============================================================

print("\n")
print("=" * 80)
print("FINAL DATASET SUMMARY")
print("=" * 80)

print(
    "\nTotal source records:",
    len(source_df)
)

print(
    "Non-target excluded:",
    len(non_target_df)
)

print(
    "Target records:",
    len(target_df)
)

print(
    "Missing lesion-mask records:",
    len(missing_pair_df)
)

print(
    "Processing errors:",
    len(error_df)
)

print(
    "Final canonical rows:",
    len(final_df)
)

print(
    "\nClass distribution:"
)

print(
    final_df[
        "class_name"
    ].value_counts()
)


print(
    "\nSplit distribution:"
)

print(
    final_df[
        "split"
    ].value_counts()
    .reindex(SPLITS)
)


print(
    "\nQC distribution:"
)

print(
    final_df[
        "qc_status"
    ].value_counts(
        dropna=False
    )
)


print(
    "\nSeverity summary:"
)

print(
    final_df[
        "severity_percent"
    ].describe()
)


# ============================================================
# STEP 18 — FINAL DATASET TREE
# ============================================================

print("\n")
print("=" * 80)
print("FINAL DATASET PATH")
print("=" * 80)

print(
    FINAL_ROOT
)

print(
    "\nCanonical CSV:"
)

print(
    CSV_PATH
)


# ============================================================
# STEP 19 — SHOW PORTABLE CSV EXAMPLES
# ============================================================

print("\n")
print("=" * 80)
print("PORTABLE CSV PATH EXAMPLES")
print("=" * 80)

print(
    final_df[
        [
            "split",
            "filename",
            "image_path",
            "lesion_mask_path",
            "leaf_mask_path"
        ]
    ].head(5).to_string(
        index=False
    )
)


# ============================================================
# STEP 20 — OPTIONAL PACKAGE ZIP
#
# Run this AFTER verifying the dataset.
# ============================================================

# import shutil
#
# ZIP_OUTPUT = Path(
#     "/content/Potato_Severity_Dataset_FINAL"
# )
#
# archive_path = shutil.make_archive(
#     str(ZIP_OUTPUT),
#     "zip",
#     root_dir=str(FINAL_ROOT.parent),
#     base_dir=FINAL_ROOT.name
# )
#
# print("ZIP created:")
# print(archive_path)
```

```python
from pathlib import Path
import pandas as pd
import cv2
import numpy as np

# ============================================================
# FINAL DATASET INSPECTION
# ============================================================

FINAL_ROOT = Path("/content/Potato_Severity_Dataset_FINAL")
CSV_PATH = FINAL_ROOT / "severity_dataset" / "potato_severity_FINAL.csv"

SPLITS = ["train", "valid", "test"]

print("=" * 80)
print("FINAL DATASET INSPECTION")
print("=" * 80)

# ------------------------------------------------------------
# 1. ROOT CHECK
# ------------------------------------------------------------

assert FINAL_ROOT.exists(), f"Dataset not found: {FINAL_ROOT}"
assert CSV_PATH.exists(), f"CSV not found: {CSV_PATH}"

print("\n[1] ROOT")
print(FINAL_ROOT)

# ------------------------------------------------------------
# 2. DIRECTORY STRUCTURE
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("[2] DIRECTORY STRUCTURE")
print("=" * 80)

for split in SPLITS:
    print(f"\n--- {split.upper()} ---")

    for subdir in [
        "images",
        "lesion_masks",
        "leaf_masks"
    ]:
        d = FINAL_ROOT / split / subdir

        files = [
            p for p in d.iterdir()
            if p.is_file()
        ] if d.exists() else []

        print(
            f"{subdir:<15}: "
            f"{len(files):4d} files | "
            f"exists={d.exists()}"
        )

# ------------------------------------------------------------
# 3. CSV
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("[3] CSV")
print("=" * 80)

df = pd.read_csv(CSV_PATH)

print("CSV rows   :", len(df))
print("CSV columns:", len(df.columns))

print("\nColumns:")
print(list(df.columns))

# ------------------------------------------------------------
# 4. EXPECTED COLUMNS
# ------------------------------------------------------------

expected_columns = [
    "split",
    "filename",
    "class_name",
    "image_path",
    "lesion_mask_path",
    "leaf_mask_path",
    "leaf_pixels",
    "lesion_pixels",
    "lesion_inside_leaf",
    "lesion_outside_leaf",
    "severity_percent",
    "leaf_area_ratio",
    "qc_status",
    "qc_flags",
    "shadow_pixels_detected",
    "shadow_pixels_after_close",
    "shadow_pixels_removed",
    "shadow_removed_ratio",
    "shadow_components_total",
    "shadow_components_removed",
]

missing_columns = [
    c for c in expected_columns
    if c not in df.columns
]

extra_columns = [
    c for c in df.columns
    if c not in expected_columns
]

print("\nMissing expected columns:", missing_columns)
print("Extra columns:", extra_columns)

assert not missing_columns

# ------------------------------------------------------------
# 5. CLASS DISTRIBUTION
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("[5] CLASS DISTRIBUTION")
print("=" * 80)

print(
    df["class_name"]
    .value_counts(dropna=False)
)

assert set(df["class_name"].dropna().unique()) <= {
    "Potato Healthy",
    "Potato Late Blight"
}

# ------------------------------------------------------------
# 6. SPLIT DISTRIBUTION
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("[6] SPLIT DISTRIBUTION")
print("=" * 80)

print(
    df["split"]
    .value_counts()
    .reindex(SPLITS)
)

# ------------------------------------------------------------
# 7. DUPLICATES
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("[7] DUPLICATES")
print("=" * 80)

dup_split_filename = df.duplicated(
    subset=["split", "filename"]
).sum()

dup_filename = df["filename"].duplicated().sum()

print("Duplicate (split, filename):", dup_split_filename)
print("Duplicate filename globally:", dup_filename)

assert dup_split_filename == 0

# ------------------------------------------------------------
# 8. MISSING VALUES
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("[8] MISSING VALUES")
print("=" * 80)

for col in [
    "image_path",
    "lesion_mask_path",
    "leaf_mask_path",
    "severity_percent",
    "leaf_pixels",
    "lesion_pixels",
    "lesion_inside_leaf",
]:
    print(
        f"{col:<25}: "
        f"{df[col].isna().sum()}"
    )

# ------------------------------------------------------------
# 9. ABSOLUTE PATH CHECK
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("[9] PATH PORTABILITY")
print("=" * 80)

path_cols = [
    "image_path",
    "lesion_mask_path",
    "leaf_mask_path",
]

for col in path_cols:

    values = df[col].astype(str)

    absolute_unix = values.str.startswith("/")
    absolute_windows = values.str.match(
        r"^[A-Za-z]:[\\/]"
    )

    content_paths = values.str.contains(
        "/content/",
        case=False,
        regex=False
    )

    bad = (
        absolute_unix
        | absolute_windows
        | content_paths
    )

    print(
        f"{col:<25}: "
        f"{bad.sum()} invalid absolute/machine paths"
    )

    assert bad.sum() == 0

# ------------------------------------------------------------
# 10. FILE EXISTENCE
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("[10] FILE EXISTENCE")
print("=" * 80)

missing_files = []

for idx, row in df.iterrows():

    for col in path_cols:

        rel_path = Path(row[col])
        full_path = FINAL_ROOT / rel_path

        if not full_path.exists():
            missing_files.append(
                (
                    idx,
                    col,
                    str(rel_path)
                )
            )

print("Missing files:", len(missing_files))

if missing_files:
    print("\nFirst missing files:")
    for x in missing_files[:20]:
        print(x)

assert len(missing_files) == 0

# ------------------------------------------------------------
# 11. IMAGE / MASK DIMENSION CHECK
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("[11] IMAGE / LESION / LEAF SIZE CHECK")
print("=" * 80)

dimension_errors = []

# inspect every record
for idx, row in df.iterrows():

    image = cv2.imread(
        str(FINAL_ROOT / row["image_path"]),
        cv2.IMREAD_GRAYSCALE
    )

    lesion = cv2.imread(
        str(FINAL_ROOT / row["lesion_mask_path"]),
        cv2.IMREAD_GRAYSCALE
    )

    leaf = cv2.imread(
        str(FINAL_ROOT / row["leaf_mask_path"]),
        cv2.IMREAD_GRAYSCALE
    )

    if image is None or lesion is None or leaf is None:
        dimension_errors.append(
            (idx, "cannot_read")
        )
        continue

    if not (
        image.shape == lesion.shape == leaf.shape
    ):
        dimension_errors.append(
            (
                idx,
                row["filename"],
                image.shape,
                lesion.shape,
                leaf.shape
            )
        )

print(
    "Dimension mismatches:",
    len(dimension_errors)
)

if dimension_errors:
    print("\nFirst mismatches:")
    for x in dimension_errors[:10]:
        print(x)

assert len(dimension_errors) == 0

# ------------------------------------------------------------
# 12. SEVERITY FORMULA VALIDATION
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("[12] SEVERITY VALIDATION")
print("=" * 80)

expected_severity = (
    100.0
    * df["lesion_inside_leaf"]
    / df["leaf_pixels"]
)

difference = (
    expected_severity
    - df["severity_percent"]
).abs()

max_difference = difference.max()

print(
    "Maximum difference:",
    max_difference
)

assert max_difference < 1e-9

# ------------------------------------------------------------
# 13. SEVERITY RANGE
# ------------------------------------------------------------

print("\nSeverity < 0:",
      (df["severity_percent"] < 0).sum())

print("Severity > 100:",
      (df["severity_percent"] > 100).sum())

assert (
    df["severity_percent"] >= 0
).all()

assert (
    df["severity_percent"] <= 100
).all()

# ------------------------------------------------------------
# 14. HEALTHY CHECK
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("[14] HEALTHY IMAGE CHECK")
print("=" * 80)

healthy = df[
    df["class_name"] == "Potato Healthy"
]

healthy_nonzero = healthy[
    healthy["severity_percent"] != 0
]

print(
    "Healthy rows:",
    len(healthy)
)

print(
    "Healthy rows with severity != 0:",
    len(healthy_nonzero)
)

if not healthy_nonzero.empty:

    print(
        healthy_nonzero[
            [
                "filename",
                "severity_percent",
                "lesion_pixels"
            ]
        ].head(20)
    )

# ------------------------------------------------------------
# 15. LESION OVERLAP CONSISTENCY
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("[15] LESION/LEAF CONSISTENCY")
print("=" * 80)

bad_overlap = df[
    (
        df["lesion_inside_leaf"]
        +
        df["lesion_outside_leaf"]
    )
    !=
    df["lesion_pixels"]
]

print(
    "Rows violating lesion decomposition:",
    len(bad_overlap)
)

assert len(bad_overlap) == 0

# ------------------------------------------------------------
# 16. FINAL SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("FINAL INSPECTION SUMMARY")
print("=" * 80)

print(f"Rows                  : {len(df)}")
print(f"Healthy               : {(df.class_name == 'Potato Healthy').sum()}")
print(f"Late Blight           : {(df.class_name == 'Potato Late Blight').sum()}")
print(f"Train                 : {(df.split == 'train').sum()}")
print(f"Valid                 : {(df.split == 'valid').sum()}")
print(f"Test                  : {(df.split == 'test').sum()}")
print(f"Missing files         : {len(missing_files)}")
print(f"Dimension errors      : {len(dimension_errors)}")
print(f"Duplicate records     : {dup_split_filename}")
print(f"Absolute paths        : 0")
print(f"Severity formula max Δ: {max_difference}")
print(f"Severity >100         : {(df.severity_percent > 100).sum()}")

print("\nQC distribution:")
print(
    df["qc_status"]
    .value_counts(dropna=False)
)

print("\n" + "=" * 80)
print("INSPECTION FINISHED")
print("=" * 80)
```

# ***`small comparison`***

```python
# ============================================================
# FIND THE ONE TARGET RECORD MISSING FROM FINAL DATASET
# ============================================================

from pathlib import Path
import pandas as pd

SOURCE_ROOT = Path("/content/Potato_Severity_Dataset")
FINAL_ROOT = Path("/content/Potato_Severity_Dataset_FINAL")

FINAL_CSV = (
    FINAL_ROOT
    / "severity_dataset"
    / "potato_severity_FINAL.csv"
)

TARGET_CLASSES = {
    "Potato Healthy",
    "Potato Late Blight"
}

SPLITS = ["train", "valid", "test"]


def infer_class(path):
    name = path.name.lower()

    if (
        "early_blight" in name
        or "early-blight" in name
        or "earlyblight" in name
    ):
        return "Potato Early Blight"

    if (
        "late_blight" in name
        or "late-blight" in name
        or "lateblight" in name
    ):
        return "Potato Late Blight"

    if "healthy" in name:
        return "Potato Healthy"

    return "Unknown"


# ------------------------------------------------------------
# Build source target set
# ------------------------------------------------------------

source_records = []

for split in SPLITS:

    image_dir = SOURCE_ROOT / split / "images"
    mask_dir = SOURCE_ROOT / split / "lesion_masks"

    mask_stems = {
        p.stem
        for p in mask_dir.iterdir()
        if p.is_file()
    }

    for image_path in image_dir.iterdir():

        if not image_path.is_file():
            continue

        class_name = infer_class(image_path)

        if class_name not in TARGET_CLASSES:
            continue

        source_records.append({
            "split": split,
            "filename": image_path.name,
            "class_name": class_name,
            "has_mask": image_path.stem in mask_stems
        })


source_df = pd.DataFrame(source_records)


# ------------------------------------------------------------
# Final dataset
# ------------------------------------------------------------

final_df = pd.read_csv(FINAL_CSV)

final_keys = set(
    zip(
        final_df["split"],
        final_df["filename"]
    )
)

source_df["in_final"] = source_df.apply(
    lambda r: (
        r["split"],
        r["filename"]
    ) in final_keys,
    axis=1
)


missing = source_df[
    ~source_df["in_final"]
].copy()


print("=" * 80)
print("TARGET RECORD COMPARISON")
print("=" * 80)

print("Source target records :", len(source_df))
print("Final records         :", len(final_df))
print("Missing target        :", len(missing))

print("\nMissing record(s):")

if missing.empty:
    print("NONE")

else:
    print(
        missing.to_string(
            index=False
        )
    )
```
