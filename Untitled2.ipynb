```python
# ============================================================
# POTATO LATE BLIGHT
# EPIDEMIOLOGY-CONSTRAINED SYNTHETIC ENVIRONMENT GENERATOR
# VERSION 0.1 — PROTOTYPE
# ============================================================

import numpy as np
import pandas as pd
from dataclasses import dataclass

SEED = 20260913
rng = np.random.default_rng(SEED)

HOURS_PER_DAY = 24
N_DAYS = 7
N_HOURS = N_DAYS * HOURS_PER_DAY

CULTIVAR = "MS"  # Moderately Susceptible

# ------------------------------------------------------------
# 1. SEVERITY REGIMES
# ------------------------------------------------------------

SEVERITY_REGIMES = {
    "healthy":    (0.0, 0.0),
    "low_1":      (0.000001, 5.0),
    "low_2":      (5.000001, 10.0),
    "moderate_1": (10.000001, 20.0),
    "moderate_2": (20.000001, 30.0),
    "high":       (30.000001, 50.0),
    "very_high":  (50.000001, 100.0),
}


# ------------------------------------------------------------
# 2. MAP IMAGE SEVERITY → LATENT ENVIRONMENTAL RISK
# ------------------------------------------------------------

def severity_to_risk_prior(severity, class_name):
    """
    Returns a latent environmental conduciveness prior in [0,1].

    IMPORTANT:
    This is NOT a deterministic severity -> environment mapping.
    It controls the probability of sampling more conducive weather.
    """

    class_name = str(class_name).lower()

    # Healthy samples are allowed to experience
    # moderately/highly conducive weather.
    if "healthy" in class_name:
        return {
            "mean": 0.25,
            "std": 0.18,
            "min": 0.02,
            "max": 0.75,
        }

    if severity <= 5:
        return {
            "mean": 0.35,
            "std": 0.20,
            "min": 0.05,
            "max": 0.85,
        }

    elif severity <= 10:
        return {
            "mean": 0.43,
            "std": 0.20,
            "min": 0.05,
            "max": 0.90,
        }

    elif severity <= 20:
        return {
            "mean": 0.53,
            "std": 0.19,
            "min": 0.10,
            "max": 0.95,
        }

    elif severity <= 30:
        return {
            "mean": 0.62,
            "std": 0.18,
            "min": 0.15,
            "max": 0.97,
        }

    elif severity <= 50:
        return {
            "mean": 0.70,
            "std": 0.16,
            "min": 0.20,
            "max": 0.99,
        }

    else:
        return {
            "mean": 0.78,
            "std": 0.13,
            "min": 0.30,
            "max": 0.999,
        }


def sample_risk_score(severity, class_name):
    prior = severity_to_risk_prior(severity, class_name)

    x = rng.normal(
        prior["mean"],
        prior["std"]
    )

    return np.clip(
        x,
        prior["min"],
        prior["max"]
    )


# ------------------------------------------------------------
# 3. DAILY TEMPERATURE REGIME
# ------------------------------------------------------------

def generate_temperature(day_index, risk_score):
    """
    Generates realistic daily temperature with diurnal variation.

    Higher risk score slightly favors cooler late-blight-relevant
    conditions, but does NOT force a fixed optimum.
    """

    # Broad disease-relevant daily baseline.
    center = rng.normal(
        loc=18.0 - 1.5 * risk_score,
        scale=2.2
    )

    # Keep inside practical simulation envelope.
    center = np.clip(center, 10.0, 25.5)

    amplitude = rng.uniform(4.0, 7.0)

    hours = np.arange(24)

    # Peak temperature around mid-afternoon.
    phase = 15.0

    temp = (
        center
        + amplitude * np.sin(
            2 * np.pi * (hours - phase) / 24.0
        )
    )

    # Autocorrelated measurement/environment noise.
    noise = rng.normal(0, 0.45, size=24)

    for i in range(1, 24):
        noise[i] = (
            0.75 * noise[i - 1]
            + 0.25 * noise[i]
        )

    temp += noise

    return temp


# ------------------------------------------------------------
# 4. HUMIDITY GENERATION
# ------------------------------------------------------------

def generate_rh(temp, risk_score):
    """
    RH is correlated with temperature and disease-risk regime.

    Night:
        high RH probability

    Day:
        lower RH

    High risk:
        greater persistence of RH >= 90%
    """

    n = len(temp)

    # Base humidity increases with risk score.
    base_rh = 76 + 17 * risk_score

    # Temperature-humidity inverse relation.
    rh = (
        base_rh
        - 0.75 * (temp - np.mean(temp))
    )

    # Nighttime humidification.
    hours = np.arange(n) % 24
    night = (
        (hours >= 21) |
        (hours <= 7)
    )

    rh[night] += rng.normal(
        6.0 + 4.0 * risk_score,
        2.0,
        size=night.sum()
    )

    # Day drying.
    day = ~night

    rh[day] -= rng.normal(
        2.0,
        1.5,
        size=day.sum()
    )

    # temporal noise
    noise = rng.normal(0, 1.5, size=n)

    for i in range(1, n):
        noise[i] = (
            0.8 * noise[i - 1]
            + 0.2 * noise[i]
        )

    rh += noise

    return np.clip(rh, 55.0, 99.5)


# ------------------------------------------------------------
# 5. RAINFALL
# ------------------------------------------------------------

def generate_rainfall(risk_score):
    """
    Discrete rainfall-event process.

    Rainfall is correlated with humid/disease-conducive regimes,
    but not deterministic.
    """

    rain = np.zeros(24)

    # probability of at least one rain event
    p_event = 0.10 + 0.25 * risk_score

    if rng.random() < p_event:

        # 1–5 hour event
        duration = rng.integers(1, 6)

        start = rng.integers(
            0,
            24 - duration + 1
        )

        # rainfall intensity
        amount = rng.gamma(
            shape=1.8,
            scale=1.5
        )

        profile = rng.uniform(
            0.6,
            1.4,
            size=duration
        )

        values = amount * profile
        rain[start:start + duration] = values

    return rain


# ------------------------------------------------------------
# 6. LEAF WETNESS
# ------------------------------------------------------------

def generate_leaf_wetness(
    temperature,
    rh,
    rainfall
):
    """
    Continuous wetness proxy in [0,1].

    NOT simply:
        LWD = RH >= 90

    Wetness has persistence and rainfall influence.
    """

    n = len(rh)

    wet = np.zeros(n)

    for t in range(n):

        score = 0.0

        # Humidity contribution
        if rh[t] >= 95:
            score += 0.75
        elif rh[t] >= 90:
            score += 0.50
        elif rh[t] >= 85:
            score += 0.20

        # Rain contribution
        if rainfall[t] >= 0.1:
            score += 0.45

        # Cooler temperature favors condensation
        if 8 <= temperature[t] <= 22:
            score += 0.15

        # Persistence
        if t > 0:
            score += 0.30 * wet[t - 1]

        probability = np.clip(
            score,
            0,
            1
        )

        wet[t] = (
            1.0
            if rng.random() < probability
            else 0.0
        )

    return wet


# ------------------------------------------------------------
# 7. SOIL MOISTURE
# ------------------------------------------------------------

def generate_soil_moisture(
    previous=None,
    rainfall=None
):
    """
    Slowly varying soil moisture state.
    Rainfall increases soil moisture gradually.
    """

    if previous is None:
        previous = rng.uniform(
            35.0,
            65.0
        )

    soil = np.zeros(24)

    current = previous

    for t in range(24):

        if rainfall is not None:
            current += 0.15 * rainfall[t]

        # natural daily drying
        current -= rng.uniform(
            0.05,
            0.25
        )

        # process noise
        current += rng.normal(
            0,
            0.25
        )

        current = np.clip(
            current,
            15.0,
            85.0
        )

        soil[t] = current

    return soil


# ============================================================
# 8. SIMCAST BLIGHT UNIT LOOKUP
# ============================================================

SIMCAST_TABLE = {

    # temperature range : {
    # cultivar : [(minimum_hours, maximum_hours, BU), ...]
    # }

    "3-7": {
        "S":  [(9, 24, 1)],
        "MS": [(12, 24, 1)],
        "MR": [(18, 24, 1)],
    },

    "8-12": {
        "S": [
            (6, 6, 0),
            (7, 7, 1),
            (8, 9, 2),
            (10, 10, 3),
            (11, 12, 4),
            (13, 15, 5),
            (16, 24, 6),
        ],
        "MS": [
            (6, 6, 0),
            (7, 9, 1),
            (10, 12, 2),
            (13, 15, 3),
            (16, 18, 4),
            (19, 24, 5),
        ],
        "MR": [
            (9, 9, 0),
            (10, 12, 1),
            (13, 15, 2),
            (16, 24, 3),
        ],
    },

    "13-22": {
        "S": [
            (6, 6, 0),
            (7, 9, 5),
            (10, 12, 6),
            (13, 24, 7),
        ],
        "MS": [
            (6, 6, 0),
            (7, 7, 1),
            (8, 8, 2),
            (9, 9, 3),
            (10, 10, 4),
            (11, 12, 5),
            (13, 24, 6),
        ],
        "MR": [
            (6, 6, 0),
            (7, 7, 1),
            (8, 8, 2),
            (9, 9, 3),
            (10, 12, 4),
            (13, 24, 5),
        ],
    },

    "23-27": {
        "S": [
            (6, 6, 0),
            (7, 9, 1),
            (10, 12, 2),
            (13, 15, 3),
            (16, 18, 4),
            (19, 24, 5),
        ],
        "MS": [
            (9, 9, 0),
            (10, 18, 1),
            (19, 24, 2),
        ],
        "MR": [
            (15, 15, 0),
            (16, 24, 1),
        ],
    },

    ">27": {
        "S":  [(24, 24, 0)],
        "MS": [(24, 24, 0)],
        "MR": [(24, 24, 0)],
    },

    "<3": {
        "S":  [(24, 24, 0)],
        "MS": [(24, 24, 0)],
        "MR": [(24, 24, 0)],
    },
}


def temperature_band(temp):
    if temp < 3:
        return "<3"
    elif temp <= 7:
        return "3-7"
    elif temp <= 12:
        return "8-12"
    elif temp <= 22:
        return "13-22"
    elif temp <= 27:
        return "23-27"
    else:
        return ">27"


def simcast_bu_for_spell(mean_temp, high_rh_hours, cultivar="MS"):

    band = temperature_band(mean_temp)

    rules = SIMCAST_TABLE[band][cultivar]

    # If there are no qualifying hours,
    # no BU should be assigned.
    if high_rh_hours <= 0:
        return 0

    for low, high, bu in rules:
        if low <= high_rh_hours <= high:
            return bu

    return 0


# ------------------------------------------------------------
# 9. FIND CONSECUTIVE HIGH-RH SPELLS
# ------------------------------------------------------------

def find_rh_spells(
    temperature,
    rh,
    threshold=90.0
):

    high = rh >= threshold

    spells = []

    start = None

    for i, flag in enumerate(high):

        if flag and start is None:
            start = i

        elif not flag and start is not None:

            end = i - 1

            spells.append({
                "start": start,
                "end": end,
                "hours": end - start + 1,
                "mean_temp": float(
                    np.mean(
                        temperature[start:end + 1]
                    )
                )
            })

            start = None

    if start is not None:

        end = len(high) - 1

        spells.append({
            "start": start,
            "end": end,
            "hours": end - start + 1,
            "mean_temp": float(
                np.mean(
                    temperature[start:end + 1]
                )
            )
        })

    return spells


# ------------------------------------------------------------
# 10. SIMCAST DAILY BU
# ------------------------------------------------------------

def calculate_simcast_daily_bu(
    temperature,
    rh,
    cultivar="MS"
):

    results = []

    for day in range(N_DAYS):

        s = day * 24
        e = s + 24

        T = temperature[s:e]
        RH = rh[s:e]

        spells = find_rh_spells(
            T,
            RH,
            threshold=90.0
        )

        day_bu = 0

        for spell in spells:

            # SIMCAST uses consecutive RH >=90%
            if spell["hours"] <= 0:
                continue

            bu = simcast_bu_for_spell(
                spell["mean_temp"],
                spell["hours"],
                cultivar
            )

            day_bu += bu

        results.append(day_bu)

    return np.array(results)


# ============================================================
# 11. WALLIN SV
# ============================================================

def wallin_sv_from_day(
    mean_temp,
    rh_ge90_hours
):

    h = rh_ge90_hours

    if h <= 0:
        return 0

    # 7.2–11.6 C
    if 7.2 <= mean_temp <= 11.6:

        if h <= 15:
            return 0
        elif h <= 18:
            return 1
        elif h <= 21:
            return 2
        elif h <= 24:
            return 3
        else:
            return 4

    # 11.7–15.0 C
    elif 11.7 <= mean_temp <= 15.0:

        if h <= 12:
            return 0
        elif h <= 15:
            return 1
        elif h <= 18:
            return 2
        elif h <= 21:
            return 3
        else:
            return 4

    # 15.1–26.6 C
    elif 15.1 <= mean_temp <= 26.6:

        if h <= 9:
            return 0
        elif h <= 12:
            return 1
        elif h <= 15:
            return 2
        elif h <= 18:
            return 3
        else:
            return 4

    # Outside Wallin's published table
    return 0


def calculate_wallin_daily_sv(
    temperature,
    rh
):

    daily_sv = []

    for day in range(N_DAYS):

        s = day * 24
        e = s + 24

        T = temperature[s:e]
        RH = rh[s:e]

        spells = find_rh_spells(
            T,
            RH,
            threshold=90.0
        )

        total_sv = 0

        for spell in spells:

            total_sv += wallin_sv_from_day(
                spell["mean_temp"],
                spell["hours"]
            )

        daily_sv.append(total_sv)

    return np.array(daily_sv)


# ============================================================
# 12. HYRE / BLITECAST RAIN-FAVORABLE DAY
# ============================================================

def hyre_favorable_day(
    temperature_day,
    rainfall_10d_total=None
):
    """
    Simplified Hyre criterion from UC IPM:
    favorable day if:
      - 5-day mean temperature < 25.5 C
      - minimum temperature >= 7.2 C
      - cumulative rainfall over previous 10 days >= 3 cm

    For the 7-day prototype, the 10-day rainfall window
    is unavailable, so this function is intentionally NOT
    used as a hard acceptance criterion yet.
    """

    five_day_proxy = np.mean(temperature_day)

    min_temp = np.min(temperature_day)

    if (
        five_day_proxy < 25.5
        and min_temp >= 7.2
    ):
        return True

    return False


# ============================================================
# 13. COMPLETE TRAJECTORY
# ============================================================

def generate_candidate_trajectory(
    severity,
    class_name,
    seed_offset=0
):

    # Independent local RNG for reproducibility
    global rng

    risk = sample_risk_score(
        severity,
        class_name
    )

    temperatures = []
    rhs = []
    rainfall = []
    wetness = []
    soil = []

    previous_soil = None

    for day in range(N_DAYS):

        temp_day = generate_temperature(
            day,
            risk
        )

        rh_day = generate_rh(
            temp_day,
            risk
        )

        rain_day = generate_rainfall(
            risk
        )

        wet_day = generate_leaf_wetness(
            temp_day,
            rh_day,
            rain_day
        )

        soil_day = generate_soil_moisture(
            previous=previous_soil,
            rainfall=rain_day
        )

        previous_soil = soil_day[-1]

        temperatures.extend(temp_day)
        rhs.extend(rh_day)
        rainfall.extend(rain_day)
        wetness.extend(wet_day)
        soil.extend(soil_day)

    temperature = np.array(temperatures)
    rh = np.array(rhs)
    rain = np.array(rainfall)
    wet = np.array(wetness)
    soil = np.array(soil)

    # Epidemiological calculations
    daily_bu = calculate_simcast_daily_bu(
        temperature,
        rh,
        CULTIVAR
    )

    daily_sv = calculate_wallin_daily_sv(
        temperature,
        rh
    )

    # Daily RH>=90 duration
    rh90_hours = []

    for day in range(N_DAYS):

        s = day * 24
        e = s + 24

        rh90_hours.append(
            np.sum(rh[s:e] >= 90)
        )

    rh90_hours = np.array(
        rh90_hours
    )

    cumulative_bu = float(
        np.sum(daily_bu)
    )

    cumulative_sv = float(
        np.sum(daily_sv)
    )

    # Wetness
    wet_hours = float(
        np.sum(wet > 0.5)
    )

    # Rain
    rainfall_total = float(
        np.sum(rain)
    )

    trajectory = pd.DataFrame({
        "hour": np.arange(N_HOURS),
        "temperature": temperature,
        "rh": rh,
        "rainfall_mm": rain,
        "leaf_wetness": wet,
        "soil_moisture": soil,
    })

    summary = {
        "severity_percent": severity,
        "class_name": class_name,
        "risk_score": risk,
        "temperature_mean": np.mean(temperature),
        "temperature_min": np.min(temperature),
        "temperature_max": np.max(temperature),
        "rh_mean": np.mean(rh),
        "rh_min": np.min(rh),
        "rh_max": np.max(rh),
        "rh_ge90_hours": int(
            np.sum(rh >= 90)
        ),
        "leaf_wetness_hours": int(
            wet_hours
        ),
        "rainfall_total_mm": rainfall_total,
        "soil_moisture_mean": np.mean(soil),
        "simcast_bu": cumulative_bu,
        "wallin_sv": cumulative_sv,
        "daily_simcast_bu": daily_bu.tolist(),
        "daily_wallin_sv": daily_sv.tolist(),
    }

    return trajectory, summary


print("Generator engine loaded successfully.")
print("Seed:", SEED)
print("Trajectory length:", N_HOURS, "hours")
print("Cultivar assumption:", CULTIVAR)
```

# **prototype - 02**

```python
# ============================================================
# POTATO LATE BLIGHT — CORRECTED PROTOTYPE ENGINE
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# CONFIG
# ------------------------------------------------------------

SEED = 42

N_DAYS = 10
HOURS_PER_DAY = 24
N_HOURS = N_DAYS * HOURS_PER_DAY

N_CANDIDATES = 10

# Example image severity
# Change this to test other severity levels later
EXAMPLE_SEVERITY = 42.0

# Assume susceptible potato cultivar for SIMCAST.
# We are NOT using cultivar resistance as a research variable yet.
CULTIVAR = "S"


# ============================================================
# 1. BASIC UTILITIES
# ============================================================

def consecutive_runs(binary_array):
    """
    Return lengths of consecutive True runs.

    Example:
    [False, True, True, False, True]
    -> [2, 1]
    """
    arr = np.asarray(binary_array, dtype=bool)

    runs = []
    current = 0

    for x in arr:
        if x:
            current += 1
        else:
            if current > 0:
                runs.append(current)
                current = 0

    if current > 0:
        runs.append(current)

    return runs


def longest_true_run(binary_array):
    runs = consecutive_runs(binary_array)
    return max(runs) if runs else 0


# ============================================================
# 2. SIMCAST / FRY BLIGHT UNITS
# ============================================================

# Published Fry/SIMCAST table.
#
# For a SUSCEPTIBLE cultivar:
#
# Temperature       BU by consecutive RH >= 90 h
#
# >27 C             24 h -> BU 0
#
# 23-27 C:
#   6 h       -> 0
#   7-9       -> 1
#   10-12     -> 2
#   13-15     -> 3
#   16-18     -> 4
#   19-24     -> 5
#
# 13-22 C:
#   6 h       -> 0
#   7-9       -> 5
#   10-12     -> 6
#   13-24     -> 7
#
# 8-12 C:
#   6 h       -> 0
#   7         -> 1
#   8-9       -> 2
#   10        -> 3
#   11-12     -> 4
#   13-15     -> 5
#   16-24     -> 6
#
# 3-7 C:
#   9 h       -> 0
#   10-12     -> 1
#   13-15     -> 2
#   16-18     -> 3
#   19-24     -> 4
#
# <3 C:
#   24 h -> 0
#
# Source: UC IPM Fry/SIMCAST table.
# ============================================================


def simcast_bu_susceptible(temp_mean, consecutive_rh90_hours):
    """
    Return Fry/SIMCAST blight units for susceptible cultivar.

    IMPORTANT:
    Uses consecutive RH >= 90 h, not total RH >= 90 h.
    """

    T = temp_mean
    H = consecutive_rh90_hours

    # No meaningful qualifying period
    if H <= 0:
        return 0

    # >27 C
    if T > 27:
        return 0

    # 23-27 C
    elif 23 <= T <= 27:

        if H < 7:
            return 0
        elif H <= 9:
            return 1
        elif H <= 12:
            return 2
        elif H <= 15:
            return 3
        elif H <= 18:
            return 4
        else:
            return 5

    # 13-22 C
    elif 13 <= T <= 22:

        if H < 7:
            return 0
        elif H <= 9:
            return 5
        elif H <= 12:
            return 6
        else:
            return 7

    # 8-12 C
    elif 8 <= T <= 12:

        if H < 7:
            return 0
        elif H == 7:
            return 1
        elif H <= 9:
            return 2
        elif H == 10:
            return 3
        elif H <= 12:
            return 4
        elif H <= 15:
            return 5
        else:
            return 6

    # 3-7 C
    elif 3 <= T <= 7:

        if H < 10:
            return 0
        elif H <= 12:
            return 1
        elif H <= 15:
            return 2
        elif H <= 18:
            return 3
        else:
            return 4

    # <3 C
    else:
        return 0


def calculate_simcast_daily(df):
    """
    Calculate daily SIMCAST/Fry BU.

    We use 24-hour periods.

    Prototype convention:
    day 0 = hours 0-23
    day 1 = hours 24-47
    ...

    For each day:
      1. Find RH >= 90 periods.
      2. For each qualifying period, determine its temperature.
      3. Use the strongest qualifying consecutive period for the
         day's BU calculation.

    This avoids the old incorrect approach of simply summing
    arbitrary BU values from disconnected RH spells.
    """

    daily_rows = []

    for day in range(N_DAYS):

        start = day * 24
        end = start + 24

        d = df.iloc[start:end].copy()

        rh90 = d["rh"] >= 90.0

        # Find each RH>=90 spell
        runs = []
        start_idx = None

        for i, flag in enumerate(rh90):

            if flag and start_idx is None:
                start_idx = i

            elif not flag and start_idx is not None:
                runs.append((start_idx, i - 1))
                start_idx = None

        if start_idx is not None:
            runs.append((start_idx, len(d) - 1))

        best_bu = 0
        best_hours = 0
        best_temp = np.nan

        for a, b in runs:

            spell = d.iloc[a:b+1]

            h = len(spell)

            mean_t = spell["temperature"].mean()

            bu = simcast_bu_susceptible(
                mean_t,
                h
            )

            # Keep the strongest spell
            if bu > best_bu or (bu == best_bu and h > best_hours):
                best_bu = bu
                best_hours = h
                best_temp = mean_t

        daily_rows.append({
            "day": day + 1,
            "rh_ge90_hours_total": int(rh90.sum()),
            "max_consecutive_rh90_hours": int(
                longest_true_run(rh90.values)
            ),
            "simcast_bu": int(best_bu),
            "simcast_temp_during_best_spell": best_temp
        })

    return pd.DataFrame(daily_rows)


# ============================================================
# 3. WALLIN SEVERITY VALUE
# ============================================================

def wallin_sv(mean_temp, rh90_hours):
    """
    Wallin severity value.

    Uses:
      - total hours RH >= 90%
      - average temperature during those RH>=90 hours

    Published table:

    7.2-11.6 C:
        15       -> 0
        16-18    -> 1
        19-21    -> 2
        22-24    -> 3
        >=25     -> 4

    11.7-15.0 C:
        12       -> 0
        13-15    -> 1
        16-18    -> 2
        19-21    -> 3
        >=22     -> 4

    15.1-26.6 C:
        9        -> 0
        10-12    -> 1
        13-15    -> 2
        16-18    -> 3
        >=19     -> 4
    """

    if rh90_hours <= 0 or np.isnan(mean_temp):
        return 0

    T = mean_temp
    H = rh90_hours

    # 7.2-11.6
    if 7.2 <= T <= 11.6:

        if H <= 15:
            return 0
        elif H <= 18:
            return 1
        elif H <= 21:
            return 2
        elif H <= 24:
            return 3
        else:
            return 4

    # 11.7-15.0
    elif 11.7 <= T <= 15.0:

        if H <= 12:
            return 0
        elif H <= 15:
            return 1
        elif H <= 18:
            return 2
        elif H <= 21:
            return 3
        else:
            return 4

    # 15.1-26.6
    elif 15.1 <= T <= 26.6:

        if H <= 9:
            return 0
        elif H <= 12:
            return 1
        elif H <= 15:
            return 2
        elif H <= 18:
            return 3
        else:
            return 4

    # Outside Wallin table
    else:
        return 0


def calculate_wallin_daily(df):

    daily_rows = []

    for day in range(N_DAYS):

        start = day * 24
        end = start + 24

        d = df.iloc[start:end].copy()

        high_rh = d[d["rh"] >= 90.0]

        H = len(high_rh)

        if H > 0:
            mean_temp = high_rh["temperature"].mean()
            min_temp = high_rh["temperature"].min()
            max_temp = high_rh["temperature"].max()
        else:
            mean_temp = np.nan
            min_temp = np.nan
            max_temp = np.nan

        sv = wallin_sv(
            mean_temp,
            H
        )

        daily_rows.append({
            "day": day + 1,
            "wallin_rh_ge90_hours": int(H),
            "wallin_temp_mean": mean_temp,
            "wallin_temp_min": min_temp,
            "wallin_temp_max": max_temp,
            "wallin_sv": int(sv)
        })

    return pd.DataFrame(daily_rows)


# ============================================================
# 4. HYRE / BLITECAST RAIN-FAVORABLE DAY
# ============================================================

def calculate_hyre(df):
    """
    Hyre criterion as used by BLITECAST:

    A day is favorable when:
      - 5-day average temperature < 25.5 C
      - minimum temperature >= 7.2 C
      - rainfall over preceding 10 days >= 30 mm

    Because our prototype itself is 10 days long,
    the 10-day rainfall condition becomes progressively available.

    We calculate the condition for each day using data available
    through that day.
    """

    daily = df.groupby("day").agg(
        temperature_mean=("temperature", "mean"),
        temperature_min=("temperature", "min"),
        temperature_max=("temperature", "max"),
        rainfall_mm=("rainfall", "sum")
    ).reset_index()

    daily["temp_5day_mean"] = (
        daily["temperature_mean"]
        .rolling(window=5, min_periods=5)
        .mean()
    )

    daily["rain_10day_total"] = (
        daily["rainfall_mm"]
        .rolling(window=10, min_periods=10)
        .sum()
    )

    daily["hyre_favorable"] = (
        (daily["temp_5day_mean"] < 25.5)
        &
        (daily["temperature_min"] >= 7.2)
        &
        (daily["rain_10day_total"] >= 30.0)
    )

    return daily


# ============================================================
# 5. WEATHER TRAJECTORY GENERATOR
# ============================================================

def generate_weather_trajectory(
    severity_percent,
    candidate_id,
    seed
):
    """
    Generate one 10-day synthetic environmental trajectory.

    IMPORTANT:
    Severity is NOT converted directly into temperature/RH.

    Instead, severity only shifts a latent environmental-risk
    tendency. Noise + temporal correlation preserve overlap
    between severity levels.
    """

    rng = np.random.default_rng(seed)

    hours = np.arange(N_HOURS)

    # --------------------------------------------------------
    # Latent risk
    # --------------------------------------------------------
    #
    # Convert severity into a soft prior.
    # This is intentionally broad.
    #
    # 0% severity -> lower risk tendency
    # 90% severity -> higher risk tendency
    #
    # BUT substantial noise is retained.
    #

    severity_norm = np.clip(
        severity_percent / 90.0,
        0,
        1
    )

    latent_risk = (
        0.25
        + 0.50 * severity_norm
        + rng.normal(0, 0.14)
    )

    latent_risk = float(
        np.clip(latent_risk, 0.0, 1.0)
    )

    # --------------------------------------------------------
    # Daily weather state
    # --------------------------------------------------------

    daily_temp_mean = []
    daily_rh_bias = []
    daily_rain = []

    temp_state = 17.0 + rng.normal(0, 1.0)
    rh_state = 0.0

    for day in range(N_DAYS):

        # Slowly varying temperature state
        temp_state = (
            0.75 * temp_state
            + 0.25 * (
                17.0
                + rng.normal(0, 2.0)
            )
        )

        # Higher latent risk tends to make humidity more persistent,
        # NOT deterministically higher.
        target_rh_bias = (
            -2
            + 14 * latent_risk
            + rng.normal(0, 4)
        )

        rh_state = (
            0.75 * rh_state
            + 0.25 * target_rh_bias
        )

        # Rain event probability
        rain_probability = (
            0.18
            + 0.40 * latent_risk
        )

        rain_event = rng.random() < rain_probability

        if rain_event:
            # Gamma gives mostly small events and occasional larger ones
            rain_total = rng.gamma(
                shape=1.8,
                scale=3.0
            )

            # Cap extreme synthetic events
            rain_total = min(rain_total, 18.0)

        else:
            rain_total = 0.0

        daily_temp_mean.append(temp_state)
        daily_rh_bias.append(rh_state)
        daily_rain.append(rain_total)

    daily_temp_mean = np.array(daily_temp_mean)
    daily_rh_bias = np.array(daily_rh_bias)
    daily_rain = np.array(daily_rain)

    # --------------------------------------------------------
    # Hourly generation
    # --------------------------------------------------------

    temperature = np.zeros(N_HOURS)
    rh = np.zeros(N_HOURS)
    rainfall = np.zeros(N_HOURS)

    for day in range(N_DAYS):

        start = day * 24
        end = start + 24

        h = np.arange(24)

        # Temperature daily cycle:
        # coolest near early morning,
        # warmest around afternoon.
        phase = 15.0

        temp_cycle = (
            3.8
            * np.sin(
                2 * np.pi * (h - phase) / 24
            )
        )

        temp_noise = rng.normal(
            0,
            0.45,
            size=24
        )

        day_temp = (
            daily_temp_mean[day]
            + temp_cycle
            + temp_noise
        )

        # Keep within plausible prototype envelope
        day_temp = np.clip(
            day_temp,
            7.0,
            27.0
        )

        temperature[start:end] = day_temp

        # ----------------------------------------------------
        # RH
        # ----------------------------------------------------

        # RH tends to be high at night/early morning
        rh_cycle = (
            7.5
            * np.cos(
                2 * np.pi * (h - 4) / 24
            )
        )

        # Rain increases humidity persistently.
        rain_effect = min(
            daily_rain[day] * 0.65,
            10.0
        )

        # AR(1)-like hourly noise
        noise = np.zeros(24)

        for k in range(1, 24):
            noise[k] = (
                0.82 * noise[k - 1]
                + rng.normal(0, 1.0)
            )

        day_rh = (
            86.0
            + daily_rh_bias[day]
            + rh_cycle
            + rain_effect
            + noise
        )

        day_rh = np.clip(
            day_rh,
            65.0,
            99.5
        )

        rh[start:end] = day_rh

        # ----------------------------------------------------
        # Rainfall
        # ----------------------------------------------------

        if daily_rain[day] > 0:

            # Concentrate rain into a few hourly events
            weights = rng.gamma(
                shape=0.8,
                scale=1.0,
                size=24
            )

            # Favor afternoon/evening slightly
            hour_factor = (
                0.7
                + 0.6
                * np.exp(
                    -((h - 16) ** 2) / 30
                )
            )

            weights *= hour_factor

            weights /= weights.sum()

            rainfall[start:end] = (
                daily_rain[day]
                * weights
            )

        else:
            rainfall[start:end] = 0.0

    # --------------------------------------------------------
    # Leaf wetness
    # --------------------------------------------------------
    #
    # Wetness is related to RH AND rainfall.
    # It is not simply RH >= 90.
    #

    leaf_wetness = np.zeros(N_HOURS, dtype=int)

    wet_state = 0

    for t in range(N_HOURS):

        rain_here = rainfall[t] > 0.05

        high_rh = rh[t] >= 90.0

        # Rain can initiate wetness.
        if rain_here:
            wet_state = 1

        # High RH sustains wetness.
        elif high_rh and wet_state == 1:
            wet_state = 1

        # If RH becomes sufficiently low,
        # leaf dries.
        elif rh[t] < 82:
            wet_state = 0

        # Intermediate RH has some persistence.
        else:
            if rng.random() < 0.15:
                wet_state = 0

        leaf_wetness[t] = wet_state

    # --------------------------------------------------------
    # Soil moisture
    # --------------------------------------------------------

    soil_moisture = np.zeros(N_HOURS)

    sm = (
        35
        + 10 * latent_risk
        + rng.normal(0, 3)
    )

    for t in range(N_HOURS):

        recharge = (
            rainfall[t]
            * 0.35
        )

        evaporation = (
            max(temperature[t] - 15, 0)
            * 0.018
        )

        drainage = (
            max(sm - 65, 0)
            * 0.03
        )

        noise = rng.normal(
            0,
            0.15
        )

        sm = (
            sm
            + recharge
            - evaporation
            - drainage
            + noise
        )

        sm = np.clip(
            sm,
            20,
            75
        )

        soil_moisture[t] = sm

    # --------------------------------------------------------
    # DataFrame
    # --------------------------------------------------------

    out = pd.DataFrame({
        "candidate_id": candidate_id,
        "day": np.repeat(
            np.arange(1, N_DAYS + 1),
            24
        ),
        "hour": np.tile(
            np.arange(24),
            N_DAYS
        ),
        "temperature": temperature,
        "rh": rh,
        "rainfall": rainfall,
        "leaf_wetness": leaf_wetness,
        "soil_moisture": soil_moisture
    })

    out["severity_percent"] = severity_percent
    out["latent_risk"] = latent_risk

    return out


# ============================================================
# 6. COMPLETE EPIDEMIOLOGICAL EVALUATION
# ============================================================

def evaluate_trajectory(df):

    simcast_daily = calculate_simcast_daily(df)
    wallin_daily = calculate_wallin_daily(df)
    hyre_daily = calculate_hyre(df)

    # Merge daily model outputs
    daily = (
        simcast_daily
        .merge(
            wallin_daily,
            on="day",
            how="left"
        )
        .merge(
            hyre_daily[
                [
                    "day",
                    "temperature_mean",
                    "temperature_min",
                    "temperature_max",
                    "rainfall_mm",
                    "temp_5day_mean",
                    "rain_10day_total",
                    "hyre_favorable"
                ]
            ],
            on="day",
            how="left"
        )
    )

    # Overall metrics
    total_rain = df["rainfall"].sum()

    rh90_hours = int(
        (df["rh"] >= 90).sum()
    )

    lwd_hours = int(
        df["leaf_wetness"].sum()
    )

    summary = {
        "candidate_id": int(df["candidate_id"].iloc[0]),

        "severity_percent":
            float(df["severity_percent"].iloc[0]),

        "temperature_mean":
            float(df["temperature"].mean()),

        "temperature_min":
            float(df["temperature"].min()),

        "temperature_max":
            float(df["temperature"].max()),

        "rh_mean":
            float(df["rh"].mean()),

        "rh_min":
            float(df["rh"].min()),

        "rh_max":
            float(df["rh"].max()),

        "rh_ge90_hours":
            rh90_hours,

        "leaf_wetness_hours":
            lwd_hours,

        "rainfall_total_mm":
            float(total_rain),

        "soil_moisture_mean":
            float(df["soil_moisture"].mean()),

        "simcast_bu":
            float(daily["simcast_bu"].sum()),

        "wallin_sv":
            float(daily["wallin_sv"].sum()),

        "hyre_favorable_days":
            int(daily["hyre_favorable"].sum()),

        "max_daily_wallin_sv":
            int(daily["wallin_sv"].max()),

        "max_daily_simcast_bu":
            int(daily["simcast_bu"].max()),

        "latent_risk":
            float(df["latent_risk"].iloc[0])
    }

    return daily, summary


# ============================================================
# 7. GENERATE 10 CANDIDATES
# ============================================================

candidate_trajectories = []
candidate_summaries = []
candidate_daily = []

for candidate_id in range(N_CANDIDATES):

    seed = SEED + candidate_id

    trajectory = generate_weather_trajectory(
        severity_percent=EXAMPLE_SEVERITY,
        candidate_id=candidate_id,
        seed=seed
    )

    daily, summary = evaluate_trajectory(
        trajectory
    )

    candidate_trajectories.append(
        trajectory
    )

    candidate_summaries.append(
        summary
    )

    daily["candidate_id"] = candidate_id

    candidate_daily.append(
        daily
    )


summary_df = pd.DataFrame(
    candidate_summaries
)

daily_df = pd.concat(
    candidate_daily,
    ignore_index=True
)

hourly_df = pd.concat(
    candidate_trajectories,
    ignore_index=True
)


# ============================================================
# 8. DISPLAY RESULTS
# ============================================================

pd.set_option(
    "display.max_columns",
    None
)

print("=" * 80)
print("10-DAY SYNTHETIC ENVIRONMENT PROTOTYPE")
print("=" * 80)

print(
    summary_df.round(3).to_string(index=False)
)


print("\n")
print("=" * 80)
print("DAILY EPIDEMIOLOGICAL OUTPUT — CANDIDATE 0")
print("=" * 80)

print(
    daily_df[
        daily_df["candidate_id"] == 0
    ].round(3).to_string(index=False)
)


# ============================================================
# 9. PLOT CANDIDATE 0
# ============================================================

c0 = hourly_df[
    hourly_df["candidate_id"] == 0
].copy()

fig, ax1 = plt.subplots(
    figsize=(15, 6)
)

ax1.plot(
    c0["hour"].values
    + (c0["day"].values - 1) * 24,
    c0["temperature"],
    label="Temperature"
)

ax1.set_xlabel("Hour")
ax1.set_ylabel("Temperature (°C)")

ax2 = ax1.twinx()

ax2.plot(
    c0["hour"].values
    + (c0["day"].values - 1) * 24,
    c0["rh"],
    label="RH"
)

ax2.axhline(
    90,
    linestyle="--",
    linewidth=1
)

ax2.set_ylabel("Relative Humidity (%)")

plt.title(
    "Corrected Prototype — 10-Day Synthetic Potato Environment"
)

plt.tight_layout()
plt.show()


# ============================================================
# 10. EPIDEMIOLOGICAL SUMMARY PLOTS
# ============================================================

fig, ax = plt.subplots(
    figsize=(12, 5)
)

ax.bar(
    summary_df["candidate_id"],
    summary_df["simcast_bu"]
)

ax.set_xlabel("Candidate")
ax.set_ylabel("Cumulative SIMCAST BU")

ax.set_title(
    "SIMCAST / Fry Cumulative BU Across Candidates"
)

plt.tight_layout()
plt.show()


fig, ax = plt.subplots(
    figsize=(12, 5)
)

ax.bar(
    summary_df["candidate_id"],
    summary_df["wallin_sv"]
)

ax.set_xlabel("Candidate")
ax.set_ylabel("Cumulative Wallin SV")

ax.set_title(
    "Wallin Cumulative Severity Values Across Candidates"
)

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# SELF-CONTAINED 10-DAY SYNTHETIC ENVIRONMENT PROTOTYPE
# APPROACH B — Epidemiology-Constrained Candidate Generation
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# GLOBAL SETTINGS
# ------------------------------------------------------------
SEED = 42
rng_global = np.random.default_rng(SEED)

DAYS = 10
HOURS_PER_DAY = 24


# ============================================================
# 1. SIMCAST / FRY BU TABLE
# ============================================================

def simcast_bu_from_spell(temp_c, rh90_hours):
    """
    Prototype implementation of Fry/SIMCAST BU assignment.

    Input:
        temp_c       = mean temperature during a qualifying RH>=90 spell
        rh90_hours   = consecutive RH>=90 hours

    Returns:
        BU value for that spell
    """

    h = int(rh90_hours)
    t = float(temp_c)

    # Very cold / very hot
    if t > 27:
        return 0

    if t < 3:
        return 0

    # 23–27 C
    if 23 <= t <= 27:
        if h <= 6:
            return 0
        elif h <= 9:
            return 1
        elif h <= 12:
            return 2
        elif h <= 15:
            return 3
        elif h <= 18:
            return 4
        else:
            return 5

    # 13–22 C
    if 13 <= t <= 22:
        if h <= 6:
            return 0
        elif h <= 9:
            return 5
        elif h <= 12:
            return 6
        else:
            return 7

    # 8–12 C
    if 8 <= t <= 12:
        if h <= 6:
            return 0
        elif h == 7:
            return 1
        elif h <= 9:
            return 2
        elif h == 10:
            return 3
        elif h <= 12:
            return 4
        elif h <= 15:
            return 5
        else:
            return 6

    # 3–7 C
    if 3 <= t <= 7:
        if h <= 9:
            return 0
        elif h <= 12:
            return 1
        elif h <= 15:
            return 2
        elif h <= 18:
            return 3
        else:
            return 4

    return 0


# ============================================================
# 2. FIND CONSECUTIVE RH >= 90 SPELLS
# ============================================================

def get_rh90_spells(rh, temp):
    """
    Find consecutive RH>=90 hour blocks.

    Returns list of dictionaries:
        start_idx
        end_idx
        duration
        mean_temp
        min_temp
        max_temp
    """

    rh = np.asarray(rh)
    temp = np.asarray(temp)

    mask = rh >= 90.0

    spells = []
    start = None

    for i, flag in enumerate(mask):

        if flag and start is None:
            start = i

        elif not flag and start is not None:

            end = i

            block_temp = temp[start:end]

            spells.append({
                "start_idx": start,
                "end_idx": end - 1,
                "duration": end - start,
                "mean_temp": float(np.mean(block_temp)),
                "min_temp": float(np.min(block_temp)),
                "max_temp": float(np.max(block_temp))
            })

            start = None

    # Handle spell extending to final hour
    if start is not None:

        end = len(mask)

        block_temp = temp[start:end]

        spells.append({
            "start_idx": start,
            "end_idx": end - 1,
            "duration": end - start,
            "mean_temp": float(np.mean(block_temp)),
            "min_temp": float(np.min(block_temp)),
            "max_temp": float(np.max(block_temp))
        })

    return spells


# ============================================================
# 3. WALLIN SEVERITY VALUE
# ============================================================

def wallin_sv_from_hours(temp_c, rh90_hours):
    """
    Wallin SV assignment.

    This uses the RH>=90 duration and average temperature
    during the wet/humid period.

    Outside the Wallin temperature table -> 0.
    """

    t = float(temp_c)
    h = int(rh90_hours)

    # 7.2–11.6 C
    if 7.2 <= t <= 11.6:

        if h <= 15:
            return 0
        elif h <= 18:
            return 1
        elif h <= 21:
            return 2
        elif h <= 24:
            return 3
        else:
            return 4

    # 11.7–15.0 C
    elif 11.7 <= t <= 15.0:

        if h <= 12:
            return 0
        elif h <= 15:
            return 1
        elif h <= 18:
            return 2
        elif h <= 21:
            return 3
        else:
            return 4

    # 15.1–26.6 C
    elif 15.1 <= t <= 26.6:

        if h <= 9:
            return 0
        elif h <= 12:
            return 1
        elif h <= 15:
            return 2
        elif h <= 18:
            return 3
        else:
            return 4

    return 0


# ============================================================
# 4. HYRE FAVORABILITY
# ============================================================

def calculate_hyre_status(daily_temp_mean,
                          daily_temp_min,
                          rainfall_mm,
                          previous_temp_means,
                          previous_rainfall):

    """
    Prototype Hyre contextual check.

    Favorable day if:
        5-day mean temp >= 25.5 C
        OR rainfall 10-day < 30 mm
        OR minimum temp >= 7.2 C

    This is used as a contextual signal rather than
    disease ground truth.
    """

    temp_history = previous_temp_means + [daily_temp_mean]
    rain_history = previous_rainfall + [rainfall_mm]

    if len(temp_history) >= 5:
        temp_5day = float(np.mean(temp_history[-5:]))
    else:
        temp_5day = np.nan

    rain_10day = float(np.sum(rain_history[-10:]))

    if len(temp_history) >= 5:

        favorable = (
            (temp_5day >= 25.5)
            and
            (rain_10day >= 30.0)
            and
            (daily_temp_min < 7.2)
        )

        # Hyre unfavorable conditions imply not favorable
        # for late-blight development.
        hyre_favorable = not favorable

    else:
        hyre_favorable = False

    return temp_5day, rain_10day, hyre_favorable


# ============================================================
# 5. LATENT RISK PRIOR FROM IMAGE SEVERITY
# ============================================================

def severity_to_latent_risk(severity_percent, rng):
    """
    Soft prior only.

    IMPORTANT:
    This does NOT define disease severity from weather.
    It only changes the probability of sampling more
    disease-conducive environmental trajectories.
    """

    s = np.clip(float(severity_percent), 0, 100)

    # Nonlinear mapping:
    # low severity -> still allows disease-conducive environments
    # higher severity -> stronger tendency toward favorable regimes

    normalized = s / 100.0

    base_risk = 0.15 + 0.65 * (normalized ** 0.70)

    noise = rng.normal(0, 0.08)

    risk = np.clip(base_risk + noise, 0.05, 0.95)

    return float(risk)


# ============================================================
# 6. HOURLY ENVIRONMENT GENERATOR
# ============================================================

def generate_hourly_environment(severity_percent, rng, days=10):

    latent_risk = severity_to_latent_risk(
        severity_percent,
        rng
    )

    total_hours = days * 24

    hours = np.arange(total_hours)

    # --------------------------------------------------------
    # Temperature
    # --------------------------------------------------------

    # Disease-favorable temperatures centered roughly around
    # mid-teens to low-20s for the prototype.

    daily_mean_base = (
        14.0
        + 6.5 * latent_risk
        + rng.normal(0, 1.2, days)
    )

    temperature = np.zeros(total_hours)

    for d in range(days):

        daily_mean = daily_mean_base[d]

        start = d * 24
        end = start + 24

        h = np.arange(24)

        # Sinusoidal diurnal pattern
        temp_day = (
            daily_mean
            + 4.0 * np.sin(
                2 * np.pi * (h - 8) / 24
            )
            + rng.normal(0, 0.65, 24)
        )

        temperature[start:end] = temp_day

    temperature = np.clip(
        temperature,
        3.0,
        27.0
    )


    # --------------------------------------------------------
    # Relative Humidity
    # --------------------------------------------------------

    rh_center = (
        78
        + 17 * latent_risk
        + rng.normal(0, 2.0, days)
    )

    rh = np.zeros(total_hours)

    for d in range(days):

        center = rh_center[d]

        start = d * 24
        end = start + 24

        h = np.arange(24)

        # Humidity rises during night / early morning
        diurnal = (
            8.0
            * np.cos(
                2 * np.pi * (h - 4) / 24
            )
        )

        rh_day = (
            center
            + diurnal
            + rng.normal(0, 3.0, 24)
        )

        rh[start:end] = rh_day

    rh = np.clip(rh, 60.0, 99.5)


    # --------------------------------------------------------
    # Rainfall
    # --------------------------------------------------------

    rain_probability = (
        0.05
        + 0.35 * latent_risk
    )

    rainfall_daily = []

    for d in range(days):

        if rng.random() < rain_probability:

            rain = rng.gamma(
                shape=1.8,
                scale=4.0
            )

        else:

            rain = 0.0

        rainfall_daily.append(
            float(rain)
        )


    # --------------------------------------------------------
    # Leaf wetness
    # --------------------------------------------------------

    leaf_wetness_hours = np.zeros(days)

    for d in range(days):

        start = d * 24
        end = start + 24

        rh_day = rh[start:end]

        # Base wetness when RH>=90
        wet_hours = np.sum(rh_day >= 90)

        # Rain additionally extends wetness
        rain_extra = min(
            6.0,
            rainfall_daily[d] * 0.7
        )

        lw = wet_hours + rain_extra

        # Small measurement/environment variation
        lw += rng.normal(0, 1.5)

        leaf_wetness_hours[d] = np.clip(
            lw,
            0,
            24
        )


    # --------------------------------------------------------
    # Soil Moisture
    # --------------------------------------------------------

    soil_moisture = np.zeros(days)

    current_sm = (
        25
        + 20 * latent_risk
        + rng.normal(0, 3)
    )

    for d in range(days):

        rain_effect = rainfall_daily[d] * 0.9

        drying = (
            0.20
            * max(
                0,
                daily_mean_base[d] - 15
            )
        )

        current_sm = (
            current_sm
            + rain_effect
            - drying
            + rng.normal(0, 1.5)
        )

        current_sm = np.clip(
            current_sm,
            10,
            70
        )

        soil_moisture[d] = current_sm


    return {
        "temperature": temperature,
        "rh": rh,
        "rainfall_daily": rainfall_daily,
        "leaf_wetness_daily": leaf_wetness_hours,
        "soil_moisture_daily": soil_moisture,
        "latent_risk": latent_risk
    }


# ============================================================
# 7. MAIN CANDIDATE GENERATOR
# ============================================================

def generate_environment_candidate(
    severity_percent,
    candidate_id,
    seed,
    days=10
):

    rng = np.random.default_rng(seed)

    env = generate_hourly_environment(
        severity_percent=severity_percent,
        rng=rng,
        days=days
    )

    temperature = env["temperature"]
    rh = env["rh"]

    rainfall_daily = env["rainfall_daily"]
    leaf_wetness_daily = env["leaf_wetness_daily"]
    soil_moisture_daily = env["soil_moisture_daily"]

    latent_risk = env["latent_risk"]


    # ========================================================
    # DAILY EPIDEMIOLOGICAL CALCULATION
    # ========================================================

    daily_records = []

    cumulative_simcast_bu = 0
    cumulative_wallin_sv = 0

    previous_temp_means = []
    previous_rainfall = []

    for day in range(days):

        start = day * 24
        end = start + 24

        temp_day = temperature[start:end]
        rh_day = rh[start:end]

        spells = get_rh90_spells(
            rh_day,
            temp_day
        )

        rh90_hours_total = int(
            np.sum(rh_day >= 90)
        )

        if len(spells) > 0:

            best_spell = max(
                spells,
                key=lambda x: x["duration"]
            )

            best_duration = best_spell["duration"]
            best_temp = best_spell["mean_temp"]

            simcast_bu = simcast_bu_from_spell(
                best_temp,
                best_duration
            )

            wallin_hours = rh90_hours_total

            rh90_indices = np.where(
                rh_day >= 90
            )[0]

            if len(rh90_indices) > 0:

                wallin_temp_values = temp_day[
                    rh90_indices
                ]

                wallin_temp_mean = float(
                    np.mean(wallin_temp_values)
                )

                wallin_temp_min = float(
                    np.min(wallin_temp_values)
                )

                wallin_temp_max = float(
                    np.max(wallin_temp_values)
                )

            else:

                wallin_temp_mean = np.nan
                wallin_temp_min = np.nan
                wallin_temp_max = np.nan

            wallin_sv = wallin_sv_from_hours(
                wallin_temp_mean
                if not np.isnan(wallin_temp_mean)
                else 0,
                wallin_hours
            )

        else:

            best_duration = 0
            best_temp = np.nan

            simcast_bu = 0

            wallin_hours = 0

            wallin_temp_mean = np.nan
            wallin_temp_min = np.nan
            wallin_temp_max = np.nan

            wallin_sv = 0


        # ----------------------------------------------------
        # Daily summary
        # ----------------------------------------------------

        daily_temp_mean = float(
            np.mean(temp_day)
        )

        daily_temp_min = float(
            np.min(temp_day)
        )

        daily_temp_max = float(
            np.max(temp_day)
        )

        rainfall_mm = float(
            rainfall_daily[day]
        )


        # ----------------------------------------------------
        # Hyre
        # ----------------------------------------------------

        temp_5day_mean, rain_10day_total, hyre_favorable = (
            calculate_hyre_status(
                daily_temp_mean=daily_temp_mean,
                daily_temp_min=daily_temp_min,
                rainfall_mm=rainfall_mm,
                previous_temp_means=previous_temp_means,
                previous_rainfall=previous_rainfall
            )
        )

        previous_temp_means.append(
            daily_temp_mean
        )

        previous_rainfall.append(
            rainfall_mm
        )


        cumulative_simcast_bu += simcast_bu
        cumulative_wallin_sv += wallin_sv


        daily_records.append({

            "day": day + 1,

            "rh_ge90_hours_total":
                rh90_hours_total,

            "max_consecutive_rh90_hours":
                best_duration,

            "simcast_bu":
                simcast_bu,

            "simcast_temp_during_best_spell":
                best_temp,

            "wallin_rh_ge90_hours":
                wallin_hours,

            "wallin_temp_mean":
                wallin_temp_mean,

            "wallin_temp_min":
                wallin_temp_min,

            "wallin_temp_max":
                wallin_temp_max,

            "wallin_sv":
                wallin_sv,

            "temperature_mean":
                daily_temp_mean,

            "temperature_min":
                daily_temp_min,

            "temperature_max":
                daily_temp_max,

            "rainfall_mm":
                rainfall_mm,

            "temp_5day_mean":
                temp_5day_mean,

            "rain_10day_total":
                rain_10day_total,

            "hyre_favorable":
                hyre_favorable,

            "leaf_wetness_hours":
                float(
                    leaf_wetness_daily[day]
                ),

            "soil_moisture":
                float(
                    soil_moisture_daily[day]
                ),

            "candidate_id":
                candidate_id
        })


    daily_df = pd.DataFrame(
        daily_records
    )


    # ========================================================
    # CANDIDATE-LEVEL SUMMARY
    # ========================================================

    candidate_summary = {

        "candidate_id":
            candidate_id,

        "severity_percent":
            float(severity_percent),

        "temperature_mean":
            float(np.mean(temperature)),

        "temperature_min":
            float(np.min(temperature)),

        "temperature_max":
            float(np.max(temperature)),

        "rh_mean":
            float(np.mean(rh)),

        "rh_min":
            float(np.min(rh)),

        "rh_max":
            float(np.max(rh)),

        "rh_ge90_hours":
            int(np.sum(rh >= 90)),

        "leaf_wetness_hours":
            float(
                np.sum(leaf_wetness_daily)
            ),

        "rainfall_total_mm":
            float(
                np.sum(rainfall_daily)
            ),

        "soil_moisture_mean":
            float(
                np.mean(soil_moisture_daily)
            ),

        "simcast_bu":
            float(
                daily_df["simcast_bu"].sum()
            ),

        "wallin_sv":
            float(
                daily_df["wallin_sv"].sum()
            ),

        "hyre_favorable_days":
            int(
                daily_df["hyre_favorable"].sum()
            ),

        "max_daily_wallin_sv":
            int(
                daily_df["wallin_sv"].max()
            ),

        "max_daily_simcast_bu":
            int(
                daily_df["simcast_bu"].max()
            ),

        "latent_risk":
            latent_risk
    }


    hourly_df = pd.DataFrame({

        "hour":
            np.arange(days * 24),

        "day":
            np.repeat(
                np.arange(1, days + 1),
                24
            ),

        "temperature":
            temperature,

        "rh":
            rh,

        "rh_ge90":
            rh >= 90,

        "candidate_id":
            candidate_id,

        "severity_percent":
            severity_percent
    })


    return (
        candidate_summary,
        daily_df,
        hourly_df
    )


# ============================================================
# 8. TEST GENERATION
# ============================================================

def generate_test_candidates(
    severity_percent,
    n_candidates=10,
    base_seed=42
):

    summaries = []
    daily_outputs = {}
    hourly_outputs = {}

    for cid in range(n_candidates):

        (
            summary,
            daily_df,
            hourly_df
        ) = generate_environment_candidate(
            severity_percent=severity_percent,
            candidate_id=cid,
            seed=base_seed + cid,
            days=DAYS
        )

        summaries.append(summary)

        daily_outputs[cid] = daily_df
        hourly_outputs[cid] = hourly_df


    summary_df = pd.DataFrame(
        summaries
    )

    return (
        summary_df,
        daily_outputs,
        hourly_outputs
    )


# ============================================================
# 9. RUN FOR MULTIPLE SEVERITY LEVELS
# ============================================================

severity_levels = [
    0,
    5,
    10,
    20,
    30,
    42,
    50,
    70,
    90
]

all_severity_results = []

for sev in severity_levels:

    summary_df, _, _ = generate_test_candidates(
        severity_percent=sev,
        n_candidates=10,
        base_seed=1000 + int(sev)
    )

    summary_df["severity_group"] = sev

    all_severity_results.append(
        summary_df
    )

severity_test_df = pd.concat(
    all_severity_results,
    ignore_index=True
)


# ============================================================
# 10. DISPLAY
# ============================================================

pd.set_option(
    "display.max_columns",
    None
)

print("=" * 90)
print("SEVERITY CONDITION TEST")
print("=" * 90)

print(
    severity_test_df[
        [
            "severity_percent",
            "candidate_id",
            "temperature_mean",
            "rh_mean",
            "rh_ge90_hours",
            "leaf_wetness_hours",
            "rainfall_total_mm",
            "soil_moisture_mean",
            "simcast_bu",
            "wallin_sv",
            "latent_risk"
        ]
    ]
    .round(3)
    .to_string(index=False)
)


# ============================================================
# 11. AGGREGATED SEVERITY SUMMARY
# ============================================================

severity_summary = (
    severity_test_df
    .groupby("severity_percent")
    .agg(
        latent_risk_mean=("latent_risk", "mean"),
        latent_risk_std=("latent_risk", "std"),

        rh90_mean=("rh_ge90_hours", "mean"),
        rh90_std=("rh_ge90_hours", "std"),

        simcast_bu_mean=("simcast_bu", "mean"),
        simcast_bu_std=("simcast_bu", "std"),

        wallin_sv_mean=("wallin_sv", "mean"),
        wallin_sv_std=("wallin_sv", "std"),

        rainfall_mean=("rainfall_total_mm", "mean"),

        temperature_mean=("temperature_mean", "mean"),
        rh_mean=("rh_mean", "mean")
    )
    .reset_index()
)

print()
print("=" * 90)
print("AGGREGATED SEVERITY SUMMARY")
print("=" * 90)

print(
    severity_summary.round(3).to_string(
        index=False
    )
)


# ============================================================
# 12. SHOW CANDIDATE 0 FOR SEVERITY = 42%
# ============================================================

summary_42, daily_42, hourly_42 = (
    generate_test_candidates(
        severity_percent=42,
        n_candidates=10,
        base_seed=5000
    )
)

print()
print("=" * 90)
print("10 CANDIDATES — SEVERITY = 42%")
print("=" * 90)

print(
    summary_42.round(3).to_string(
        index=False
    )
)

print()
print("=" * 90)
print("DAILY EPIDEMIOLOGICAL OUTPUT — CANDIDATE 0")
print("=" * 90)

print(
    daily_42[0].round(3).to_string(
        index=False
    )
)


# ============================================================
# 13. PLOT — CANDIDATE 0
# ============================================================

h0 = hourly_42[0]

fig, ax1 = plt.subplots(
    figsize=(14, 5)
)

ax1.plot(
    h0["hour"],
    h0["temperature"],
    label="Temperature (°C)"
)

ax1.set_xlabel("Hour")
ax1.set_ylabel("Temperature (°C)")

ax2 = ax1.twinx()

ax2.plot(
    h0["hour"],
    h0["rh"],
    label="RH (%)"
)

ax2.axhline(
    90,
    linestyle="--",
    linewidth=1
)

ax2.set_ylabel("Relative Humidity (%)")

plt.title(
    "10-Day Synthetic Environment — Candidate 0 — Severity 42%"
)

plt.tight_layout()

plt.show()


# ============================================================
# 14. CHECK SEVERITY = 0%
# ============================================================

summary_0, daily_0, hourly_0 = (
    generate_test_candidates(
        severity_percent=0,
        n_candidates=10,
        base_seed=9000
    )
)

print()
print("=" * 90)
print("HEALTHY CASE TEST — SEVERITY = 0%")
print("=" * 90)

print(
    summary_0.round(3).to_string(
        index=False
    )
)
```

```python
# ============================================================
# PROTOTYPE VALIDATION / ACCEPTANCE GATE
# ============================================================

import numpy as np
import pandas as pd


# ------------------------------------------------------------
# 1. Generate MANY candidates per severity
# ------------------------------------------------------------

severity_levels = [
    0, 5, 10, 20, 30, 42, 50, 70, 90
]

N_CANDIDATES = 100

validation_rows = []

for sev in severity_levels:

    summary_df, _, _ = generate_test_candidates(
        severity_percent=sev,
        n_candidates=N_CANDIDATES,
        base_seed=20000 + sev
    )

    validation_rows.append(summary_df)

validation_df = pd.concat(
    validation_rows,
    ignore_index=True
)


# ------------------------------------------------------------
# 2. Physical plausibility checks
# ------------------------------------------------------------

validation_df["temp_ok"] = (
    (validation_df["temperature_min"] >= 3) &
    (validation_df["temperature_max"] <= 27)
)

validation_df["rh_ok"] = (
    (validation_df["rh_min"] >= 50) &
    (validation_df["rh_max"] <= 99.5)
)

validation_df["soil_ok"] = (
    (validation_df["soil_moisture_mean"] >= 10) &
    (validation_df["soil_moisture_mean"] <= 70)
)

validation_df["rain_ok"] = (
    (validation_df["rainfall_total_mm"] >= 0) &
    (validation_df["rainfall_total_mm"] <= 100)
)

validation_df["wetness_ok"] = (
    (validation_df["leaf_wetness_hours"] >= 0) &
    (validation_df["leaf_wetness_hours"] <= 240)
)

validation_df["all_physical_ok"] = (
    validation_df["temp_ok"] &
    validation_df["rh_ok"] &
    validation_df["soil_ok"] &
    validation_df["rain_ok"] &
    validation_df["wetness_ok"]
)


# ------------------------------------------------------------
# 3. Calculate severity-group statistics
# ------------------------------------------------------------

group_stats = (
    validation_df
    .groupby("severity_percent")
    .agg(
        n=("candidate_id", "count"),

        risk_mean=("latent_risk", "mean"),
        risk_std=("latent_risk", "std"),

        rh90_mean=("rh_ge90_hours", "mean"),
        rh90_std=("rh_ge90_hours", "std"),

        simcast_mean=("simcast_bu", "mean"),
        simcast_std=("simcast_bu", "std"),

        wallin_mean=("wallin_sv", "mean"),
        wallin_std=("wallin_sv", "std"),

        temperature_mean=("temperature_mean", "mean"),
        rh_mean=("rh_mean", "mean"),

        rainfall_mean=("rainfall_total_mm", "mean"),
        soil_mean=("soil_moisture_mean", "mean"),

        physical_pass_rate=("all_physical_ok", "mean")
    )
    .reset_index()
)


print("=" * 100)
print("PROTOTYPE VALIDATION SUMMARY")
print("=" * 100)

print(
    group_stats.round(3).to_string(
        index=False
    )
)


# ------------------------------------------------------------
# 4. Monotonicity check
# ------------------------------------------------------------

def monotonic_increasing(values):
    values = np.asarray(values)

    return np.all(
        np.diff(values) >= 0
    )


risk_monotonic = monotonic_increasing(
    group_stats["risk_mean"]
)

rh90_monotonic = monotonic_increasing(
    group_stats["rh90_mean"]
)

simcast_monotonic = monotonic_increasing(
    group_stats["simcast_mean"]
)

wallin_monotonic = monotonic_increasing(
    group_stats["wallin_mean"]
)


print()
print("=" * 100)
print("MONOTONICITY CHECK")
print("=" * 100)

print(
    "Latent risk :", risk_monotonic
)

print(
    "RH>=90 hours:", rh90_monotonic
)

print(
    "SIMCAST BU   :", simcast_monotonic
)

print(
    "Wallin SV    :", wallin_monotonic
)


# ------------------------------------------------------------
# 5. Healthy-case risk inspection
# ------------------------------------------------------------

healthy = validation_df[
    validation_df["severity_percent"] == 0
]

late_severity_70 = validation_df[
    validation_df["severity_percent"] == 70
]

late_severity_90 = validation_df[
    validation_df["severity_percent"] == 90
]


print()
print("=" * 100)
print("HEALTHY CASE")
print("=" * 100)

print(
    healthy[
        [
            "rh_ge90_hours",
            "simcast_bu",
            "wallin_sv",
            "rainfall_total_mm",
            "rh_mean",
            "temperature_mean"
        ]
    ]
    .describe()
    .round(3)
)


print()
print("=" * 100)
print("SEVERITY 70 CASE")
print("=" * 100)

print(
    late_severity_70[
        [
            "rh_ge90_hours",
            "simcast_bu",
            "wallin_sv",
            "rainfall_total_mm",
            "rh_mean",
            "temperature_mean"
        ]
    ]
    .describe()
    .round(3)
)


print()
print("=" * 100)
print("SEVERITY 90 CASE")
print("=" * 100)

print(
    late_severity_90[
        [
            "rh_ge90_hours",
            "simcast_bu",
            "wallin_sv",
            "rainfall_total_mm",
            "rh_mean",
            "temperature_mean"
        ]
    ]
    .describe()
    .round(3)
)


# ------------------------------------------------------------
# 6. Candidate diversity
# ------------------------------------------------------------

print()
print("=" * 100)
print("CANDIDATE DIVERSITY CHECK")
print("=" * 100)

for sev in severity_levels:

    subset = validation_df[
        validation_df["severity_percent"] == sev
    ]

    print(
        f"\nSeverity {sev}%"
    )

    print(
        "BU range:",
        subset["simcast_bu"].min(),
        "to",
        subset["simcast_bu"].max()
    )

    print(
        "Wallin range:",
        subset["wallin_sv"].min(),
        "to",
        subset["wallin_sv"].max()
    )

    print(
        "RH90 range:",
        subset["rh_ge90_hours"].min(),
        "to",
        subset["rh_ge90_hours"].max()
    )

    print(
        "Temp range:",
        round(subset["temperature_mean"].min(), 2),
        "to",
        round(subset["temperature_mean"].max(), 2)
    )


# ------------------------------------------------------------
# 7. Overall physical QC
# ------------------------------------------------------------

physical_pass_rate = (
    validation_df["all_physical_ok"].mean()
)

print()
print("=" * 100)
print("OVERALL PHYSICAL QC")
print("=" * 100)

print(
    f"Physical pass rate: "
    f"{physical_pass_rate * 100:.2f}%"
)


# ------------------------------------------------------------
# 8. Flag suspicious high-risk healthy cases
# ------------------------------------------------------------

healthy_suspicious = healthy[
    (
        (healthy["simcast_bu"] >= 20) |
        (healthy["wallin_sv"] >= 5)
    )
]

print()
print("=" * 100)
print("SUSPICIOUS HEALTHY CASES")
print("=" * 100)

print(
    f"Count: {len(healthy_suspicious)} "
    f"/ {len(healthy)}"
)

if len(healthy_suspicious) > 0:

    print(
        healthy_suspicious[
            [
                "candidate_id",
                "rh_ge90_hours",
                "simcast_bu",
                "wallin_sv",
                "rainfall_total_mm",
                "temperature_mean",
                "rh_mean"
            ]
        ]
        .head(20)
        .round(3)
        .to_string(index=False)
    )
```

```python
# ============================================================
# FINAL POTATO LATE-BLIGHT SYNTHETIC ENVIRONMENT GENERATOR
# ============================================================
#
# SOURCE:
# /content/Potato_Severity_Dataset_FINAL/
#   Potato_Severity_Dataset_FINAL/
#     severity_dataset/
#       potato_severity_FINAL.csv
#
# IMPORTANT:
# - CSV "split" column is authoritative.
# - Filename train/val tokens are NOT used.
# - severity_percent is kept exactly as observed.
# - 10 synthetic trajectories per image.
# - SIMCAST/Fry = primary epidemiological constraint.
# - Wallin SV = secondary cross-check.
# - Environment is synthetic auxiliary data, NOT measured sensor data.
# ============================================================

import numpy as np
import pandas as pd
from pathlib import Path
import json
import hashlib
import warnings

warnings.filterwarnings("ignore")

# ============================================================
# 1. CONFIGURATION
# ============================================================

MANIFEST_PATH = (
    "/content/Potato_Severity_Dataset_FINAL/"
    "Potato_Severity_Dataset_FINAL/"
    "severity_dataset/"
    "potato_severity_FINAL.csv"
)

OUTPUT_DIR = Path(
    "/content/Potato_Synthetic_Environment_FINAL"
)

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

N_CANDIDATES_PER_IMAGE = 10
TRAJECTORY_DAYS = 10
BASE_SEED = 20260913


# ------------------------------------------------------------
# Physical constraints
# ------------------------------------------------------------

TEMP_MIN_ALLOWED = 3.0
TEMP_MAX_ALLOWED = 27.0

RH_MIN_ALLOWED = 50.0
RH_MAX_ALLOWED = 99.5

SOIL_MIN_ALLOWED = 10.0
SOIL_MAX_ALLOWED = 70.0

MAX_10DAY_RAIN = 100.0


# ------------------------------------------------------------
# Healthy-case constraint
#
# Healthy image does not mean zero-risk environment.
# We therefore allow modest environmental risk but reject
# clearly extreme exposure.
# ------------------------------------------------------------

MAX_HEALTHY_SIMCAST_BU = 10
MAX_HEALTHY_WALLIN_SV = 4


# ------------------------------------------------------------
# High-severity constraint
#
# Very high observed severity should not receive an environment
# that is almost completely unfavorable to late blight.
# ------------------------------------------------------------

HIGH_SEVERITY_THRESHOLD = 60.0
HIGH_SEVERITY_MIN_RH90 = 40
HIGH_SEVERITY_MIN_SIMCAST = 5


# ============================================================
# 2. LOAD FINAL DATASET
# ============================================================

df = pd.read_csv(MANIFEST_PATH)

print("=" * 90)
print("LOADED FINAL SEVERITY DATASET")
print("=" * 90)

print("Rows:", len(df))
print("Columns:", len(df.columns))


# ============================================================
# 3. REQUIRED COLUMN CHECK
# ============================================================

required_columns = [
    "split",
    "filename",
    "class_name",
    "severity_percent",
    "final_decision"
]

missing_columns = [
    c for c in required_columns
    if c not in df.columns
]

if missing_columns:
    raise ValueError(
        f"Missing required columns: {missing_columns}"
    )


# ============================================================
# 4. KEEP ONLY FINAL APPROVED ROWS
# ============================================================

print("\nFinal decision counts:")
print(
    df["final_decision"]
    .value_counts(dropna=False)
)

df = df[
    df["final_decision"]
    .astype(str)
    .str.upper()
    .eq("KEEP")
].copy()

print("\nRows after KEEP filter:", len(df))


# ============================================================
# 5. BASIC DATA VALIDATION
# ============================================================

assert df["filename"].notna().all()
assert df["class_name"].notna().all()
assert df["split"].notna().all()
assert df["severity_percent"].notna().all()

assert df["severity_percent"].between(
    0, 100
).all()

assert df["filename"].is_unique


# ============================================================
# 6. AUTHORITATIVE SPLIT VALIDATION
# ============================================================

allowed_splits = {
    "train",
    "valid",
    "test"
}

unexpected_splits = set(
    df["split"].unique()
) - allowed_splits

if unexpected_splits:
    raise ValueError(
        f"Unexpected split values: "
        f"{unexpected_splits}"
    )

print("\nSplit counts:")
print(
    df["split"]
    .value_counts()
)


# ============================================================
# 7. CLASS VALIDATION
# ============================================================

print("\nClass counts:")
print(
    df["class_name"]
    .value_counts()
)


# ============================================================
# 8. SEVERITY VALIDATION
# ============================================================

print("\nSeverity statistics:")
print(
    df["severity_percent"]
    .describe()
    .round(4)
)


# ============================================================
# 9. STABLE REPRODUCIBLE SEED
# ============================================================

def stable_seed(text, base_seed):
    digest = hashlib.sha256(
        str(text).encode("utf-8")
    ).hexdigest()

    value = int(
        digest[:8],
        16
    )

    return (
        base_seed + value
    ) % (2**32)


# ============================================================
# 10. SIMCAST / FRY BU
# ============================================================

def simcast_bu_from_spell(
    temp_c,
    rh90_hours
):
    """
    Prototype implementation of the published Fry/SIMCAST
    temperature-duration table.

    Input:
        temp_c      = mean temperature during qualifying
                       RH >= 90% spell
        rh90_hours  = consecutive RH >= 90 hours
    """

    t = float(temp_c)
    h = int(rh90_hours)

    # >27 C
    if t > 27:
        return 0

    # <3 C
    if t < 3:
        return 0

    # 23–27 C
    if 23 <= t <= 27:

        if h <= 6:
            return 0

        elif h <= 9:
            return 1

        elif h <= 12:
            return 2

        elif h <= 15:
            return 3

        elif h <= 18:
            return 4

        else:
            return 5

    # 13–22 C
    if 13 <= t <= 22:

        if h <= 6:
            return 0

        elif h <= 9:
            return 5

        elif h <= 12:
            return 6

        else:
            return 7

    # 8–12 C
    if 8 <= t <= 12:

        if h <= 6:
            return 0

        elif h == 7:
            return 1

        elif h <= 9:
            return 2

        elif h == 10:
            return 3

        elif h <= 12:
            return 4

        elif h <= 15:
            return 5

        else:
            return 6

    # 3–7 C
    if 3 <= t <= 7:

        if h <= 9:
            return 0

        elif h <= 12:
            return 1

        elif h <= 15:
            return 2

        elif h <= 18:
            return 3

        else:
            return 4

    return 0


# ============================================================
# 11. WALLIN SEVERITY VALUE
# ============================================================

def wallin_sv_from_hours(
    temp_c,
    rh90_hours
):

    t = float(temp_c)
    h = int(rh90_hours)

    # 7.2–11.6 C
    if 7.2 <= t <= 11.6:

        if h <= 15:
            return 0

        elif h <= 18:
            return 1

        elif h <= 21:
            return 2

        elif h <= 24:
            return 3

        else:
            return 4

    # 11.7–15.0 C
    if 11.7 <= t <= 15.0:

        if h <= 12:
            return 0

        elif h <= 15:
            return 1

        elif h <= 18:
            return 2

        elif h <= 21:
            return 3

        else:
            return 4

    # 15.1–26.6 C
    if 15.1 <= t <= 26.6:

        if h <= 9:
            return 0

        elif h <= 12:
            return 1

        elif h <= 15:
            return 2

        elif h <= 18:
            return 3

        else:
            return 4

    # Outside Wallin temperature table
    return 0


# ============================================================
# 12. FIND RH >= 90 SPELLS
# ============================================================

def get_rh90_spells(
    rh,
    temperature
):

    rh = np.asarray(rh)
    temperature = np.asarray(temperature)

    mask = rh >= 90.0

    spells = []

    start = None

    for i, flag in enumerate(mask):

        if flag and start is None:

            start = i

        elif not flag and start is not None:

            end = i

            block_temp = temperature[
                start:end
            ]

            spells.append({

                "start": start,

                "end": end - 1,

                "duration":
                    end - start,

                "mean_temp":
                    float(
                        np.mean(block_temp)
                    ),

                "min_temp":
                    float(
                        np.min(block_temp)
                    ),

                "max_temp":
                    float(
                        np.max(block_temp)
                    )
            })

            start = None

    # Handle spell extending to final hour
    if start is not None:

        end = len(mask)

        block_temp = temperature[
            start:end
        ]

        spells.append({

            "start": start,

            "end": end - 1,

            "duration":
                end - start,

            "mean_temp":
                float(
                    np.mean(block_temp)
                ),

            "min_temp":
                float(
                    np.min(block_temp)
                ),

            "max_temp":
                float(
                    np.max(block_temp)
                )
        })

    return spells


# ============================================================
# 13. SEVERITY -> SOFT ENVIRONMENTAL PRIOR
# ============================================================

def severity_to_risk_prior(
    severity,
    rng
):
    """
    IMPORTANT:

    This is NOT:
        severity -> disease risk equation

    It is only a sampling prior that changes the probability
    of generating more disease-conducive environmental
    trajectories.

    Multiple environments remain possible for the same severity.
    """

    s = np.clip(
        float(severity),
        0,
        100
    ) / 100.0

    base_risk = (
        0.10
        + 0.72 * (s ** 0.72)
    )

    noise = rng.normal(
        0,
        0.07
    )

    return float(
        np.clip(
            base_risk + noise,
            0.03,
            0.97
        )
    )


# ============================================================
# 14. GENERATE 10-DAY HOURLY ENVIRONMENT
# ============================================================

def generate_environment(
    severity,
    seed
):

    rng = np.random.default_rng(seed)

    latent_risk = severity_to_risk_prior(
        severity,
        rng
    )

    total_hours = (
        TRAJECTORY_DAYS * 24
    )


    # --------------------------------------------------------
    # Temperature
    # --------------------------------------------------------

    daily_temp_mean = (
        14.0
        + 6.0 * latent_risk
        + rng.normal(
            0,
            1.15,
            TRAJECTORY_DAYS
        )
    )

    temperature = np.zeros(
        total_hours
    )

    for day in range(
        TRAJECTORY_DAYS
    ):

        start = day * 24
        end = start + 24

        h = np.arange(24)

        values = (
            daily_temp_mean[day]

            + 4.0 * np.sin(
                2 * np.pi * (h - 8) / 24
            )

            + rng.normal(
                0,
                0.7,
                24
            )
        )

        temperature[
            start:end
        ] = values

    temperature = np.clip(
        temperature,
        TEMP_MIN_ALLOWED,
        TEMP_MAX_ALLOWED
    )


    # --------------------------------------------------------
    # Relative humidity
    # --------------------------------------------------------

    daily_rh_center = (
        77.0
        + 16.0 * latent_risk
        + rng.normal(
            0,
            2.0,
            TRAJECTORY_DAYS
        )
    )

    rh = np.zeros(
        total_hours
    )

    for day in range(
        TRAJECTORY_DAYS
    ):

        start = day * 24
        end = start + 24

        h = np.arange(24)

        diurnal = (
            8.0
            * np.cos(
                2 * np.pi * (h - 4) / 24
            )
        )

        values = (
            daily_rh_center[day]
            + diurnal
            + rng.normal(
                0,
                3.0,
                24
            )
        )

        rh[
            start:end
        ] = values

    rh = np.clip(
        rh,
        RH_MIN_ALLOWED,
        RH_MAX_ALLOWED
    )


    # --------------------------------------------------------
    # Rainfall
    # --------------------------------------------------------

    rain_probability = (
        0.04
        + 0.35 * latent_risk
    )

    rainfall = []

    for day in range(
        TRAJECTORY_DAYS
    ):

        if rng.random() < rain_probability:

            rain = rng.gamma(
                shape=1.8,
                scale=4.0
            )

        else:

            rain = 0.0

        rainfall.append(
            float(rain)
        )


    # --------------------------------------------------------
    # Leaf wetness
    # --------------------------------------------------------

    leaf_wetness = []

    for day in range(
        TRAJECTORY_DAYS
    ):

        start = day * 24
        end = start + 24

        rh_day = rh[
            start:end
        ]

        rh_wet_hours = int(
            np.sum(
                rh_day >= 90
            )
        )

        rain_extension = min(
            6.0,
            rainfall[day] * 0.7
        )

        wetness = (
            rh_wet_hours
            + rain_extension
            + rng.normal(
                0,
                1.5
            )
        )

        wetness = np.clip(
            wetness,
            0,
            24
        )

        leaf_wetness.append(
            float(wetness)
        )


    # --------------------------------------------------------
    # Soil moisture
    # --------------------------------------------------------

    soil_moisture = []

    current_soil = (
        25.0
        + 20.0 * latent_risk
        + rng.normal(
            0,
            3.0
        )
    )

    for day in range(
        TRAJECTORY_DAYS
    ):

        rainfall_effect = (
            rainfall[day] * 0.9
        )

        drying = (
            0.20
            * max(
                0,
                daily_temp_mean[day] - 15
            )
        )

        current_soil = (
            current_soil
            + rainfall_effect
            - drying
            + rng.normal(
                0,
                1.5
            )
        )

        current_soil = np.clip(
            current_soil,
            SOIL_MIN_ALLOWED,
            SOIL_MAX_ALLOWED
        )

        soil_moisture.append(
            float(current_soil)
        )


    return {

        "temperature":
            temperature,

        "rh":
            rh,

        "rainfall":
            rainfall,

        "leaf_wetness":
            leaf_wetness,

        "soil_moisture":
            soil_moisture,

        "latent_risk":
            latent_risk
    }


# ============================================================
# 15. EPIDEMIOLOGICAL ANALYSIS
# ============================================================

def analyse_environment(
    env,
    severity,
    image_id,
    candidate_id,
    split,
    class_name,
    filename
):

    temperature = env["temperature"]
    rh = env["rh"]

    rainfall = env["rainfall"]
    leaf_wetness = env["leaf_wetness"]
    soil = env["soil_moisture"]

    daily_rows = []

    cumulative_simcast_bu = 0
    cumulative_wallin_sv = 0


    for day in range(
        TRAJECTORY_DAYS
    ):

        start = day * 24
        end = start + 24

        t = temperature[
            start:end
        ]

        r = rh[
            start:end
        ]


        # ----------------------------------------------------
        # RH >= 90 spells
        # ----------------------------------------------------

        spells = get_rh90_spells(
            r,
            t
        )

        rh90_total = int(
            np.sum(
                r >= 90
            )
        )


        if spells:

            # Prototype convention:
            # strongest/longest daily qualifying spell
            # is used for daily SIMCAST assignment.

            best_spell = max(
                spells,
                key=lambda x: x["duration"]
            )

            best_duration = int(
                best_spell["duration"]
            )

            best_temp = float(
                best_spell["mean_temp"]
            )

            simcast_bu = simcast_bu_from_spell(
                best_temp,
                best_duration
            )


            # ------------------------------------------------
            # Wallin
            # ------------------------------------------------

            wet_idx = np.where(
                r >= 90
            )[0]

            if len(wet_idx) > 0:

                wallin_temp = float(
                    np.mean(
                        t[wet_idx]
                    )
                )

                wallin_temp_min = float(
                    np.min(
                        t[wet_idx]
                    )
                )

                wallin_temp_max = float(
                    np.max(
                        t[wet_idx]
                    )
                )

            else:

                wallin_temp = np.nan
                wallin_temp_min = np.nan
                wallin_temp_max = np.nan

            wallin_sv = wallin_sv_from_hours(
                wallin_temp
                if not np.isnan(wallin_temp)
                else 0,
                rh90_total
            )

        else:

            best_duration = 0
            best_temp = np.nan

            simcast_bu = 0

            wallin_temp = np.nan
            wallin_temp_min = np.nan
            wallin_temp_max = np.nan

            wallin_sv = 0


        cumulative_simcast_bu += (
            simcast_bu
        )

        cumulative_wallin_sv += (
            wallin_sv
        )


        daily_rows.append({

            "filename":
                filename,

            "image_id":
                image_id,

            "split":
                split,

            "class_name":
                class_name,

            "candidate_id":
                candidate_id,

            "day":
                day + 1,

            "severity_percent":
                severity,

            "temperature_mean":
                float(
                    np.mean(t)
                ),

            "temperature_min":
                float(
                    np.min(t)
                ),

            "temperature_max":
                float(
                    np.max(t)
                ),

            "rh_mean":
                float(
                    np.mean(r)
                ),

            "rh_min":
                float(
                    np.min(r)
                ),

            "rh_max":
                float(
                    np.max(r)
                ),

            "rh_ge90_hours":
                rh90_total,

            "max_consecutive_rh90_hours":
                best_duration,

            "simcast_bu":
                simcast_bu,

            "wallin_temp_mean":
                wallin_temp,

            "wallin_temp_min":
                wallin_temp_min,

            "wallin_temp_max":
                wallin_temp_max,

            "wallin_sv":
                wallin_sv,

            "rainfall_mm":
                rainfall[day],

            "leaf_wetness_hours":
                leaf_wetness[day],

            "soil_moisture":
                soil[day]
        })


    daily_df = pd.DataFrame(
        daily_rows
    )


    # ========================================================
    # IMAGE-LEVEL SUMMARY
    # ========================================================

    summary = {

        "filename":
            filename,

        "image_id":
            image_id,

        "split":
            split,

        "class_name":
            class_name,

        "candidate_id":
            candidate_id,

        "severity_percent":
            severity,

        "temperature_mean":
            float(
                np.mean(
                    temperature
                )
            ),

        "temperature_min":
            float(
                np.min(
                    temperature
                )
            ),

        "temperature_max":
            float(
                np.max(
                    temperature
                )
            ),

        "rh_mean":
            float(
                np.mean(
                    rh
                )
            ),

        "rh_min":
            float(
                np.min(
                    rh
                )
            ),

        "rh_max":
            float(
                np.max(
                    rh
                )
            ),

        "rh_ge90_hours":
            int(
                np.sum(
                    rh >= 90
                )
            ),

        "leaf_wetness_hours":
            float(
                np.sum(
                    leaf_wetness
                )
            ),

        "rainfall_total_mm":
            float(
                np.sum(
                    rainfall
                )
            ),

        "soil_moisture_mean":
            float(
                np.mean(
                    soil
                )
            ),

        "simcast_bu":
            float(
                cumulative_simcast_bu
            ),

        "wallin_sv":
            float(
                cumulative_wallin_sv
            ),

        "max_daily_simcast_bu":
            int(
                daily_df[
                    "simcast_bu"
                ].max()
            ),

        "max_daily_wallin_sv":
            int(
                daily_df[
                    "wallin_sv"
                ].max()
            ),

        "latent_risk":
            env["latent_risk"]
    }


    # ========================================================
    # HOURLY TABLE
    # ========================================================

    hourly_df = pd.DataFrame({

        "filename":
            filename,

        "image_id":
            image_id,

        "split":
            split,

        "class_name":
            class_name,

        "candidate_id":
            candidate_id,

        "hour":
            np.arange(
                TRAJECTORY_DAYS * 24
            ),

        "day":
            np.repeat(
                np.arange(
                    1,
                    TRAJECTORY_DAYS + 1
                ),
                24
            ),

        "temperature":
            temperature,

        "rh":
            rh,

        "rh_ge90":
            rh >= 90,

        "severity_percent":
            severity
    })


    return (
        summary,
        daily_df,
        hourly_df
    )


# ============================================================
# 16. ACCEPTANCE / REJECTION RULE
# ============================================================

def accept_candidate(row):

    severity = float(
        row["severity_percent"]
    )


    # --------------------------------------------------------
    # Physical temperature
    # --------------------------------------------------------

    if not (
        TEMP_MIN_ALLOWED
        <= row["temperature_min"]
        <= row["temperature_max"]
        <= TEMP_MAX_ALLOWED
    ):

        return False, "temperature"


    # --------------------------------------------------------
    # Physical RH
    # --------------------------------------------------------

    if not (
        RH_MIN_ALLOWED
        <= row["rh_min"]
        <= row["rh_max"]
        <= RH_MAX_ALLOWED
    ):

        return False, "rh"


    # --------------------------------------------------------
    # Soil
    # --------------------------------------------------------

    if not (
        SOIL_MIN_ALLOWED
        <= row["soil_moisture_mean"]
        <= SOIL_MAX_ALLOWED
    ):

        return False, "soil_moisture"


    # --------------------------------------------------------
    # Rainfall
    # --------------------------------------------------------

    if not (
        0
        <= row["rainfall_total_mm"]
        <= MAX_10DAY_RAIN
    ):

        return False, "rainfall"


    # --------------------------------------------------------
    # Healthy image
    # --------------------------------------------------------

    if severity == 0:

        if (
            row["simcast_bu"]
            > MAX_HEALTHY_SIMCAST_BU
        ):

            return False, "healthy_high_simcast"

        if (
            row["wallin_sv"]
            > MAX_HEALTHY_WALLIN_SV
        ):

            return False, "healthy_high_wallin"


    # --------------------------------------------------------
    # High severity
    # --------------------------------------------------------

    if severity >= HIGH_SEVERITY_THRESHOLD:

        if (
            row["rh_ge90_hours"]
            < HIGH_SEVERITY_MIN_RH90
            and
            row["simcast_bu"]
            < HIGH_SEVERITY_MIN_SIMCAST
        ):

            return False, "high_severity_low_risk"


    return True, "accepted"


# ============================================================
# 17. GENERATE FINAL DATASET
# ============================================================

summary_rows = []
daily_tables = []
hourly_tables = []
rejection_rows = []


print()
print("=" * 90)
print("STARTING FULL DATASET GENERATION")
print("=" * 90)

print(
    "Original images:",
    len(df)
)

print(
    "Candidates per image:",
    N_CANDIDATES_PER_IMAGE
)

print(
    "Expected final trajectories:",
    len(df)
    * N_CANDIDATES_PER_IMAGE
)

print()


for idx, source_row in df.iterrows():

    filename = source_row["filename"]

    severity = float(
        source_row["severity_percent"]
    )

    split = source_row["split"]

    class_name = source_row["class_name"]

    accepted_count = 0
    attempt = 0

    max_attempts = (
        N_CANDIDATES_PER_IMAGE * 5
    )


    while (
        accepted_count
        < N_CANDIDATES_PER_IMAGE
        and
        attempt
        < max_attempts
    ):

        candidate_id = (
            accepted_count
        )

        seed = stable_seed(
            f"{filename}|"
            f"{candidate_id}|"
            f"{attempt}",
            BASE_SEED
        )


        # ----------------------------------------------------
        # Generate environment
        # ----------------------------------------------------

        env = generate_environment(
            severity=severity,
            seed=seed
        )


        # ----------------------------------------------------
        # Analyse epidemiology
        # ----------------------------------------------------

        (
            summary,
            daily_df,
            hourly_df
        ) = analyse_environment(

            env=env,

            severity=severity,

            image_id=filename,

            candidate_id=candidate_id,

            split=split,

            class_name=class_name,

            filename=filename
        )


        # ----------------------------------------------------
        # Accept / reject
        # ----------------------------------------------------

        accepted, reason = (
            accept_candidate(
                summary
            )
        )


        if accepted:

            summary["generation_method"] = (
                "Approach_B_SIMCAST_Fry_plus_Wallin"
            )

            summary["random_seed"] = seed

            summary["qc_status"] = "OK"

            summary_rows.append(
                summary
            )

            daily_tables.append(
                daily_df
            )

            hourly_tables.append(
                hourly_df
            )

            accepted_count += 1


        else:

            rejection_rows.append({

                "filename":
                    filename,

                "split":
                    split,

                "class_name":
                    class_name,

                "severity_percent":
                    severity,

                "attempt":
                    attempt,

                "candidate_id":
                    candidate_id,

                "random_seed":
                    seed,

                "rejection_reason":
                    reason
            })


        attempt += 1


    # --------------------------------------------------------
    # If not enough accepted trajectories
    # --------------------------------------------------------

    if accepted_count < N_CANDIDATES_PER_IMAGE:

        raise RuntimeError(

            f"Could not generate "
            f"{N_CANDIDATES_PER_IMAGE} "
            f"accepted candidates for:\n"
            f"{filename}\n"
            f"Accepted: {accepted_count}\n"
            f"Attempts: {attempt}"
        )


    # Progress
    if (
        (idx + 1) % 100 == 0
        or
        (idx + 1) == len(df)
    ):

        print(
            f"Processed "
            f"{idx + 1}/{len(df)} images"
        )


# ============================================================
# 18. BUILD FINAL DATAFRAMES
# ============================================================

summary_final = pd.DataFrame(
    summary_rows
)

daily_final = pd.concat(
    daily_tables,
    ignore_index=True
)

hourly_final = pd.concat(
    hourly_tables,
    ignore_index=True
)

rejected_final = pd.DataFrame(
    rejection_rows
    if rejection_rows
    else [],
    columns=[
        "filename",
        "split",
        "class_name",
        "severity_percent",
        "attempt",
        "candidate_id",
        "random_seed",
        "rejection_reason"
    ]
)


# ============================================================
# 19. FINAL QC — ROW COUNT
# ============================================================

expected_rows = (
    len(df)
    * N_CANDIDATES_PER_IMAGE
)

actual_rows = len(
    summary_final
)

print()
print("=" * 90)
print("FINAL ROW COUNT")
print("=" * 90)

print(
    "Expected:",
    expected_rows
)

print(
    "Actual:",
    actual_rows
)

assert actual_rows == expected_rows


# ============================================================
# 20. FINAL QC — CANDIDATES PER IMAGE
# ============================================================

candidate_counts = (
    summary_final
    .groupby("filename")
    .size()
)

wrong_candidate_count = (
    candidate_counts
    != N_CANDIDATES_PER_IMAGE
)

print()
print(
    "Images with incorrect candidate count:",
    int(wrong_candidate_count.sum())
)

assert wrong_candidate_count.sum() == 0


# ============================================================
# 21. FINAL QC — SPLIT INTEGRITY
# ============================================================

split_per_image = (
    summary_final
    .groupby("filename")["split"]
    .nunique()
)

images_spanning_splits = (
    split_per_image > 1
).sum()

print(
    "Images spanning multiple splits:",
    int(images_spanning_splits)
)

assert images_spanning_splits == 0


# ============================================================
# 22. FINAL QC — ORIGINAL SPLIT COUNTS
# ============================================================

print()
print("Original image split:")
print(
    df["split"].value_counts()
)

print()
print(
    "Final trajectory split:"
)

print(
    summary_final[
        ["filename", "split"]
    ]
    .drop_duplicates()["split"]
    .value_counts()
)


# ============================================================
# 23. FINAL QC — SEVERITY PRESERVATION
# ============================================================

original_severity = (
    df.set_index(
        "filename"
    )["severity_percent"]
)

generated_severity = (
    summary_final
    .drop_duplicates("filename")
    .set_index(
        "filename"
    )["severity_percent"]
)

common = (
    original_severity.index
    .intersection(
        generated_severity.index
    )
)

severity_difference = (
    original_severity.loc[common]
    -
    generated_severity.loc[common]
)

max_severity_difference = (
    np.abs(
        severity_difference
    ).max()
)

print()
print(
    "Maximum severity mismatch:",
    max_severity_difference
)

assert (
    max_severity_difference
    < 1e-10
)


# ============================================================
# 24. HEALTHY SEVERITY CHECK
# ============================================================

healthy_generated = summary_final[
    summary_final["class_name"]
    .eq("Potato Healthy")
]

healthy_nonzero = (
    np.abs(
        healthy_generated[
            "severity_percent"
        ]
    ) > 1e-10
).sum()

print(
    "Healthy trajectories with nonzero severity:",
    int(healthy_nonzero)
)

assert healthy_nonzero == 0


# ============================================================
# 25. LATE BLIGHT SEVERITY CHECK
# ============================================================

late_generated = summary_final[
    summary_final["class_name"]
    .eq("Potato Late Blight")
]

print()
print(
    "Late Blight trajectories:",
    len(late_generated)
)

print(
    "Late Blight severity range:",
    late_generated[
        "severity_percent"
    ].min(),
    "to",
    late_generated[
        "severity_percent"
    ].max()
)


# ============================================================
# 26. DUPLICATE TRAJECTORY CHECK
# ============================================================

duplicate_trajectory_rows = (
    summary_final
    .duplicated(
        subset=[
            "filename",
            "candidate_id"
        ]
    )
    .sum()
)

print(
    "Duplicate trajectory IDs:",
    int(duplicate_trajectory_rows)
)

assert duplicate_trajectory_rows == 0


# ============================================================
# 27. MISSING VALUE CHECK
# ============================================================

summary_missing = (
    summary_final.isna()
    .sum()
    .sum()
)

daily_missing = (
    daily_final.isna()
    .sum()
    .sum()
)

hourly_missing = (
    hourly_final.isna()
    .sum()
    .sum()
)

print()
print(
    "Summary missing cells:",
    int(summary_missing)
)

print(
    "Daily missing cells:",
    int(daily_missing)
)

print(
    "Hourly missing cells:",
    int(hourly_missing)
)

# Hourly table should have no missing values.
assert hourly_missing == 0


# ============================================================
# 28. PHYSICAL RANGE QC
# ============================================================

physical_temperature_ok = (
    (summary_final["temperature_min"]
     >= TEMP_MIN_ALLOWED)
    &
    (summary_final["temperature_max"]
     <= TEMP_MAX_ALLOWED)
)

physical_rh_ok = (
    (summary_final["rh_min"]
     >= RH_MIN_ALLOWED)
    &
    (summary_final["rh_max"]
     <= RH_MAX_ALLOWED)
)

physical_soil_ok = (
    (summary_final["soil_moisture_mean"]
     >= SOIL_MIN_ALLOWED)
    &
    (summary_final["soil_moisture_mean"]
     <= SOIL_MAX_ALLOWED)
)

physical_rain_ok = (
    (summary_final["rainfall_total_mm"]
     >= 0)
    &
    (summary_final["rainfall_total_mm"]
     <= MAX_10DAY_RAIN)
)

all_physical_ok = (
    physical_temperature_ok
    &
    physical_rh_ok
    &
    physical_soil_ok
    &
    physical_rain_ok
)

print()
print(
    "Physical QC pass rate:",
    round(
        100 *
        all_physical_ok.mean(),
        3
    ),
    "%"
)

assert all_physical_ok.all()


# ============================================================
# 29. SAVE FILES
# ============================================================

SUMMARY_PATH = (
    OUTPUT_DIR
    / "synthetic_environment_image_level.csv"
)

DAILY_PATH = (
    OUTPUT_DIR
    / "synthetic_environment_daily.csv"
)

HOURLY_PATH = (
    OUTPUT_DIR
    / "synthetic_environment_hourly.csv"
)

REJECT_PATH = (
    OUTPUT_DIR
    / "synthetic_environment_rejected_candidates.csv"
)

CONFIG_PATH = (
    OUTPUT_DIR
    / "generation_config.json"
)


summary_final.to_csv(
    SUMMARY_PATH,
    index=False
)

daily_final.to_csv(
    DAILY_PATH,
    index=False
)

hourly_final.to_csv(
    HOURLY_PATH,
    index=False
)

rejected_final.to_csv(
    REJECT_PATH,
    index=False
)


# ============================================================
# 30. SAVE GENERATION CONFIG
# ============================================================

config = {

    "source_file":
        MANIFEST_PATH,

    "method":
        "Approach B — Epidemiology-Constrained Synthetic Environment",

    "primary_model":
        "SIMCAST/Fry",

    "secondary_model":
        "Wallin Severity Value",

    "trajectory_days":
        TRAJECTORY_DAYS,

    "candidates_per_image":
        N_CANDIDATES_PER_IMAGE,

    "base_seed":
        BASE_SEED,

    "source_images":
        int(len(df)),

    "train_images":
        int(
            (df["split"] == "train").sum()
        ),

    "valid_images":
        int(
            (df["split"] == "valid").sum()
        ),

    "test_images":
        int(
            (df["split"] == "test").sum()
        ),

    "healthy_images":
        int(
            (
                df["class_name"]
                == "Potato Healthy"
            ).sum()
        ),

    "late_blight_images":
        int(
            (
                df["class_name"]
                == "Potato Late Blight"
            ).sum()
        ),

    "final_trajectory_rows":
        int(len(summary_final)),

    "rejected_candidate_attempts":
        int(len(rejected_final)),

    "severity_is_ground_truth":
        False,

    "environment_is_synthetic":
        True,

    "split_column_authoritative":
        True
}

with open(
    CONFIG_PATH,
    "w"
) as f:

    json.dump(
        config,
        f,
        indent=2
    )


# ============================================================
# 31. FINAL REPORT
# ============================================================

print()
print("=" * 90)
print("FINAL SYNTHETIC DATASET GENERATION COMPLETE")
print("=" * 90)

print(
    "Original images          :",
    len(df)
)

print(
    "Train images             :",
    (df["split"] == "train").sum()
)

print(
    "Valid images             :",
    (df["split"] == "valid").sum()
)

print(
    "Test images              :",
    (df["split"] == "test").sum()
)

print(
    "Healthy images           :",
    (
        df["class_name"]
        == "Potato Healthy"
    ).sum()
)

print(
    "Late Blight images       :",
    (
        df["class_name"]
        == "Potato Late Blight"
    ).sum()
)

print(
    "Candidates per image     :",
    N_CANDIDATES_PER_IMAGE
)

print(
    "Expected trajectories    :",
    expected_rows
)

print(
    "Generated trajectories   :",
    len(summary_final)
)

print(
    "Rejected attempts        :",
    len(rejected_final)
)

print()
print("OUTPUT FILES")
print("------------------------")

print(
    "Image-level:",
    SUMMARY_PATH
)

print(
    "Daily:",
    DAILY_PATH
)

print(
    "Hourly:",
    HOURLY_PATH
)

print(
    "Rejected:",
    REJECT_PATH
)

print(
    "Config:",
    CONFIG_PATH
)

print()
print("ALL FINAL QC CHECKS PASSED.")
```

```python
# ============================================================
# POTATO LATE-BLIGHT — FINAL SYNTHETIC ENVIRONMENT AUDIT
# ============================================================

import os
import json
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

warnings.filterwarnings("ignore")

# ------------------------------------------------------------
# 0. PATHS
# ------------------------------------------------------------

BASE_DIR = "/content/Potato_Synthetic_Environment_FINAL"

IMG_PATH = os.path.join(
    BASE_DIR, "synthetic_environment_image_level.csv"
)

DAILY_PATH = os.path.join(
    BASE_DIR, "synthetic_environment_daily.csv"
)

HOURLY_PATH = os.path.join(
    BASE_DIR, "synthetic_environment_hourly.csv"
)

CONFIG_PATH = os.path.join(
    BASE_DIR, "generation_config.json"
)

print("=" * 80)
print("POTATO SYNTHETIC ENVIRONMENT — FINAL AUDIT")
print("=" * 80)

for path in [IMG_PATH, DAILY_PATH, HOURLY_PATH, CONFIG_PATH]:
    print(f"{os.path.basename(path):45s} -> {'FOUND' if os.path.exists(path) else 'MISSING'}")

if not os.path.exists(IMG_PATH):
    raise FileNotFoundError(f"Missing: {IMG_PATH}")

if not os.path.exists(DAILY_PATH):
    raise FileNotFoundError(f"Missing: {DAILY_PATH}")


# ------------------------------------------------------------
# 1. LOAD DATA
# ------------------------------------------------------------

image_df = pd.read_csv(IMG_PATH)
daily_df = pd.read_csv(DAILY_PATH)

hourly_df = None
if os.path.exists(HOURLY_PATH):
    hourly_df = pd.read_csv(HOURLY_PATH)

print("\nLoaded:")
print("Image-level:", image_df.shape)
print("Daily:", daily_df.shape)
if hourly_df is not None:
    print("Hourly:", hourly_df.shape)


# ------------------------------------------------------------
# 2. BASIC STRUCTURE AUDIT
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("2. BASIC STRUCTURE AUDIT")
print("=" * 80)

print("\nImage-level columns:")
print(image_df.columns.tolist())

print("\nDaily columns:")
print(daily_df.columns.tolist())

print("\nFirst 3 image-level rows:")
display(image_df.head(3))

print("\nFirst 3 daily rows:")
display(daily_df.head(3))


# ------------------------------------------------------------
# 3. IMAGE-LEVEL INTEGRITY
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("3. IMAGE-LEVEL INTEGRITY")
print("=" * 80)

required_img_cols = [
    "split",
    "filename",
    "class_name",
    "severity_percent",
]

missing = [c for c in required_img_cols if c not in image_df.columns]

if missing:
    raise ValueError(f"Missing image-level columns: {missing}")

print("\nRows:", len(image_df))

print("\nSplit counts:")
print(image_df["split"].value_counts(dropna=False))

print("\nClass counts:")
print(image_df["class_name"].value_counts(dropna=False))

print("\nSeverity summary:")
print(image_df["severity_percent"].describe())

print("\nMissing values in key columns:")
print(image_df[required_img_cols].isna().sum())

print("\nDuplicate rows:", image_df.duplicated().sum())
print("Duplicate filenames:", image_df["filename"].duplicated().sum())


# ------------------------------------------------------------
# 4. 10 TRAJECTORIES / IMAGE CHECK
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("4. TRAJECTORY COUNT PER IMAGE")
print("=" * 80)

traj_count = (
    image_df
    .groupby("filename")
    .size()
    .value_counts()
    .sort_index()
)

print(traj_count)

bad_traj = (
    image_df
    .groupby("filename")
    .size()
)

bad_traj = bad_traj[bad_traj != 10]

print("\nImages not having exactly 10 trajectories:", len(bad_traj))

if len(bad_traj):
    print(bad_traj.head(20))


# ------------------------------------------------------------
# 5. SEVERITY PRESERVATION
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("5. SEVERITY PRESERVATION ACROSS TRAJECTORIES")
print("=" * 80)

severity_unique = (
    image_df
    .groupby("filename")["severity_percent"]
    .nunique()
)

bad_severity = severity_unique[severity_unique != 1]

print("Images with inconsistent severity across trajectories:",
      len(bad_severity))

if len(bad_severity):
    print(bad_severity.head(20))


# ------------------------------------------------------------
# 6. HEALTHY / LATE-BLIGHT CHECK
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("6. CLASS ↔ SEVERITY CONSISTENCY")
print("=" * 80)

healthy_nonzero = image_df[
    (image_df["class_name"].str.lower() == "healthy") &
    (image_df["severity_percent"] > 0)
]

lb_zero = image_df[
    (image_df["class_name"].str.lower() != "healthy") &
    (image_df["severity_percent"] <= 0)
]

print("Healthy with severity > 0:", len(healthy_nonzero))
print("Late-blight with severity <= 0:", len(lb_zero))

if len(healthy_nonzero):
    display(healthy_nonzero.head())

if len(lb_zero):
    display(lb_zero.head())


# ------------------------------------------------------------
# 7. DAILY NaN AUDIT
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("7. DAILY NaN AUDIT")
print("=" * 80)

print("\nMissing values by column:")
nan_counts = daily_df.isna().sum()
print(nan_counts[nan_counts > 0].sort_values(ascending=False))

wallin_cols = [
    "wallin_temp_mean",
    "wallin_temp_min",
    "wallin_temp_max"
]

for col in wallin_cols:
    if col not in daily_df.columns:
        print(f"WARNING: {col} not found")


# ------------------------------------------------------------
# 8. CHECK WHETHER WALLIN NaNs OCCUR ONLY WHEN RH90 = 0
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("8. WALLIN NaN ↔ RH≥90 EVENT CHECK")
print("=" * 80)

if "rh_ge90_hours" in daily_df.columns:

    daily_df["rh90_event"] = (
        daily_df["rh_ge90_hours"] > 0
    ).astype(int)

    daily_df["wallin_temp_defined"] = (
        daily_df["rh90_event"] == 1
    ).astype(int)

    for col in wallin_cols:

        if col not in daily_df.columns:
            continue

        nan_mask = daily_df[col].isna()

        print(f"\n{col}")
        print("NaN rows:", nan_mask.sum())

        print("NaN rows with RH90 event:")
        print(
            daily_df.loc[nan_mask, "rh90_event"]
            .value_counts(dropna=False)
        )

        print("Non-NaN rows without RH90 event:")
        non_nan_no_event = (
            (~nan_mask) &
            (daily_df["rh90_event"] == 0)
        ).sum()

        print(non_nan_no_event)


# ------------------------------------------------------------
# 9. SAVE CLEANED DAILY DATASET
# ------------------------------------------------------------

DAILY_V2_PATH = os.path.join(
    BASE_DIR,
    "synthetic_environment_daily_v2.csv"
)

daily_df.to_csv(DAILY_V2_PATH, index=False)

print("\nSaved cleaned daily table:")
print(DAILY_V2_PATH)


# ------------------------------------------------------------
# 10. IMAGE-LEVEL FEATURE CORRELATION
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("10. SEVERITY ↔ ENVIRONMENT CORRELATIONS")
print("=" * 80)

env_cols = [
    "temperature_mean",
    "temperature_min",
    "temperature_max",
    "rh_mean",
    "rh_min",
    "rh_max",
    "rh_ge90_hours",
    "leaf_wetness_hours",
    "rainfall_total_mm",
    "soil_moisture_mean",
    "simcast_bu",
    "wallin_sv",
]

available_env = [
    c for c in env_cols
    if c in image_df.columns
]

corr_rows = []

for col in available_env:

    x = image_df[[col, "severity_percent"]].dropna()

    if len(x) < 3:
        continue

    pearson = x[col].corr(
        x["severity_percent"],
        method="pearson"
    )

    spearman = x[col].corr(
        x["severity_percent"],
        method="spearman"
    )

    corr_rows.append({
        "feature": col,
        "pearson_r": pearson,
        "spearman_r": spearman,
        "n": len(x)
    })

corr_df = (
    pd.DataFrame(corr_rows)
    .sort_values("spearman_r", ascending=False)
)

display(corr_df)


# ------------------------------------------------------------
# 11. HEALTHY VS LATE-BLIGHT ENVIRONMENT
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("11. HEALTHY VS LATE-BLIGHT ENVIRONMENT")
print("=" * 80)

group_col = "class_name"

group_summary = (
    image_df
    .groupby(group_col)[available_env]
    .agg(["mean", "std", "median"])
)

display(group_summary)


# ------------------------------------------------------------
# 12. TRAIN / VALID / TEST DISTRIBUTION
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("12. TRAIN / VALID / TEST CONSISTENCY")
print("=" * 80)

split_summary = (
    image_df
    .groupby("split")
    .agg(
        n=("filename", "count"),
        unique_images=("filename", "nunique"),
        severity_mean=("severity_percent", "mean"),
        severity_median=("severity_percent", "median"),
        severity_std=("severity_percent", "std"),
    )
)

display(split_summary)

split_env_summary = (
    image_df
    .groupby("split")[available_env]
    .mean()
)

display(split_env_summary)


# ------------------------------------------------------------
# 13. SPLIT OVERLAP / LEAKAGE AUDIT
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("13. SPLIT LEAKAGE AUDIT")
print("=" * 80)

split_counts_per_image = (
    image_df
    .groupby("filename")["split"]
    .nunique()
)

multi_split = split_counts_per_image[
    split_counts_per_image > 1
]

print("Images appearing in multiple splits:",
      len(multi_split))

if len(multi_split):
    print(multi_split.head(20))


# ============================================================
# REPLACEMENT: SECTIONS 14–21
# ============================================================

print("\n" + "=" * 80)
print("14. HEALTHY vs LATE-BLIGHT ENVIRONMENT OVERLAP")
print("=" * 80)

# Explicit class masks
healthy_mask = image_df["class_name"].str.contains(
    "Healthy", case=False, na=False
)

late_mask = image_df["class_name"].str.contains(
    "Late Blight", case=False, na=False
)

print("Healthy trajectories:", healthy_mask.sum())
print("Late-blight trajectories:", late_mask.sum())

overlap_rows = []

for col in available_env:

    healthy = image_df.loc[healthy_mask, col].dropna()
    late = image_df.loc[late_mask, col].dropna()

    if len(healthy) < 2 or len(late) < 2:
        continue

    pooled_sd = np.sqrt(
        (
            (len(healthy) - 1) * healthy.var(ddof=1)
            +
            (len(late) - 1) * late.var(ddof=1)
        )
        /
        (len(healthy) + len(late) - 2)
    )

    if pooled_sd > 0:
        effect = abs(
            healthy.mean() - late.mean()
        ) / pooled_sd
    else:
        effect = np.nan

    overlap_rows.append({
        "feature": col,
        "healthy_mean": healthy.mean(),
        "late_blight_mean": late.mean(),
        "healthy_median": healthy.median(),
        "late_blight_median": late.median(),
        "healthy_std": healthy.std(),
        "late_blight_std": late.std(),
        "standardized_mean_difference": effect,
    })

overlap_df = pd.DataFrame(overlap_rows)

if overlap_df.empty:
    print("ERROR: overlap table is empty.")
else:
    overlap_df = overlap_df.sort_values(
        "standardized_mean_difference",
        ascending=False
    )

    display(overlap_df)


# ============================================================
# 15. HEALTHY / LATE-BLIGHT QUANTILE OVERLAP
# ============================================================

print("\n" + "=" * 80)
print("15. HEALTHY / LATE-BLIGHT QUANTILE OVERLAP")
print("=" * 80)

quantile_rows = []

for col in available_env:

    healthy = image_df.loc[healthy_mask, col].dropna()
    late = image_df.loc[late_mask, col].dropna()

    if len(healthy) == 0 or len(late) == 0:
        continue

    q = [0.05, 0.25, 0.50, 0.75, 0.95]

    hq = healthy.quantile(q)
    lq = late.quantile(q)

    quantile_rows.append({
        "feature": col,

        "healthy_q05": hq.loc[0.05],
        "healthy_q25": hq.loc[0.25],
        "healthy_q50": hq.loc[0.50],
        "healthy_q75": hq.loc[0.75],
        "healthy_q95": hq.loc[0.95],

        "late_q05": lq.loc[0.05],
        "late_q25": lq.loc[0.25],
        "late_q50": lq.loc[0.50],
        "late_q75": lq.loc[0.75],
        "late_q95": lq.loc[0.95],
    })

quantile_df = pd.DataFrame(quantile_rows)

display(quantile_df)


# ============================================================
# 16. OVERLAP USING RANGE INTERSECTION
# ============================================================

print("\n" + "=" * 80)
print("16. RANGE OVERLAP")
print("=" * 80)

range_rows = []

for col in available_env:

    healthy = image_df.loc[healthy_mask, col].dropna()
    late = image_df.loc[late_mask, col].dropna()

    if len(healthy) == 0 or len(late) == 0:
        continue

    h_min, h_max = healthy.min(), healthy.max()
    l_min, l_max = late.min(), late.max()

    overlap_min = max(h_min, l_min)
    overlap_max = min(h_max, l_max)

    if overlap_max >= overlap_min:
        overlap_width = overlap_max - overlap_min
    else:
        overlap_width = 0

    total_min = min(h_min, l_min)
    total_max = max(h_max, l_max)
    total_width = total_max - total_min

    overlap_fraction = (
        overlap_width / total_width
        if total_width > 0 else np.nan
    )

    range_rows.append({
        "feature": col,
        "healthy_min": h_min,
        "healthy_max": h_max,
        "late_min": l_min,
        "late_max": l_max,
        "overlap_min": overlap_min,
        "overlap_max": overlap_max,
        "overlap_fraction_of_total_range": overlap_fraction
    })

range_overlap_df = pd.DataFrame(range_rows)

display(
    range_overlap_df.sort_values(
        "overlap_fraction_of_total_range",
        ascending=True
    )
)


# ============================================================
# 17. CLASS-ONLY CORRECTNESS CHECK
# ============================================================

print("\n" + "=" * 80)
print("17. CLASS ↔ SEVERITY CONSISTENCY — CORRECTED")
print("=" * 80)

healthy_nonzero = image_df[
    healthy_mask &
    (image_df["severity_percent"] > 0)
]

late_zero = image_df[
    late_mask &
    (image_df["severity_percent"] <= 0)
]

print("Healthy with severity > 0:", len(healthy_nonzero))
print("Late-blight with severity <= 0:", len(late_zero))

if len(healthy_nonzero):
    display(healthy_nonzero.head())

if len(late_zero):
    display(late_zero.head())


# ============================================================
# 18. ENVIRONMENT-ONLY SINGLE FEATURE CLASS SEPARABILITY
# ============================================================

print("\n" + "=" * 80)
print("18. ENVIRONMENT-ONLY SINGLE-FEATURE SEPARABILITY")
print("=" * 80)

binary_df = image_df.copy()

binary_df["is_late_blight"] = late_mask.astype(int)

threshold_rows = []

for col in available_env:

    x = binary_df[[col, "is_late_blight"]].dropna()

    if len(x) < 20 or x[col].nunique() < 2:
        continue

    v = x[col].values
    y = x["is_late_blight"].values

    thresholds = np.unique(
        np.quantile(
            v,
            np.linspace(0.01, 0.99, 200)
        )
    )

    best_acc = -1
    best_bal_acc = -1
    best_threshold = None
    best_direction = None

    for t in thresholds:

        # High = late blight
        pred_high = (v >= t).astype(int)

        # Low = late blight
        pred_low = (v <= t).astype(int)

        for pred, direction in [
            (pred_high, "late_blight_if_high"),
            (pred_low, "late_blight_if_low")
        ]:

            acc = (pred == y).mean()

            # Balanced accuracy
            tp = ((pred == 1) & (y == 1)).sum()
            tn = ((pred == 0) & (y == 0)).sum()

            p = (y == 1).sum()
            n = (y == 0).sum()

            sensitivity = tp / p if p else 0
            specificity = tn / n if n else 0

            bal_acc = (
                sensitivity + specificity
            ) / 2

            if bal_acc > best_bal_acc:
                best_bal_acc = bal_acc
                best_acc = acc
                best_threshold = t
                best_direction = direction

    threshold_rows.append({
        "feature": col,
        "accuracy": best_acc,
        "balanced_accuracy": best_bal_acc,
        "threshold": best_threshold,
        "direction": best_direction
    })

threshold_df = (
    pd.DataFrame(threshold_rows)
    .sort_values(
        "balanced_accuracy",
        ascending=False
    )
)

display(threshold_df)


# ============================================================
# 19. VERY IMPORTANT: ENVIRONMENT-ONLY REGRESSION SIGNAL
# ============================================================

print("\n" + "=" * 80)
print("19. ENVIRONMENT-ONLY SEVERITY SIGNAL")
print("=" * 80)

# Simple Spearman correlations among environmental variables
# and severity. This helps assess how much information about
# the target exists without the image.

severity_signal = corr_df.copy()

severity_signal["abs_spearman"] = (
    severity_signal["spearman_r"].abs()
)

display(
    severity_signal.sort_values(
        "abs_spearman",
        ascending=False
    )
)


# ============================================================
# 20. CHECK DUPLICATED IMAGE IDs
# ============================================================

print("\n" + "=" * 80)
print("20. MULTI-TRAJECTORY GROUPING CHECK")
print("=" * 80)

image_group_counts = (
    image_df
    .groupby("image_id")
    .agg(
        rows=("candidate_id", "size"),
        unique_split=("split", "nunique"),
        unique_severity=("severity_percent", "nunique"),
        unique_class=("class_name", "nunique")
    )
)

print("Image groups:", len(image_group_counts))

print(
    "\nRows/group distribution:"
)
print(
    image_group_counts["rows"]
    .value_counts()
    .sort_index()
)

print(
    "\nGroups with split conflicts:",
    (image_group_counts["unique_split"] > 1).sum()
)

print(
    "Groups with severity conflicts:",
    (image_group_counts["unique_severity"] > 1).sum()
)

print(
    "Groups with class conflicts:",
    (image_group_counts["unique_class"] > 1).sum()
)


# ============================================================
# 21. FINAL AUDIT DECISION HELPERS
# ============================================================

print("\n" + "=" * 80)
print("21. AUTOMATED WARNING FLAGS")
print("=" * 80)

# Highest severity correlation
max_corr_row = corr_df.iloc[
    corr_df["spearman_r"].abs().argmax()
]

max_corr = max_corr_row["spearman_r"]
max_corr_feature = max_corr_row["feature"]

print(
    f"Highest |Spearman(severity, environment)|: "
    f"{max_corr_feature} = {max_corr:.3f}"
)

# Highest class separability
if not threshold_df.empty:

    best_class_feature = threshold_df.iloc[0]

    print(
        f"Best single-feature balanced accuracy: "
        f"{best_class_feature['feature']} = "
        f"{best_class_feature['balanced_accuracy']:.3f}"
    )

    if best_class_feature["balanced_accuracy"] >= 0.90:
        print(
            "\n⚠️ WARNING: A single environment feature gives "
            "≥90% balanced accuracy for Healthy/Late-Blight."
        )
    elif best_class_feature["balanced_accuracy"] >= 0.80:
        print(
            "\n⚠️ CAUTION: A single environment feature gives "
            "≥80% balanced accuracy."
        )
    else:
        print(
            "\n✓ No single environment feature has very strong "
            "class separation."
        )

# Extremely high severity correlations
if abs(max_corr) >= 0.90:
    print(
        "⚠️ WARNING: severity/environment correlation "
        "may be excessively deterministic."
    )
elif abs(max_corr) >= 0.80:
    print(
        "⚠️ CAUTION: strong severity/environment coupling."
    )
else:
    print(
        "✓ Severity/environment correlation is not "
        "extremely deterministic."
    )


# ============================================================
# 22. SAVE AUDIT RESULTS
# ============================================================

corr_path = os.path.join(
    BASE_DIR,
    "AUDIT_severity_environment_correlations.csv"
)

overlap_path = os.path.join(
    BASE_DIR,
    "AUDIT_healthy_late_overlap.csv"
)

quantile_path = os.path.join(
    BASE_DIR,
    "AUDIT_healthy_late_quantiles.csv"
)

range_path = os.path.join(
    BASE_DIR,
    "AUDIT_healthy_late_range_overlap.csv"
)

threshold_path = os.path.join(
    BASE_DIR,
    "AUDIT_environment_only_class_separability.csv"
)

corr_df.to_csv(corr_path, index=False)
overlap_df.to_csv(overlap_path, index=False)
quantile_df.to_csv(quantile_path, index=False)
range_overlap_df.to_csv(range_path, index=False)
threshold_df.to_csv(threshold_path, index=False)

print("\nSaved:")
print(corr_path)
print(overlap_path)
print(quantile_path)
print(range_path)
print(threshold_path)


# ============================================================
# 23. COMPACT FINAL REPORT
# ============================================================

print("\n" + "=" * 80)
print("FINAL AUDIT REPORT")
print("=" * 80)

print(f"""
Total trajectories:              {len(image_df):,}
Unique images:                   {image_df['image_id'].nunique():,}
Trajectories/image:              {len(image_df) / image_df['image_id'].nunique():.1f}

Healthy trajectories:            {healthy_mask.sum():,}
Late-blight trajectories:        {late_mask.sum():,}

Cross-split image groups:        {(image_group_counts['unique_split'] > 1).sum()}
Severity-inconsistent groups:   {(image_group_counts['unique_severity'] > 1).sum()}

Healthy severity > 0:            {len(healthy_nonzero)}
Late-blight severity <= 0:       {len(late_zero)}

Highest severity correlation:
    {max_corr_feature} = {max_corr:.3f}

Best environment-only
class balanced accuracy:
    {threshold_df.iloc[0]['feature'] if not threshold_df.empty else 'N/A'}
    {threshold_df.iloc[0]['balanced_accuracy'] if not threshold_df.empty else np.nan:.3f}
""")

print("=" * 80)
print("AUDIT CONTINUED SUCCESSFULLY")
print("=" * 80)
```

```python
# ============================================================
# POTATO LATE-BLIGHT
# ENVIRONMENT-ONLY CONTINUOUS SEVERITY REGRESSION BASELINE
# ============================================================

import os
import numpy as np
import pandas as pd

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

from scipy.stats import spearmanr


# ============================================================
# 1. PATH
# ============================================================

BASE_DIR = "/content/Potato_Synthetic_Environment_FINAL"

IMG_PATH = os.path.join(
    BASE_DIR,
    "synthetic_environment_image_level.csv"
)

df = pd.read_csv(IMG_PATH)

print("=" * 80)
print("ENVIRONMENT-ONLY SEVERITY REGRESSION")
print("=" * 80)

print("Dataset shape:", df.shape)


# ============================================================
# 2. TARGET
# ============================================================

TARGET = "severity_percent"

# IMPORTANT:
# Do NOT use these as predictors:
# - severity_percent     -> target leakage
# - latent_risk          -> internal generator variable
# - candidate_id         -> metadata
# - random_seed          -> metadata
# - generation_method    -> metadata
# - qc_status            -> metadata
# - filename             -> identity/metadata
# - image_id             -> grouping identifier
# - split                -> split metadata


# ============================================================
# 3. RAW ENVIRONMENT FEATURES
# ============================================================

RAW_ENV_FEATURES = [
    "temperature_mean",
    "temperature_min",
    "temperature_max",
    "rh_mean",
    "rh_min",
    "rh_max",
    "rh_ge90_hours",
    "leaf_wetness_hours",
    "rainfall_total_mm",
    "soil_moisture_mean",
]

EPI_FEATURES = [
    "simcast_bu",
    "wallin_sv",
]

ALL_ENV_FEATURES = RAW_ENV_FEATURES + EPI_FEATURES

available_raw = [
    c for c in RAW_ENV_FEATURES
    if c in df.columns
]

available_all = [
    c for c in ALL_ENV_FEATURES
    if c in df.columns
]

print("\nRaw environmental features:")
print(available_raw)

print("\nAll environmental features:")
print(available_all)


# ============================================================
# 4. GROUP INTEGRITY CHECK
# ============================================================

print("\n" + "=" * 80)
print("GROUP INTEGRITY")
print("=" * 80)

group_check = (
    df.groupby("image_id")
      .agg(
          rows=("candidate_id", "size"),
          split_count=("split", "nunique"),
          severity_count=("severity_percent", "nunique"),
          class_count=("class_name", "nunique")
      )
)

print("Unique image groups:", len(group_check))
print("Rows/group distribution:")
print(group_check["rows"].value_counts().sort_index())

print(
    "Groups spanning multiple splits:",
    (group_check["split_count"] > 1).sum()
)

print(
    "Groups with severity inconsistency:",
    (group_check["severity_count"] > 1).sum()
)

print(
    "Groups with class inconsistency:",
    (group_check["class_count"] > 1).sum()
)

assert (group_check["split_count"] == 1).all()
assert (group_check["severity_count"] == 1).all()
assert (group_check["class_count"] == 1).all()


# ============================================================
# 5. HELPER: METRICS
# ============================================================

def calculate_metrics(y_true, y_pred):
    mae = mean_absolute_error(y_true, y_pred)

    rmse = np.sqrt(
        mean_squared_error(y_true, y_pred)
    )

    r2 = r2_score(
        y_true,
        y_pred
    )

    rho, rho_p = spearmanr(
        y_true,
        y_pred
    )

    return {
        "MAE": mae,
        "RMSE": rmse,
        "R2": r2,
        "Spearman": rho,
        "Spearman_p": rho_p
    }


# ============================================================
# 6. MODEL
# ============================================================

def build_model():

    return Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),

        (
            "scaler",
            StandardScaler()
        ),

        (
            "model",
            Ridge(alpha=10.0)
        )
    ])


# ============================================================
# 7. EVALUATE USING PREDEFINED SPLITS
# ============================================================

def run_baseline(feature_cols, experiment_name):

    print("\n" + "=" * 80)
    print(experiment_name)
    print("=" * 80)

    X = df[feature_cols].copy()
    y = df[TARGET].copy()

    predictions = np.zeros(len(df), dtype=float)

    metrics_rows = []

    # --------------------------------------------------------
    # IMPORTANT:
    # Train only on rows from train images.
    # Validation/test contain completely unseen image groups.
    # --------------------------------------------------------

    train_mask = df["split"] == "train"
    valid_mask = df["split"] == "valid"
    test_mask  = df["split"] == "test"

    model = build_model()

    model.fit(
        X.loc[train_mask],
        y.loc[train_mask]
    )

    predictions[valid_mask] = model.predict(
        X.loc[valid_mask]
    )

    predictions[test_mask] = model.predict(
        X.loc[test_mask]
    )

    # --------------------------------------------------------
    # TRAIN METRIC
    # --------------------------------------------------------

    train_pred = model.predict(
        X.loc[train_mask]
    )

    train_metrics = calculate_metrics(
        y.loc[train_mask],
        train_pred
    )

    metrics_rows.append({
        "split": "train",
        **train_metrics
    })

    # --------------------------------------------------------
    # VALIDATION
    # --------------------------------------------------------

    valid_metrics = calculate_metrics(
        y.loc[valid_mask],
        predictions[valid_mask]
    )

    metrics_rows.append({
        "split": "valid",
        **valid_metrics
    })

    # --------------------------------------------------------
    # TEST
    # --------------------------------------------------------

    test_metrics = calculate_metrics(
        y.loc[test_mask],
        predictions[test_mask]
    )

    metrics_rows.append({
        "split": "test",
        **test_metrics
    })

    metrics_df = pd.DataFrame(metrics_rows)

    print("\nFeature set:")
    print(feature_cols)

    display(metrics_df)

    return model, metrics_df, predictions


# ============================================================
# 8. BASELINE A
# RAW ENVIRONMENT ONLY
# ============================================================

raw_model, raw_metrics, raw_pred = run_baseline(
    available_raw,
    "BASELINE A — RAW ENVIRONMENT ONLY"
)


# ============================================================
# 9. BASELINE B
# RAW ENVIRONMENT + EPIDEMIOLOGICAL FEATURES
# ============================================================

all_model, all_metrics, all_pred = run_baseline(
    available_all,
    "BASELINE B — RAW ENVIRONMENT + SIMCAST + WALLIN"
)


# ============================================================
# 10. COMPARE THE TWO BASELINES
# ============================================================

print("\n" + "=" * 80)
print("BASELINE COMPARISON")
print("=" * 80)

comparison = pd.DataFrame([
    {
        "experiment": "Raw environment",
        "test_MAE": raw_metrics.loc[
            raw_metrics["split"] == "test",
            "MAE"
        ].iloc[0],
        "test_RMSE": raw_metrics.loc[
            raw_metrics["split"] == "test",
            "RMSE"
        ].iloc[0],
        "test_R2": raw_metrics.loc[
            raw_metrics["split"] == "test",
            "R2"
        ].iloc[0],
        "test_Spearman": raw_metrics.loc[
            raw_metrics["split"] == "test",
            "Spearman"
        ].iloc[0],
    },

    {
        "experiment": "Raw + SIMCAST + Wallin",
        "test_MAE": all_metrics.loc[
            all_metrics["split"] == "test",
            "MAE"
        ].iloc[0],
        "test_RMSE": all_metrics.loc[
            all_metrics["split"] == "test",
            "RMSE"
        ].iloc[0],
        "test_R2": all_metrics.loc[
            all_metrics["split"] == "test",
            "R2"
        ].iloc[0],
        "test_Spearman": all_metrics.loc[
            all_metrics["split"] == "test",
            "Spearman"
        ].iloc[0],
    }
])

display(comparison)


# ============================================================
# 11. SINGLE-TRAJECTORY-PER-IMAGE BASELINE
# ============================================================
#
# This removes the 10x duplicated-image effect.
#
# For each image_id, candidate 0 is selected.
#
# This is useful as the cleanest "1 image = 1 sample"
# environment-only baseline.
# ============================================================

print("\n" + "=" * 80)
print("SINGLE-TRAJECTORY / IMAGE BASELINE")
print("=" * 80)

single_df = (
    df.sort_values(
        ["image_id", "candidate_id"]
    )
    .groupby("image_id", as_index=False)
    .first()
)

print("Single-trajectory rows:", len(single_df))

print(
    single_df["split"]
    .value_counts()
)

print(
    single_df["class_name"]
    .value_counts()
)


# ------------------------------------------------------------
# Function for single-image dataset
# ------------------------------------------------------------

def run_single_baseline(
    data,
    feature_cols,
    experiment_name
):

    print("\n" + "-" * 80)
    print(experiment_name)
    print("-" * 80)

    X = data[feature_cols].copy()
    y = data[TARGET].copy()

    train_mask = data["split"] == "train"
    valid_mask = data["split"] == "valid"
    test_mask  = data["split"] == "test"

    model = build_model()

    model.fit(
        X.loc[train_mask],
        y.loc[train_mask]
    )

    valid_pred = model.predict(
        X.loc[valid_mask]
    )

    test_pred = model.predict(
        X.loc[test_mask]
    )

    valid_metrics = calculate_metrics(
        y.loc[valid_mask],
        valid_pred
    )

    test_metrics = calculate_metrics(
        y.loc[test_mask],
        test_pred
    )

    result = pd.DataFrame([
        {
            "split": "valid",
            **valid_metrics
        },
        {
            "split": "test",
            **test_metrics
        }
    ])

    display(result)

    return model, result


single_raw_model, single_raw_metrics = run_single_baseline(
    single_df,
    available_raw,
    "SINGLE IMAGE — RAW ENVIRONMENT"
)

single_all_model, single_all_metrics = run_single_baseline(
    single_df,
    available_all,
    "SINGLE IMAGE — RAW + SIMCAST + WALLIN"
)


# ============================================================
# 12. SAVE RESULTS
# ============================================================

RESULT_DIR = os.path.join(
    BASE_DIR,
    "environment_baseline_results"
)

os.makedirs(
    RESULT_DIR,
    exist_ok=True
)

raw_metrics.to_csv(
    os.path.join(
        RESULT_DIR,
        "raw_environment_metrics.csv"
    ),
    index=False
)

all_metrics.to_csv(
    os.path.join(
        RESULT_DIR,
        "raw_plus_epidemiology_metrics.csv"
    ),
    index=False
)

comparison.to_csv(
    os.path.join(
        RESULT_DIR,
        "baseline_comparison.csv"
    ),
    index=False
)

single_raw_metrics.to_csv(
    os.path.join(
        RESULT_DIR,
        "single_image_raw_metrics.csv"
    ),
    index=False
)

single_all_metrics.to_csv(
    os.path.join(
        RESULT_DIR,
        "single_image_raw_plus_epi_metrics.csv"
    ),
    index=False
)


# ============================================================
# 13. FINAL INTERPRETATION TEMPLATE
# ============================================================

test_raw = raw_metrics.loc[
    raw_metrics["split"] == "test"
].iloc[0]

test_all = all_metrics.loc[
    all_metrics["split"] == "test"
].iloc[0]

print("\n" + "=" * 80)
print("ENVIRONMENT-ONLY BASELINE SUMMARY")
print("=" * 80)

print(f"""
RAW ENVIRONMENT
---------------
Test MAE       : {test_raw['MAE']:.4f}
Test RMSE      : {test_raw['RMSE']:.4f}
Test R²        : {test_raw['R2']:.4f}
Test Spearman  : {test_raw['Spearman']:.4f}


RAW + SIMCAST + WALLIN
----------------------
Test MAE       : {test_all['MAE']:.4f}
Test RMSE      : {test_all['RMSE']:.4f}
Test R²        : {test_all['R2']:.4f}
Test Spearman  : {test_all['Spearman']:.4f}
""")

print("=" * 80)
print("BASELINE COMPLETE")
print("=" * 80)
```

```python
# ============================================================
# POTATO LATE-BLIGHT
# IMAGE-ONLY SEVERITY REGRESSION BASELINE
# ResNet50 ImageNet pretrained
# ============================================================

import os
import copy
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from PIL import Image

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from torchvision import models, transforms

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

from scipy.stats import spearmanr


# ============================================================
# 1. CONFIG
# ============================================================

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)

DEVICE = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Device:", DEVICE)

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))


BASE_DIR = "/content/Potato_Synthetic_Environment_FINAL"

IMAGE_CSV = os.path.join(
    BASE_DIR,
    "synthetic_environment_image_level.csv"
)

df_full = pd.read_csv(IMAGE_CSV)

print("\nFull rows:", len(df_full))
print("Unique images:", df_full["image_id"].nunique())


# ============================================================
# 2. CREATE ONE ROW PER IMAGE
# ============================================================

# Every image has 10 trajectories.
# For image-only modeling, we must keep exactly one row/image.

df = (
    df_full
    .sort_values(["image_id", "candidate_id"])
    .groupby("image_id", as_index=False)
    .first()
)

print("\nImage-level dataset:")
print(df.shape)

print(df["split"].value_counts())

assert len(df) == 1153
assert df["image_id"].nunique() == 1153


# ============================================================
# 3. CHECK IMAGE PATH COLUMN
# ============================================================

required_cols = [
    "image_id",
    "filename",
    "split",
    "severity_percent"
]

missing = [
    c for c in required_cols
    if c not in df.columns
]

if missing:
    raise ValueError(
        f"Missing columns: {missing}"
    )


# The image path is NOT present in the synthetic environment CSV
# shown in your audit, so recover it from the original filename
# only if the files are actually available.
#
# IMPORTANT:
# EDIT IMAGE_ROOT below to wherever your finalized image dataset
# exists in Colab.

IMAGE_ROOT = "/content/Potato_Severity_Dataset_FINAL"


# ============================================================
# 4. FIND IMAGE FILES
# ============================================================

# Build a recursive filename index once.

from pathlib import Path

print("\nIndexing image files...")

all_image_files = {}

valid_extensions = {
    ".jpg",
    ".jpeg",
    ".png",
    ".JPG",
    ".JPEG",
    ".PNG"
}

for p in Path(IMAGE_ROOT).rglob("*"):

    if p.is_file() and p.suffix in valid_extensions:

        all_image_files[p.name] = str(p)

print(
    "Indexed image files:",
    len(all_image_files)
)


# ============================================================
# 5. MAP CSV FILENAMES TO ACTUAL IMAGE PATHS
# ============================================================

df["image_path"] = df["filename"].map(
    all_image_files
)

missing_images = df["image_path"].isna().sum()

print(
    "Images missing from filesystem:",
    missing_images
)

if missing_images > 0:

    print("\nExamples of missing filenames:")

    display(
        df.loc[
            df["image_path"].isna(),
            ["filename", "image_id", "split"]
        ].head(20)
    )

    raise FileNotFoundError(
        "Some images could not be found. "
        "Check IMAGE_ROOT."
    )


# ============================================================
# 6. DATASET CLASS
# ============================================================

class PotatoSeverityDataset(Dataset):

    def __init__(
        self,
        dataframe,
        transform=None
    ):

        self.data = dataframe.reset_index(
            drop=True
        )

        self.transform = transform

    def __len__(self):
        return len(self.data)

    def __getitem__(self, idx):

        row = self.data.iloc[idx]

        image = Image.open(
            row["image_path"]
        ).convert("RGB")

        if self.transform:
            image = self.transform(image)

        severity = torch.tensor(
            row["severity_percent"],
            dtype=torch.float32
        )

        return image, severity


# ============================================================
# 7. TRANSFORMS
# ============================================================

IMG_SIZE = 224

train_transform = transforms.Compose([
    transforms.Resize(
        (IMG_SIZE, IMG_SIZE)
    ),

    transforms.RandomHorizontalFlip(
        p=0.5
    ),

    transforms.RandomVerticalFlip(
        p=0.2
    ),

    transforms.RandomRotation(
        degrees=10
    ),

    transforms.ColorJitter(
        brightness=0.15,
        contrast=0.15,
        saturation=0.10
    ),

    transforms.ToTensor(),

    transforms.Normalize(
        mean=[
            0.485,
            0.456,
            0.406
        ],
        std=[
            0.229,
            0.224,
            0.225
        ]
    )
])


eval_transform = transforms.Compose([
    transforms.Resize(
        (IMG_SIZE, IMG_SIZE)
    ),

    transforms.ToTensor(),

    transforms.Normalize(
        mean=[
            0.485,
            0.456,
            0.406
        ],
        std=[
            0.229,
            0.224,
            0.225
        ]
    )
])


# ============================================================
# 8. SPLITS
# ============================================================

train_df = df[df["split"] == "train"].copy()
valid_df = df[df["split"] == "valid"].copy()
test_df  = df[df["split"] == "test"].copy()

print("\nSplit sizes:")
print("Train:", len(train_df))
print("Valid:", len(valid_df))
print("Test :", len(test_df))


train_ds = PotatoSeverityDataset(
    train_df,
    train_transform
)

valid_ds = PotatoSeverityDataset(
    valid_df,
    eval_transform
)

test_ds = PotatoSeverityDataset(
    test_df,
    eval_transform
)


# ============================================================
# 9. DATALOADERS
# ============================================================

BATCH_SIZE = 32

train_loader = DataLoader(
    train_ds,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=2,
    pin_memory=True
)

valid_loader = DataLoader(
    valid_ds,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)

test_loader = DataLoader(
    test_ds,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)


# ============================================================
# 10. MODEL
# ============================================================

weights = (
    models.ResNet50_Weights.IMAGENET1K_V2
)

model = models.resnet50(
    weights=weights
)

num_features = model.fc.in_features

model.fc = nn.Sequential(
    nn.Dropout(0.30),
    nn.Linear(num_features, 1)
)

model = model.to(DEVICE)


# ============================================================
# 11. LOSS / OPTIMIZER
# ============================================================

criterion = nn.SmoothL1Loss(
    beta=5.0
)

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=1e-4,
    weight_decay=1e-4
)


# ============================================================
# 12. METRIC FUNCTION
# ============================================================

def regression_metrics(
    y_true,
    y_pred
):

    mae = mean_absolute_error(
        y_true,
        y_pred
    )

    rmse = np.sqrt(
        mean_squared_error(
            y_true,
            y_pred
        )
    )

    r2 = r2_score(
        y_true,
        y_pred
    )

    rho, p = spearmanr(
        y_true,
        y_pred
    )

    return {
        "MAE": mae,
        "RMSE": rmse,
        "R2": r2,
        "Spearman": rho,
        "Spearman_p": p
    }


# ============================================================
# 13. EVALUATION
# ============================================================

@torch.no_grad()
def evaluate(
    model,
    loader
):

    model.eval()

    all_true = []
    all_pred = []

    for images, targets in loader:

        images = images.to(
            DEVICE,
            non_blocking=True
        )

        outputs = model(
            images
        ).squeeze(1)

        all_true.extend(
            targets.numpy()
        )

        all_pred.extend(
            outputs.cpu().numpy()
        )

    all_true = np.asarray(
        all_true
    )

    all_pred = np.asarray(
        all_pred
    )

    metrics = regression_metrics(
        all_true,
        all_pred
    )

    return metrics, all_true, all_pred


# ============================================================
# 14. TRAINING LOOP
# ============================================================

EPOCHS = 15

best_val_r2 = -np.inf

best_state = None

history = []

for epoch in range(1, EPOCHS + 1):

    model.train()

    running_loss = 0.0

    n_samples = 0

    for images, targets in train_loader:

        images = images.to(
            DEVICE,
            non_blocking=True
        )

        targets = targets.to(
            DEVICE,
            non_blocking=True
        )

        optimizer.zero_grad(
            set_to_none=True
        )

        predictions = (
            model(images)
            .squeeze(1)
        )

        loss = criterion(
            predictions,
            targets
        )

        loss.backward()

        optimizer.step()

        batch_size = images.size(0)

        running_loss += (
            loss.item() * batch_size
        )

        n_samples += batch_size

    train_loss = (
        running_loss /
        n_samples
    )

    train_metrics, _, _ = evaluate(
        model,
        train_loader
    )

    val_metrics, _, _ = evaluate(
        model,
        valid_loader
    )

    history.append({
        "epoch": epoch,
        "train_loss": train_loss,
        "train_MAE": train_metrics["MAE"],
        "train_RMSE": train_metrics["RMSE"],
        "train_R2": train_metrics["R2"],
        "val_MAE": val_metrics["MAE"],
        "val_RMSE": val_metrics["RMSE"],
        "val_R2": val_metrics["R2"],
        "val_Spearman": val_metrics["Spearman"]
    })

    print(
        f"Epoch {epoch:02d}/{EPOCHS} | "
        f"Loss={train_loss:.4f} | "
        f"Train R²={train_metrics['R2']:.4f} | "
        f"Val R²={val_metrics['R2']:.4f} | "
        f"Val MAE={val_metrics['MAE']:.4f}"
    )

    if val_metrics["R2"] > best_val_r2:

        best_val_r2 = val_metrics["R2"]

        best_state = copy.deepcopy(
            model.state_dict()
        )


# ============================================================
# 15. RESTORE BEST MODEL
# ============================================================

model.load_state_dict(
    best_state
)


# ============================================================
# 16. FINAL TRAIN / VALID / TEST
# ============================================================

train_metrics, train_true, train_pred = evaluate(
    model,
    train_loader
)

valid_metrics, valid_true, valid_pred = evaluate(
    model,
    valid_loader
)

test_metrics, test_true, test_pred = evaluate(
    model,
    test_loader
)


results = pd.DataFrame([
    {
        "split": "train",
        **train_metrics
    },
    {
        "split": "valid",
        **valid_metrics
    },
    {
        "split": "test",
        **test_metrics
    }
])

print("\n" + "=" * 80)
print("IMAGE-ONLY SEVERITY REGRESSION RESULTS")
print("=" * 80)

display(results)


# ============================================================
# 17. SAVE RESULTS
# ============================================================

RESULT_DIR = os.path.join(
    BASE_DIR,
    "image_baseline_results"
)

os.makedirs(
    RESULT_DIR,
    exist_ok=True
)

results.to_csv(
    os.path.join(
        RESULT_DIR,
        "image_only_metrics.csv"
    ),
    index=False
)

pd.DataFrame(history).to_csv(
    os.path.join(
        RESULT_DIR,
        "training_history.csv"
    ),
    index=False
)


# ============================================================
# 18. PREDICTION TABLE
# ============================================================

prediction_df = test_df[
    [
        "image_id",
        "filename",
        "severity_percent"
    ]
].copy()

prediction_df["prediction"] = (
    test_pred
)

prediction_df["absolute_error"] = (
    np.abs(
        prediction_df["severity_percent"]
        -
        prediction_df["prediction"]
    )
)

prediction_df.to_csv(
    os.path.join(
        RESULT_DIR,
        "test_predictions.csv"
    ),
    index=False
)


# ============================================================
# 19. PLOT — TRUE vs PREDICTED
# ============================================================

plt.figure(figsize=(7, 7))

plt.scatter(
    test_true,
    test_pred,
    alpha=0.6,
    s=30
)

lims = [
    min(
        test_true.min(),
        test_pred.min()
    ),
    max(
        test_true.max(),
        test_pred.max()
    )
]

plt.plot(
    lims,
    lims,
    linestyle="--"
)

plt.xlabel("True Severity (%)")
plt.ylabel("Predicted Severity (%)")
plt.title("Image-only Severity Regression")

plt.grid(
    alpha=0.25
)

plt.show()


# ============================================================
# 20. PLOT — ERROR VS TRUE SEVERITY
# ============================================================

errors = (
    test_pred -
    test_true
)

plt.figure(figsize=(8, 5))

plt.scatter(
    test_true,
    errors,
    alpha=0.6,
    s=30
)

plt.axhline(
    0,
    linestyle="--"
)

plt.xlabel("True Severity (%)")
plt.ylabel("Prediction Error")
plt.title("Image-only Prediction Error")

plt.grid(
    alpha=0.25
)

plt.show()


# ============================================================
# 21. FINAL TEST SUMMARY
# ============================================================

print("\n" + "=" * 80)
print("FINAL TEST SUMMARY")
print("=" * 80)

print(
    f"MAE      : {test_metrics['MAE']:.4f}"
)

print(
    f"RMSE     : {test_metrics['RMSE']:.4f}"
)

print(
    f"R²       : {test_metrics['R2']:.4f}"
)

print(
    f"Spearman : {test_metrics['Spearman']:.4f}"
)

print("\nResults saved to:")
print(RESULT_DIR)
```

```python
# ============================================================
# IMAGE-ONLY BASELINE — DETAILED TEST AUDIT
# ============================================================

import os
import numpy as np
import pandas as pd

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

from scipy.stats import spearmanr


BASE_DIR = "/content/Potato_Synthetic_Environment_FINAL"

PRED_PATH = os.path.join(
    BASE_DIR,
    "image_baseline_results",
    "test_predictions.csv"
)

pred_df = pd.read_csv(PRED_PATH)

print("=" * 80)
print("IMAGE-ONLY TEST PREDICTION AUDIT")
print("=" * 80)

print(pred_df.head())
print("\nRows:", len(pred_df))


# ------------------------------------------------------------
# 1. BASIC METRICS
# ------------------------------------------------------------

y_true = pred_df["severity_percent"].values
y_pred = pred_df["prediction"].values

mae = mean_absolute_error(y_true, y_pred)

rmse = np.sqrt(
    mean_squared_error(y_true, y_pred)
)

r2 = r2_score(y_true, y_pred)

rho, p = spearmanr(
    y_true,
    y_pred
)

print("\nALL TEST IMAGES")
print("----------------")
print("MAE      :", mae)
print("RMSE     :", rmse)
print("R2       :", r2)
print("Spearman :", rho)


# ------------------------------------------------------------
# 2. HEALTHY-ONLY
# ------------------------------------------------------------

healthy_df = pred_df[
    pred_df["severity_percent"] == 0
].copy()

print("\n" + "=" * 80)
print("HEALTHY-ONLY TEST")
print("=" * 80)

if len(healthy_df):

    healthy_mae = mean_absolute_error(
        healthy_df["severity_percent"],
        healthy_df["prediction"]
    )

    healthy_rmse = np.sqrt(
        mean_squared_error(
            healthy_df["severity_percent"],
            healthy_df["prediction"]
        )
    )

    print("Samples:", len(healthy_df))
    print("MAE:", healthy_mae)
    print("RMSE:", healthy_rmse)

    print(
        "\nHealthy prediction summary:"
    )

    print(
        healthy_df["prediction"].describe()
    )


# ------------------------------------------------------------
# 3. LATE-BLIGHT ONLY
# ------------------------------------------------------------

lb_df = pred_df[
    pred_df["severity_percent"] > 0
].copy()

print("\n" + "=" * 80)
print("LATE-BLIGHT-ONLY TEST")
print("=" * 80)

if len(lb_df):

    lb_mae = mean_absolute_error(
        lb_df["severity_percent"],
        lb_df["prediction"]
    )

    lb_rmse = np.sqrt(
        mean_squared_error(
            lb_df["severity_percent"],
            lb_df["prediction"]
        )
    )

    lb_r2 = r2_score(
        lb_df["severity_percent"],
        lb_df["prediction"]
    )

    lb_rho, lb_p = spearmanr(
        lb_df["severity_percent"],
        lb_df["prediction"]
    )

    print("Samples:", len(lb_df))
    print("MAE:", lb_mae)
    print("RMSE:", lb_rmse)
    print("R2:", lb_r2)
    print("Spearman:", lb_rho)


# ------------------------------------------------------------
# 4. BASELINE: ALWAYS PREDICT TRAIN MEAN
# ------------------------------------------------------------

# Need original training target distribution
IMG_CSV = os.path.join(
    BASE_DIR,
    "synthetic_environment_image_level.csv"
)

full_df = pd.read_csv(IMG_CSV)

one_per_image = (
    full_df
    .sort_values(["image_id", "candidate_id"])
    .groupby("image_id", as_index=False)
    .first()
)

train_mean = one_per_image.loc[
    one_per_image["split"] == "train",
    "severity_percent"
].mean()

mean_pred = np.full_like(
    y_true,
    train_mean,
    dtype=float
)

baseline_mae = mean_absolute_error(
    y_true,
    mean_pred
)

baseline_rmse = np.sqrt(
    mean_squared_error(
        y_true,
        mean_pred
    )
)

baseline_r2 = r2_score(
    y_true,
    mean_pred
)

print("\n" + "=" * 80)
print("MEAN-PREDICTION BASELINE")
print("=" * 80)

print("Training severity mean:", train_mean)
print("MAE :", baseline_mae)
print("RMSE:", baseline_rmse)
print("R2  :", baseline_r2)


# ------------------------------------------------------------
# 5. LARGEST ERRORS
# ------------------------------------------------------------

pred_df["error"] = (
    pred_df["prediction"]
    -
    pred_df["severity_percent"]
)

pred_df["abs_error"] = (
    pred_df["error"].abs()
)

print("\n" + "=" * 80)
print("LARGEST TEST ERRORS")
print("=" * 80)

display(
    pred_df.sort_values(
        "abs_error",
        ascending=False
    ).head(15)
)


# ------------------------------------------------------------
# 6. NEGATIVE PREDICTIONS
# ------------------------------------------------------------

negative_predictions = (
    pred_df["prediction"] < 0
).sum()

print(
    "\nNegative predictions:",
    negative_predictions
)

# Predictions above 100%
above_100 = (
    pred_df["prediction"] > 100
).sum()

print(
    "Predictions >100%:",
    above_100
)


# ------------------------------------------------------------
# 7. ERROR BY SEVERITY BAND
# ------------------------------------------------------------

pred_df["severity_band"] = pd.cut(
    pred_df["severity_percent"],
    bins=[
        -0.001,
        0,
        10,
        20,
        40,
        60,
        100
    ],
    labels=[
        "Healthy (0)",
        "0–10",
        "10–20",
        "20–40",
        "40–60",
        "60+"
    ]
)

band_summary = (
    pred_df
    .groupby(
        "severity_band",
        observed=False
    )
    .agg(
        n=("severity_percent", "size"),
        MAE=("abs_error", "mean"),
        mean_error=("error", "mean")
    )
)

print("\n" + "=" * 80)
print("ERROR BY SEVERITY BAND")
print("=" * 80)

display(band_summary)
```

```python
# ============================================================
# NEAR-DUPLICATE / CROSS-SPLIT IMAGE AUDIT
# ============================================================

import os
import numpy as np
import pandas as pd

from PIL import Image
from pathlib import Path

from imagehash import phash


# ------------------------------------------------------------
# PATHS
# ------------------------------------------------------------

BASE_DIR = "/content/Potato_Synthetic_Environment_FINAL"

CSV_PATH = os.path.join(
    BASE_DIR,
    "synthetic_environment_image_level.csv"
)

IMAGE_ROOT = "/content/Potato_Severity_Dataset_FINAL"


# ------------------------------------------------------------
# LOAD ONE ROW PER IMAGE
# ------------------------------------------------------------

full_df = pd.read_csv(CSV_PATH)

df = (
    full_df
    .sort_values(["image_id", "candidate_id"])
    .groupby("image_id", as_index=False)
    .first()
)

print("Unique images:", len(df))


# ------------------------------------------------------------
# INDEX FILES
# ------------------------------------------------------------

file_index = {}

extensions = {
    ".jpg",
    ".jpeg",
    ".png",
    ".JPG",
    ".JPEG",
    ".PNG"
}

for p in Path(IMAGE_ROOT).rglob("*"):

    if p.is_file() and p.suffix in extensions:
        file_index[p.name] = str(p)

df["image_path"] = df["filename"].map(file_index)

missing = df["image_path"].isna().sum()

print("Missing images:", missing)

if missing:
    raise FileNotFoundError(
        "Some images cannot be located."
    )


# ------------------------------------------------------------
# COMPUTE PERCEPTUAL HASH
# ------------------------------------------------------------

print("\nComputing pHash...")

hashes = []

for i, row in df.iterrows():

    try:

        img = Image.open(
            row["image_path"]
        ).convert("RGB")

        h = phash(img)

        hashes.append(str(h))

    except Exception as e:

        print(
            "ERROR:",
            row["filename"],
            e
        )

        hashes.append(None)


df["phash"] = hashes


# ------------------------------------------------------------
# HASH DISTANCE
# ------------------------------------------------------------

def hamming_distance(h1, h2):

    return bin(
        int(h1, 16) ^ int(h2, 16)
    ).count("1")


# ------------------------------------------------------------
# WITHIN / CROSS SPLIT PAIRS
# ------------------------------------------------------------

records = []

n = len(df)

for i in range(n):

    h1 = df.iloc[i]["phash"]

    if h1 is None:
        continue

    for j in range(i + 1, n):

        h2 = df.iloc[j]["phash"]

        if h2 is None:
            continue

        d = hamming_distance(
            h1,
            h2
        )

        # Focus on potentially similar images
        if d <= 12:

            records.append({
                "image_id_1": df.iloc[i]["image_id"],
                "image_id_2": df.iloc[j]["image_id"],

                "filename_1": df.iloc[i]["filename"],
                "filename_2": df.iloc[j]["filename"],

                "split_1": df.iloc[i]["split"],
                "split_2": df.iloc[j]["split"],

                "severity_1": df.iloc[i]["severity_percent"],
                "severity_2": df.iloc[j]["severity_percent"],

                "phash_distance": d
            })


pairs = pd.DataFrame(records)

print("\nPotentially similar pairs:",
      len(pairs))


# ------------------------------------------------------------
# CROSS-SPLIT ONLY
# ------------------------------------------------------------

if len(pairs):

    cross_split = pairs[
        pairs["split_1"] != pairs["split_2"]
    ].copy()

else:

    cross_split = pd.DataFrame()


print(
    "Potential cross-split near-duplicates:",
    len(cross_split)
)


# ------------------------------------------------------------
# SHOW RESULTS
# ------------------------------------------------------------

if len(cross_split):

    print("\n" + "=" * 80)
    print("CROSS-SPLIT POTENTIAL NEAR-DUPLICATES")
    print("=" * 80)

    display(
        cross_split.sort_values(
            "phash_distance"
        ).head(50)
    )

else:

    print(
        "\n✓ No cross-split pairs found "
        "within pHash distance ≤ 12."
    )


# ------------------------------------------------------------
# SAVE
# ------------------------------------------------------------

OUT_PATH = os.path.join(
    BASE_DIR,
    "AUDIT_cross_split_near_duplicates.csv"
)

cross_split.to_csv(
    OUT_PATH,
    index=False
)

print("\nSaved:")
print(OUT_PATH)
```
