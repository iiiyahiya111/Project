```python

# ============================================================
# METHOD 2: POTATO LEAF SEGMENTATION V5
# Shadow-Aware Lesion-Guided GrabCut
#
# Main improvements over V4:
#   1. Better shadow-component aggregation
#   2. Original-shadow evidence retained after morphology
#   3. Thicker boundary support region
#   4. Deep-shadow fallback using distance transform
#   5. Safer shadow removal
#   6. Lesion-guided GrabCut retained
#   7. Contour cleanup
#   8. Detailed debug statistics
#   9. Severity calculated from lesion ∩ leaf
# ============================================================


# ============================================================
# STEP 0 — GOOGLE DRIVE + ZIP EXTRACTION
# ============================================================

from google.colab import drive
drive.mount('/content/drive')

from pathlib import Path
import zipfile
import os

ZIP_PATH = Path("/content/drive/MyDrive/Potato_Severity_Dataset.zip")
EXTRACT_DIR = Path("/content/Potato_Severity_Dataset")

if not ZIP_PATH.exists():
    raise FileNotFoundError(
        f"ZIP file not found:\n{ZIP_PATH}"
    )

print("ZIP found:", ZIP_PATH)
print(
    f"ZIP size: "
    f"{ZIP_PATH.stat().st_size / (1024**3):.2f} GB"
)

EXTRACT_DIR.mkdir(parents=True, exist_ok=True)

print("\nExtracting dataset...")

with zipfile.ZipFile(ZIP_PATH, "r") as z:
    z.extractall(EXTRACT_DIR)

print("Extraction completed.")
print("Dataset root:", EXTRACT_DIR)


# ============================================================
# STEP 1 — IMPORTS
# ============================================================

import cv2
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from pathlib import Path
import math
import warnings

warnings.filterwarnings("ignore")


# ============================================================
# STEP 2 — PATH CONFIGURATION
# ============================================================

ROOT = Path("/content/Potato_Severity_Dataset")

VALID_IMAGES_DIR = ROOT / "valid" / "images"
VALID_MASKS_DIR  = ROOT / "valid" / "lesion_masks"

OUTPUT_DIR = ROOT / "leaf_segmentation_v5"

MASK_DIR      = OUTPUT_DIR / "leaf_masks"
OVERLAY_DIR   = OUTPUT_DIR / "overlays"
COMBINED_DIR  = OUTPUT_DIR / "combined"
DEBUG_DIR     = OUTPUT_DIR / "debug"
BEFORE_DIR    = DEBUG_DIR / "before_shadow"
SHADOW_DIR    = DEBUG_DIR / "shadow"
FINAL_DIR     = DEBUG_DIR / "final"

for d in [
    OUTPUT_DIR,
    MASK_DIR,
    OVERLAY_DIR,
    COMBINED_DIR,
    DEBUG_DIR,
    BEFORE_DIR,
    SHADOW_DIR,
    FINAL_DIR,
]:
    d.mkdir(parents=True, exist_ok=True)


# ============================================================
# STEP 3 — GENERAL PARAMETERS
# ============================================================

# --------------------------
# Image / GrabCut
# --------------------------

BORDER_SIZE = 8
GRABCUT_ITERATIONS = 7

MIN_COMPONENT_AREA_RATIO = 0.003

# --------------------------
# Component scoring
# --------------------------

W_LESION = 0.45
W_COLOR  = 0.25
W_CENTER = 0.15
W_BORDER = 0.15

# --------------------------
# QC
# --------------------------

MIN_LEAF_AREA_RATIO = 0.03
MAX_LEAF_AREA_RATIO = 0.90

# ============================================================
# SHADOW PARAMETERS
# ============================================================

# These preserve your validated V4 thresholds.
SHADOW_V_MAX = 90
SHADOW_S_MAX = 85
SHADOW_CHROMA_MAX = 28

# Fragmentation repair
SHADOW_CLOSE_KERNEL = 7
SHADOW_CLOSE_ITERATIONS = 2

# Component gate
SHADOW_MIN_AREA = 40

# Original evidence required after morphology
SHADOW_MIN_ORIGINAL_RATIO = 0.35

# Original V4 value retained
SHADOW_MIN_RATIO = 0.55

# Boundary support
SHADOW_BOUNDARY_TOUCH_RATIO = 0.02
SHADOW_BOUNDARY_DILATION_KERNEL = 7

# Deep shadow
SHADOW_MIN_DEPTH = 3.0

# Safety
MAX_SHADOW_REMOVAL_RATIO = 0.25

# ============================================================
# CONTOUR PARAMETERS
# ============================================================

MIN_FINAL_COMPONENT_RATIO = 0.003

CONTOUR_CLOSE_KERNEL = 5
CONTOUR_OPEN_KERNEL = 3

# Remove tiny isolated structures
MIN_COMPONENT_KEEP_RATIO = 0.003

# ============================================================
# PILOT
# ============================================================

# Set None -> process all
PILOT_N = None

IMAGE_EXTS = {
    ".jpg",
    ".jpeg",
    ".png",
    ".bmp",
    ".webp"
}


# ============================================================
# STEP 4 — IMAGE I/O
# ============================================================

def read_rgb(path):
    """
    Read image as RGB uint8.
    """
    img = cv2.imread(
        str(path),
        cv2.IMREAD_COLOR
    )

    if img is None:
        raise FileNotFoundError(
            f"Could not read image:\n{path}"
        )

    return cv2.cvtColor(
        img,
        cv2.COLOR_BGR2RGB
    )


def read_binary_mask(path, shape):
    """
    Read lesion mask and resize using nearest-neighbor.
    """
    mask = cv2.imread(
        str(path),
        cv2.IMREAD_GRAYSCALE
    )

    if mask is None:
        raise FileNotFoundError(
            f"Could not read mask:\n{path}"
        )

    if mask.shape != shape:
        mask = cv2.resize(
            mask,
            (shape[1], shape[0]),
            interpolation=cv2.INTER_NEAREST
        )

    return (
        (mask > 127)
        .astype(np.uint8)
        * 255
    )


def infer_class(path):
    text = str(path).lower()

    if "healthy" in text:
        return "Potato Healthy"

    if (
        "late_blight" in text
        or "late-blight" in text
        or "lateblight" in text
    ):
        return "Potato Late Blight"

    return "Unknown"


def has_lesion(mask):
    if mask is None:
        return False

    return bool(
        np.any(mask > 0)
    )


# ============================================================
# STEP 5 — COLOR CANDIDATE
# ============================================================

def green_candidate(rgb):

    hsv = cv2.cvtColor(
        rgb,
        cv2.COLOR_RGB2HSV
    )

    lab = cv2.cvtColor(
        rgb,
        cv2.COLOR_RGB2LAB
    )

    h, s, v = cv2.split(hsv)
    L, a, b = cv2.split(lab)

    condition_1 = (
        (h >= 28) &
        (h <= 92) &
        (s >= 50) &
        (v >= 35)
    )

    condition_2 = (
        (a <= 140) &
        (s >= 45) &
        (v >= 35)
    )

    candidate = (
        condition_1 |
        condition_2
    )

    mask = (
        candidate
        .astype(np.uint8)
        * 255
    )

    open_kernel = cv2.getStructuringElement(
        cv2.MORPH_ELLIPSE,
        (3, 3)
    )

    close_kernel = cv2.getStructuringElement(
        cv2.MORPH_ELLIPSE,
        (5, 5)
    )

    mask = cv2.morphologyEx(
        mask,
        cv2.MORPH_OPEN,
        open_kernel,
        iterations=1
    )

    mask = cv2.morphologyEx(
        mask,
        cv2.MORPH_CLOSE,
        close_kernel,
        iterations=1
    )

    return mask


# ============================================================
# STEP 6 — GENERIC LEAF CANDIDATE
# ============================================================

def generic_leaf_candidate(rgb):

    h_img, w_img = rgb.shape[:2]

    lab = cv2.cvtColor(
        rgb,
        cv2.COLOR_RGB2LAB
    )

    hsv = cv2.cvtColor(
        rgb,
        cv2.COLOR_RGB2HSV
    )

    L, A, B = cv2.split(lab)
    H, S, V = cv2.split(hsv)

    # --------------------------------------------------------
    # Border
    # --------------------------------------------------------

    b = BORDER_SIZE

    border_mask = np.zeros(
        (h_img, w_img),
        dtype=bool
    )

    border_mask[:b, :] = True
    border_mask[-b:, :] = True
    border_mask[:, :b] = True
    border_mask[:, -b:] = True

    border_lab = lab[
        border_mask
    ].astype(np.float32)

    median_lab = np.median(
        border_lab,
        axis=0
    )

    std_lab = (
        np.std(
            border_lab,
            axis=0
        ) + 5.0
    )

    diff_L = (
        L.astype(np.float32)
        - median_lab[0]
    ) / std_lab[0]

    diff_A = (
        A.astype(np.float32)
        - median_lab[1]
    ) / std_lab[1]

    diff_B = (
        B.astype(np.float32)
        - median_lab[2]
    ) / std_lab[2]

    color_distance = np.sqrt(
        diff_L**2 +
        diff_A**2 +
        diff_B**2
    )

    # --------------------------------------------------------
    # Existing V4 evidence
    # --------------------------------------------------------

    dark_difference = (
        median_lab[0]
        - L.astype(np.float32)
    )

    dark_leaf = (
        dark_difference > 10
    )

    saturated = (
        S >= 35
    )

    strong_difference = (
        color_distance > 2.0
    )

    candidate = (
        strong_difference &
        (
            dark_leaf |
            saturated
        )
    )

    # Existing green candidate
    green = green_candidate(rgb)

    candidate = (
        candidate |
        (green > 0)
    )

    # Remove image border
    candidate[:b, :] = False
    candidate[-b:, :] = False
    candidate[:, :b] = False
    candidate[:, -b:] = False

    mask = (
        candidate
        .astype(np.uint8)
        * 255
    )

    # Morphology
    open_kernel = cv2.getStructuringElement(
        cv2.MORPH_ELLIPSE,
        (3, 3)
    )

    close_kernel = cv2.getStructuringElement(
        cv2.MORPH_ELLIPSE,
        (5, 5)
    )

    mask = cv2.morphologyEx(
        mask,
        cv2.MORPH_OPEN,
        open_kernel,
        iterations=1
    )

    mask = cv2.morphologyEx(
        mask,
        cv2.MORPH_CLOSE,
        close_kernel,
        iterations=2
    )

    return mask


# ============================================================
# STEP 7 — CENTER SCORE
# ============================================================

def center_score(component):

    ys, xs = np.where(
        component > 0
    )

    if len(xs) == 0:
        return 0.0

    h, w = component.shape

    cx = xs.mean()
    cy = ys.mean()

    center_x = (
        (w - 1) / 2.0
    )

    center_y = (
        (h - 1) / 2.0
    )

    dx = (
        (cx - center_x)
        / (w / 2.0)
    )

    dy = (
        (cy - center_y)
        / (h / 2.0)
    )

    distance = np.sqrt(
        dx**2 + dy**2
    )

    return float(
        1.0 - min(
            distance,
            1.0
        )
    )


# ============================================================
# STEP 8 — BORDER CONTACT
# ============================================================

def border_contact_ratio(
    component,
    border_size=BORDER_SIZE
):

    h, w = component.shape

    border = np.zeros(
        (h, w),
        dtype=bool
    )

    border[:border_size, :] = True
    border[-border_size:, :] = True
    border[:, :border_size] = True
    border[:, -border_size:] = True

    comp = component > 0

    area = comp.sum()

    if area == 0:
        return 1.0

    contact = (
        np.logical_and(
            comp,
            border
        ).sum()
    )

    return float(
        contact / area
    )


# ============================================================
# STEP 9 — SELECT BEST LEAF COMPONENT
# ============================================================

def select_leaf_component(
    binary_mask,
    lesion_mask,
    generic_mask
):

    binary = (
        binary_mask > 0
    ).astype(np.uint8)

    n_labels, labels, stats, _ = (
        cv2.connectedComponentsWithStats(
            binary,
            connectivity=8
        )
    )

    if n_labels <= 1:
        return binary_mask.copy()

    lesion_bool = (
        lesion_mask > 0
    )

    generic_bool = (
        generic_mask > 0
    )

    total_pixels = binary.size

    lesion_total = lesion_bool.sum()

    best_label = None
    best_score = -999999.0

    for label in range(
        1,
        n_labels
    ):

        area = stats[
            label,
            cv2.CC_STAT_AREA
        ]

        if (
            area <
            total_pixels *
            MIN_COMPONENT_AREA_RATIO
        ):
            continue

        component = (
            labels == label
        )

        # ----------------------------------------------------
        # Lesion coverage
        # ----------------------------------------------------

        if lesion_total > 0:

            lesion_inside = (
                np.logical_and(
                    component,
                    lesion_bool
                ).sum()
            )

            lesion_coverage = (
                lesion_inside /
                lesion_total
            )

        else:
            lesion_coverage = 0.0

        # ----------------------------------------------------
        # Generic candidate fraction
        # ----------------------------------------------------

        generic_inside = (
            np.logical_and(
                component,
                generic_bool
            ).sum()
        )

        generic_fraction = (
            generic_inside /
            area
        )

        # ----------------------------------------------------
        # Geometry
        # ----------------------------------------------------

        c_score = center_score(
            component.astype(
                np.uint8
            )
        )

        b_score = border_contact_ratio(
            component.astype(
                np.uint8
            )
        )

        # ----------------------------------------------------
        # Final score
        # ----------------------------------------------------

        if lesion_total > 0:

            score = (
                W_LESION *
                lesion_coverage
                +
                W_COLOR *
                generic_fraction
                +
                W_CENTER *
                c_score
                -
                W_BORDER *
                b_score
            )

        else:

            score = (
                0.60 *
                generic_fraction
                +
                0.25 *
                c_score
                -
                0.15 *
                b_score
            )

        if score > best_score:

            best_score = score
            best_label = label

    # fallback
    if best_label is None:

        areas = stats[
            1:,
            cv2.CC_STAT_AREA
        ]

        best_label = (
            1 +
            np.argmax(areas)
        )

    return (
        labels == best_label
    ).astype(np.uint8) * 255


# ============================================================
# STEP 10 — SHADOW DETECTION
# ============================================================

def detect_shadow(rgb, leaf_mask, lesion_mask=None):

    h, w = rgb.shape[:2]

    leaf = (
        leaf_mask > 0
    )

    if lesion_mask is None:

        lesion = np.zeros(
            (h, w),
            dtype=bool
        )

    else:

        lesion = (
            lesion_mask > 0
        )

    hsv = cv2.cvtColor(
        rgb,
        cv2.COLOR_RGB2HSV
    )

    lab = cv2.cvtColor(
        rgb,
        cv2.COLOR_RGB2LAB
    )

    H, S, V = cv2.split(hsv)

    L, A, B = cv2.split(lab)

    chroma = np.sqrt(
        (
            A.astype(np.float32)
            - 128.0
        ) ** 2
        +
        (
            B.astype(np.float32)
            - 128.0
        ) ** 2
    )

    # --------------------------------------------------------
    # Existing validated threshold
    # --------------------------------------------------------

    dark = (
        V <= SHADOW_V_MAX
    )

    low_sat = (
        S <= SHADOW_S_MAX
    )

    low_chroma = (
        chroma <= SHADOW_CHROMA_MAX
    )

    shadow_like = (
        dark &
        low_sat &
        low_chroma
    )

    # Only current leaf
    shadow_like &= leaf

    # Never classify known lesion as shadow
    shadow_like &= ~lesion

    return shadow_like.astype(
        np.uint8
    ) * 255


# ============================================================
# STEP 11 — SHADOW COMPONENT REFINEMENT
# ============================================================

def remove_conservative_shadow(
    rgb,
    leaf_mask,
    lesion_mask=None,
    debug=False
):

    refined = leaf_mask.copy()

    h, w = rgb.shape[:2]

    leaf = (
        refined > 0
    )

    if lesion_mask is None:

        lesion = np.zeros(
            (h, w),
            dtype=bool
        )

    else:

        lesion = (
            lesion_mask > 0
        )

    original_leaf_area = int(
        leaf.sum()
    )

    if original_leaf_area == 0:
        return refined, None

    # ========================================================
    # ORIGINAL SHADOW DETECTION
    # ========================================================

    shadow_original_u8 = detect_shadow(
        rgb,
        refined,
        lesion_mask
    )

    shadow_original = (
        shadow_original_u8 > 0
    )

    original_shadow_pixels = int(
        shadow_original.sum()
    )

    if original_shadow_pixels == 0:

        debug_info = {
            "original_shadow_pixels": 0,
            "closed_shadow_pixels": 0,
            "components_total": 0,
            "components_removed": 0,
            "removed_pixels": 0,
        }

        return refined, debug_info

    # ========================================================
    # FIX 1 — MERGE FRAGMENTED SHADOW
    # ========================================================

    close_kernel = cv2.getStructuringElement(
        cv2.MORPH_ELLIPSE,
        (
            SHADOW_CLOSE_KERNEL,
            SHADOW_CLOSE_KERNEL
        )
    )

    shadow_closed_u8 = cv2.morphologyEx(
        shadow_original_u8,
        cv2.MORPH_CLOSE,
        close_kernel,
        iterations=SHADOW_CLOSE_ITERATIONS
    )

    shadow_closed = (
        shadow_closed_u8 > 0
    )

    # Keep it inside leaf
    shadow_closed &= leaf

    # Never cross lesion
    shadow_closed &= ~lesion

    # ========================================================
    # FIX 2 — THICKER BOUNDARY REGION
    # ========================================================

    boundary = cv2.morphologyEx(
        refined,
        cv2.MORPH_GRADIENT,
        np.ones(
            (3, 3),
            np.uint8
        )
    ) > 0

    boundary = cv2.dilate(
        boundary.astype(np.uint8),
        cv2.getStructuringElement(
            cv2.MORPH_ELLIPSE,
            (
                SHADOW_BOUNDARY_DILATION_KERNEL,
                SHADOW_BOUNDARY_DILATION_KERNEL
            )
        )
    ) > 0

    # ========================================================
    # CONNECTED COMPONENTS
    # ========================================================

    n_labels, labels, stats, _ = (
        cv2.connectedComponentsWithStats(
            shadow_closed.astype(
                np.uint8
            ),
            connectivity=8
        )
    )

    shadow_remove = np.zeros(
        (h, w),
        dtype=np.uint8
    )

    total_remove = 0
    components_removed = 0

    component_debug = []

    # ========================================================
    # COMPONENT LOOP
    # ========================================================

    for label in range(
        1,
        n_labels
    ):

        component = (
            labels == label
        )

        area = int(
            stats[
                label,
                cv2.CC_STAT_AREA
            ]
        )

        if area < SHADOW_MIN_AREA:
            continue

        # ----------------------------------------------------
        # How much of component was REALLY shadow originally?
        # ----------------------------------------------------

        original_shadow_pixels_component = int(
            np.logical_and(
                component,
                shadow_original
            ).sum()
        )

        original_shadow_ratio = (
            original_shadow_pixels_component
            / area
        )

        # ----------------------------------------------------
        # Boundary contact
        # ----------------------------------------------------

        boundary_pixels = int(
            np.logical_and(
                component,
                boundary
            ).sum()
        )

        boundary_ratio = (
            boundary_pixels /
            area
        )

        # ----------------------------------------------------
        # Distance/depth
        # ----------------------------------------------------

        component_u8 = (
            component.astype(
                np.uint8
            )
        )

        distance_map = cv2.distanceTransform(
            component_u8,
            cv2.DIST_L2,
            5
        )

        max_depth = float(
            distance_map.max()
        )

        # ----------------------------------------------------
        # Candidate decision
        # ----------------------------------------------------

        evidence_ok = (
            original_shadow_ratio
            >= SHADOW_MIN_ORIGINAL_RATIO
        )

        strong_shadow_ok = (
            original_shadow_ratio
            >= SHADOW_MIN_RATIO
        )

        boundary_ok = (
            boundary_ratio
            >= SHADOW_BOUNDARY_TOUCH_RATIO
        )

        deep_shadow_ok = (
            max_depth
            >= SHADOW_MIN_DEPTH
            and
            strong_shadow_ok
        )

        # ----------------------------------------------------
        # Main gate
        #
        # Either:
        #   1. touches expanded boundary
        # OR
        #   2. is deep enough inside current leaf
        #
        # But original shadow evidence is always required.
        # ----------------------------------------------------

        keep_as_shadow = (
            evidence_ok
            and
            (
                boundary_ok
                or
                deep_shadow_ok
            )
        )

        if keep_as_shadow:

            # ------------------------------------------------
            # Global safety limit
            # ------------------------------------------------

            if (
                total_remove + area
                <=
                original_leaf_area *
                MAX_SHADOW_REMOVAL_RATIO
            ):

                shadow_remove[
                    component
                ] = 255

                total_remove += area

                components_removed += 1

        if debug:

            component_debug.append({
                "label": label,
                "area": area,
                "original_shadow_pixels":
                    original_shadow_pixels_component,
                "original_shadow_ratio":
                    original_shadow_ratio,
                "boundary_ratio":
                    boundary_ratio,
                "max_depth":
                    max_depth,
                "evidence_ok":
                    evidence_ok,
                "boundary_ok":
                    boundary_ok,
                "deep_shadow_ok":
                    deep_shadow_ok,
                "removed":
                    bool(keep_as_shadow)
            })

    # ========================================================
    # APPLY REMOVAL
    # ========================================================

    refined[
        shadow_remove > 0
    ] = 0

    # Preserve lesion for local consistency
    refined[
        lesion
    ] = 255

    debug_info = {

        "original_shadow_pixels":
            original_shadow_pixels,

        "closed_shadow_pixels":
            int(shadow_closed.sum()),

        "shadow_ratio_before":
            (
                original_shadow_pixels /
                original_leaf_area
            ),

        "removed_pixels":
            total_remove,

        "removed_ratio":
            (
                total_remove /
                original_leaf_area
            ),

        "components_total":
            n_labels - 1,

        "components_removed":
            components_removed,

        "component_debug":
            component_debug
    }

    return refined, debug_info


# ============================================================
# STEP 12 — GRABCUT SEGMENTATION
# ============================================================

def segment_leaf(rgb, lesion_mask=None):

    h, w = rgb.shape[:2]

    bgr = cv2.cvtColor(
        rgb,
        cv2.COLOR_RGB2BGR
    )

    if lesion_mask is None:

        lesion_mask = np.zeros(
            (h, w),
            dtype=np.uint8
        )

    lesion_exists = has_lesion(
        lesion_mask
    )

    # --------------------------------------------------------
    # Generic candidate
    # --------------------------------------------------------

    generic = generic_leaf_candidate(
        rgb
    )

    generic_bool = (
        generic > 0
    )

    # --------------------------------------------------------
    # Detect potential shadow BEFORE GrabCut
    # --------------------------------------------------------

    initial_shadow = detect_shadow(
        rgb,
        generic,
        lesion_mask
    )

    shadow_bool = (
        initial_shadow > 0
    )

    # --------------------------------------------------------
    # GrabCut mask
    # --------------------------------------------------------

    gc_mask = np.full(
        (h, w),
        cv2.GC_PR_BGD,
        dtype=np.uint8
    )

    b = BORDER_SIZE

    # Definite background = image border
    gc_mask[:b, :] = cv2.GC_BGD
    gc_mask[-b:, :] = cv2.GC_BGD
    gc_mask[:, :b] = cv2.GC_BGD
    gc_mask[:, -b:] = cv2.GC_BGD

    # --------------------------------------------------------
    # Generic candidate -> probable foreground
    # --------------------------------------------------------

    generic_bool[:b, :] = False
    generic_bool[-b:, :] = False
    generic_bool[:, :b] = False
    generic_bool[:, -b:] = False

    gc_mask[
        generic_bool
    ] = cv2.GC_PR_FGD

    # --------------------------------------------------------
    # Shadow -> probable background
    #
    # IMPORTANT:
    # We don't immediately call all shadow pixels definite BG.
    # They are probable background.
    # --------------------------------------------------------

    shadow_bool[:b, :] = False
    shadow_bool[-b:, :] = False
    shadow_bool[:, :b] = False
    shadow_bool[:, -b:] = False

    gc_mask[
        shadow_bool
    ] = cv2.GC_PR_BGD

    # --------------------------------------------------------
    # Lesion -> definite foreground
    # --------------------------------------------------------

    if lesion_exists:

        lesion_bool = (
            lesion_mask > 0
        )

        lesion_bool[:b, :] = False
        lesion_bool[-b:, :] = False
        lesion_bool[:, :b] = False
        lesion_bool[:, -b:] = False

        gc_mask[
            lesion_bool
        ] = cv2.GC_FGD

        # ----------------------------------------------------
        # Lesion neighbourhood = probable foreground
        # ----------------------------------------------------

        lesion_dilate = cv2.dilate(
            lesion_mask,
            cv2.getStructuringElement(
                cv2.MORPH_ELLIPSE,
                (21, 21)
            ),
            iterations=1
        ) > 0

        lesion_dilate[:b, :] = False
        lesion_dilate[-b:, :] = False
        lesion_dilate[:, :b] = False
        lesion_dilate[:, -b:] = False

        gc_mask[
            lesion_dilate &
            (~lesion_bool)
        ] = cv2.GC_PR_FGD

    # --------------------------------------------------------
    # Restore definite border background
    # --------------------------------------------------------

    gc_mask[:b, :] = cv2.GC_BGD
    gc_mask[-b:, :] = cv2.GC_BGD
    gc_mask[:, :b] = cv2.GC_BGD
    gc_mask[:, -b:] = cv2.GC_BGD

    gc_mask[:, :b] = cv2.GC_BGD
    gc_mask[:, -b:] = cv2.GC_BGD

    # --------------------------------------------------------
    # Check foreground quantity
    # --------------------------------------------------------

    fg_pixels = np.sum(
        (gc_mask == cv2.GC_FGD) |
        (gc_mask == cv2.GC_PR_FGD)
    )

    # --------------------------------------------------------
    # Fallback if too little foreground evidence
    # --------------------------------------------------------

    if fg_pixels < (
        h * w * 0.005
    ):

        lab = cv2.cvtColor(
            rgb,
            cv2.COLOR_RGB2LAB
        )

        L = lab[:, :, 0]

        center = np.zeros(
            (h, w),
            dtype=bool
        )

        x1 = int(w * 0.10)
        x2 = int(w * 0.90)

        y1 = int(h * 0.10)
        y2 = int(h * 0.90)

        center[
            y1:y2,
            x1:x2
        ] = True

        threshold = np.percentile(
            L[center],
            45
        )

        fallback = (
            L <= threshold
        ) & center

        fallback[:b, :] = False
        fallback[-b:, :] = False
        fallback[:, :b] = False
        fallback[:, -b:] = False

        gc_mask[
            fallback
        ] = cv2.GC_PR_FGD

    # --------------------------------------------------------
    # GrabCut
    # --------------------------------------------------------

    bg_model = np.zeros(
        (1, 65),
        np.float64
    )

    fg_model = np.zeros(
        (1, 65),
        np.float64
    )

    cv2.grabCut(
        bgr,
        gc_mask,
        None,
        bg_model,
        fg_model,
        GRABCUT_ITERATIONS,
        cv2.GC_INIT_WITH_MASK
    )

    # --------------------------------------------------------
    # Binary foreground
    # --------------------------------------------------------

    leaf_candidate = (
        np.isin(
            gc_mask,
            [
                cv2.GC_FGD,
                cv2.GC_PR_FGD
            ]
        ).astype(np.uint8)
        * 255
    )

    # --------------------------------------------------------
    # Basic cleanup
    # --------------------------------------------------------

    close_kernel = cv2.getStructuringElement(
        cv2.MORPH_ELLIPSE,
        (5, 5)
    )

    open_kernel = cv2.getStructuringElement(
        cv2.MORPH_ELLIPSE,
        (3, 3)
    )

    leaf_candidate = cv2.morphologyEx(
        leaf_candidate,
        cv2.MORPH_CLOSE,
        close_kernel,
        iterations=1
    )

    leaf_candidate = cv2.morphologyEx(
        leaf_candidate,
        cv2.MORPH_OPEN,
        open_kernel,
        iterations=1
    )

    # --------------------------------------------------------
    # Select main leaf component
    # --------------------------------------------------------

    leaf = select_leaf_component(
        leaf_candidate,
        lesion_mask,
        generic
    )

    # IMPORTANT:
    # We do NOT force lesion into the leaf here.
    # Lesion is used for guidance/QC.
    # This avoids artificially expanding leaf mask.
    #
    # If required for robustness, lesion-connected pixels
    # can later be considered with a controlled rule.

    return (
        leaf,
        generic,
        initial_shadow,
        gc_mask
    )


# ============================================================
# STEP 13 — FINAL CONTOUR CLEANUP
# ============================================================

def refine_leaf_contour(
    leaf_mask,
    lesion_mask=None
):

    refined = leaf_mask.copy()

    h, w = refined.shape

    # --------------------------------------------------------
    # Morphological cleanup
    # --------------------------------------------------------

    close_kernel = cv2.getStructuringElement(
        cv2.MORPH_ELLIPSE,
        (
            CONTOUR_CLOSE_KERNEL,
            CONTOUR_CLOSE_KERNEL
        )
    )

    open_kernel = cv2.getStructuringElement(
        cv2.MORPH_ELLIPSE,
        (
            CONTOUR_OPEN_KERNEL,
            CONTOUR_OPEN_KERNEL
        )
    )

    refined = cv2.morphologyEx(
        refined,
        cv2.MORPH_CLOSE,
        close_kernel,
        iterations=1
    )

    refined = cv2.morphologyEx(
        refined,
        cv2.MORPH_OPEN,
        open_kernel,
        iterations=1
    )

    # --------------------------------------------------------
    # Keep strongest component
    # --------------------------------------------------------

    binary = (
        refined > 0
    ).astype(np.uint8)

    n_labels, labels, stats, _ = (
        cv2.connectedComponentsWithStats(
            binary,
            connectivity=8
        )
    )

    if n_labels <= 1:
        return refined

    total_pixels = h * w

    best_label = None
    best_score = -1e9

    lesion_bool = (
        lesion_mask > 0
        if lesion_mask is not None
        else np.zeros(
            (h, w),
            dtype=bool
        )
    )

    lesion_total = lesion_bool.sum()

    for label in range(
        1,
        n_labels
    ):

        area = int(
            stats[
                label,
                cv2.CC_STAT_AREA
            ]
        )

        if (
            area <
            total_pixels *
            MIN_COMPONENT_KEEP_RATIO
        ):
            continue

        comp = (
            labels == label
        )

        # center
        c_score = center_score(
            comp.astype(
                np.uint8
            )
        )

        # lesion support
        if lesion_total > 0:

            lesion_inside = (
                np.logical_and(
                    comp,
                    lesion_bool
                ).sum()
            )

            lesion_support = (
                lesion_inside /
                lesion_total
            )

        else:

            lesion_support = 0.0

        # Slight preference for larger object
        area_score = (
            area /
            float(total_pixels)
        )

        score = (
            0.30 * c_score
            +
            0.45 * lesion_support
            +
            0.25 * min(
                area_score * 5.0,
                1.0
            )
        )

        if score > best_score:

            best_score = score
            best_label = label

    if best_label is not None:

        refined = (
            labels == best_label
        ).astype(np.uint8) * 255

    return refined


# ============================================================
# STEP 14 — OVERLAY
# ============================================================

def make_overlay(
    rgb,
    leaf_mask,
    lesion_mask=None
):

    overlay = rgb.copy()

    # Leaf boundary
    boundary = cv2.morphologyEx(
        leaf_mask,
        cv2.MORPH_GRADIENT,
        np.ones(
            (3, 3),
            np.uint8
        )
    )

    overlay[
        boundary > 0
    ] = [0, 255, 0]

    # Lesion
    if lesion_mask is not None:

        lesion = (
            lesion_mask > 0
        )

        lesion_overlay = (
            overlay.copy()
        )

        lesion_overlay[
            lesion
        ] = [255, 0, 0]

        overlay = cv2.addWeighted(
            overlay,
            0.70,
            lesion_overlay,
            0.30,
            0
        )

        # redraw leaf boundary
        overlay[
            boundary > 0
        ] = [0, 255, 0]

    return overlay


# ============================================================
# STEP 15 — QC + SEVERITY
# ============================================================

def calculate_qc(
    rgb,
    lesion_mask,
    leaf_mask
):

    h, w = rgb.shape[:2]

    total = h * w

    lesion = (
        lesion_mask > 0
    )

    leaf = (
        leaf_mask > 0
    )

    lesion_pixels = int(
        lesion.sum()
    )

    leaf_pixels = int(
        leaf.sum()
    )

    # --------------------------------------------------------
    # Lesion outside leaf
    # --------------------------------------------------------

    lesion_outside = int(
        np.logical_and(
            lesion,
            ~leaf
        ).sum()
    )

    # --------------------------------------------------------
    # Lesion inside leaf
    # --------------------------------------------------------

    lesion_inside = int(
        np.logical_and(
            lesion,
            leaf
        ).sum()
    )

    # --------------------------------------------------------
    # Ratios
    # --------------------------------------------------------

    leaf_ratio = (
        leaf_pixels / total
        if total > 0
        else np.nan
    )

    lesion_ratio = (
        lesion_pixels / total
        if total > 0
        else np.nan
    )

    # --------------------------------------------------------
    # Severity
    #
    # IMPORTANT:
    # Only lesion pixels inside leaf count.
    # --------------------------------------------------------

    severity = (
        100.0 *
        lesion_inside /
        leaf_pixels
        if leaf_pixels > 0
        else np.nan
    )

    # --------------------------------------------------------
    # Border
    # --------------------------------------------------------

    b = BORDER_SIZE

    border = np.zeros(
        (h, w),
        dtype=bool
    )

    border[:b, :] = True
    border[-b:, :] = True
    border[:, :b] = True
    border[:, -b:] = True

    border_touch = bool(
        np.logical_and(
            leaf,
            border
        ).sum() > 0
    )

    # --------------------------------------------------------
    # QC flags
    # --------------------------------------------------------

    flags = []

    if leaf_pixels == 0:
        flags.append(
            "empty_leaf_mask"
        )

    if (
        leaf_ratio <
        MIN_LEAF_AREA_RATIO
    ):
        flags.append(
            "leaf_too_small"
        )

    if (
        leaf_ratio >
        MAX_LEAF_AREA_RATIO
    ):
        flags.append(
            "leaf_too_large"
        )

    if lesion_outside > 0:
        flags.append(
            "lesion_outside_leaf"
        )

    if border_touch:
        flags.append(
            "leaf_touches_border"
        )

    status = (
        "OK"
        if len(flags) == 0
        else "CHECK"
    )

    return {

        "leaf_pixels":
            leaf_pixels,

        "lesion_pixels":
            lesion_pixels,

        "lesion_inside_leaf":
            lesion_inside,

        "leaf_area_ratio":
            leaf_ratio,

        "lesion_area_ratio":
            lesion_ratio,

        "lesion_outside_leaf":
            lesion_outside,

        "severity_percent":
            severity,

        "leaf_touches_border":
            border_touch,

        "qc_status":
            status,

        "qc_flags":
            ";".join(flags)
    }


# ============================================================
# STEP 16 — BUILD IMAGE/MASK RECORDS
# ============================================================

def build_records(
    image_dir,
    mask_dir
):

    records = []

    for image_path in sorted(
        image_dir.rglob("*")
    ):

        if not image_path.is_file():
            continue

        if (
            image_path.suffix.lower()
            not in IMAGE_EXTS
        ):
            continue

        possible_masks = [

            mask_dir /
            f"{image_path.stem}.png",

            mask_dir /
            f"{image_path.stem}.jpg",

            mask_dir /
            f"{image_path.stem}.jpeg",

            mask_dir /
            f"{image_path.stem}.bmp",
        ]

        mask_path = None

        for p in possible_masks:

            if p.exists():

                mask_path = p
                break

        if mask_path is None:
            continue

        records.append({

            "image_path":
                str(image_path),

            "mask_path":
                str(mask_path),

            "filename":
                image_path.name,

            "class_name":
                infer_class(
                    image_path
                )
        })

    return pd.DataFrame(
        records
    )


validation_df = build_records(
    VALID_IMAGES_DIR,
    VALID_MASKS_DIR
)

if validation_df.empty:

    raise ValueError(
        "No image-mask pairs found."
    )

print(
    "=============================================="
)

print(
    "POTATO LEAF SEGMENTATION V5"
)

print(
    "Shadow-Aware Lesion-Guided GrabCut"
)

print(
    "=============================================="
)

print(
    f"\nFound image-mask pairs: "
    f"{len(validation_df)}"
)

print(
    "\nClass distribution:"
)

print(
    validation_df[
        "class_name"
    ].value_counts(
        dropna=False
    )
)


# ============================================================
# STEP 17 — FULL DATASET / SPLIT CONFIGURATION
# ============================================================

PILOT_N = None

SPLITS = ["train", "valid", "test"]
all_results = []

# Keep every split visually/data-wise separate so identical filenames
# across train/valid/test cannot overwrite each other.
SPLIT_DIRS = {}
for split in SPLITS:
    split_root = OUTPUT_DIR / split
    split_dirs = {
        "root": split_root,
        "mask": split_root / "leaf_masks",
        "overlay": split_root / "overlays",
        "combined": split_root / "combined",
        "debug": split_root / "debug",
        "before": split_root / "debug" / "before_shadow",
        "shadow": split_root / "debug" / "shadow",
        "final": split_root / "debug" / "final",
    }
    for d in split_dirs.values():
        if isinstance(d, Path):
            d.mkdir(parents=True, exist_ok=True)
    SPLIT_DIRS[split] = split_dirs


# ============================================================
# STEP 18 — PROCESS ONE SPLIT
# ============================================================

def process_split(split, images_dir, masks_dir):

    df = build_records(images_dir, masks_dir)
    print("\n" + "=" * 70)
    print(f"SPLIT: {split}")
    print(f"Image-mask pairs: {len(df)}")
    print("=" * 70)

    if df.empty:
        return []

    if PILOT_N is not None:
        selected = []

        for class_name, group in df.groupby("class_name", sort=False):
            proportion = len(group) / len(df)
            n_take = max(1, round(PILOT_N * proportion))
            n_take = min(n_take, len(group))
            selected.append(
                group.sample(n=n_take, random_state=42)
            )

        df = pd.concat(selected, ignore_index=True)

        if len(df) > PILOT_N:
            df = (
                df.sample(n=PILOT_N, random_state=42)
                .reset_index(drop=True)
            )

    print(f"Processing: {len(df)} images")

    split_dirs = SPLIT_DIRS[split]
    results = []

    for idx, row in df.iterrows():

        image_path = Path(row["image_path"])
        mask_path = Path(row["mask_path"])

        print("\n" + "-" * 70)
        print(f"[{idx + 1}/{len(df)}] {image_path.name}")

        try:

            # ----------------------------------------------------
            # Read
            # ----------------------------------------------------
            rgb = read_rgb(image_path)

            lesion_mask = read_binary_mask(
                mask_path,
                rgb.shape[:2]
            )

            # ----------------------------------------------------
            # GrabCut
            # ----------------------------------------------------
            (
                leaf_before_shadow,
                generic_candidate,
                initial_shadow,
                grabcut_debug
            ) = segment_leaf(
                rgb,
                lesion_mask
            )

            # ----------------------------------------------------
            # Shadow refinement
            # ----------------------------------------------------
            (
                leaf_after_shadow,
                shadow_debug
            ) = remove_conservative_shadow(
                rgb,
                leaf_before_shadow,
                lesion_mask,
                debug=True
            )

            # ----------------------------------------------------
            # Contour refinement
            # ----------------------------------------------------
            leaf_final = refine_leaf_contour(
                leaf_after_shadow,
                lesion_mask
            )

            # ----------------------------------------------------
            # QC + severity
            # ----------------------------------------------------
            qc = calculate_qc(
                rgb,
                lesion_mask,
                leaf_final
            )

            # ----------------------------------------------------
            # Overlay
            # ----------------------------------------------------
            overlay = make_overlay(
                rgb,
                leaf_final,
                lesion_mask
            )

            # ----------------------------------------------------
            # Save final leaf mask
            # ----------------------------------------------------
            leaf_mask_path = (
                split_dirs["mask"] /
                f"{image_path.stem}_leaf.png"
            )
            cv2.imwrite(
                str(leaf_mask_path),
                leaf_final
            )

            # ----------------------------------------------------
            # Save overlay
            # ----------------------------------------------------
            overlay_path = (
                split_dirs["overlay"] /
                f"{image_path.stem}_overlay.jpg"
            )
            cv2.imwrite(
                str(overlay_path),
                cv2.cvtColor(
                    overlay,
                    cv2.COLOR_RGB2BGR
                )
            )

            # ----------------------------------------------------
            # Save before-shadow mask
            # ----------------------------------------------------
            before_path = (
                split_dirs["before"] /
                f"{image_path.stem}_before.png"
            )
            cv2.imwrite(
                str(before_path),
                leaf_before_shadow
            )

            # ----------------------------------------------------
            # Save detected shadow mask
            # ----------------------------------------------------
            shadow_path = (
                split_dirs["shadow"] /
                f"{image_path.stem}_shadow.png"
            )
            cv2.imwrite(
                str(shadow_path),
                initial_shadow
            )

            # ----------------------------------------------------
            # Save final debug mask
            # ----------------------------------------------------
            final_debug_path = (
                split_dirs["final"] /
                f"{image_path.stem}_final.png"
            )
            cv2.imwrite(
                str(final_debug_path),
                leaf_final
            )

            # ----------------------------------------------------
            # Save combined visual panel
            # ----------------------------------------------------
            fig, axes = plt.subplots(
                1,
                5,
                figsize=(22, 5)
            )

            axes[0].imshow(rgb)
            axes[0].set_title("Original")

            axes[1].imshow(
                leaf_before_shadow,
                cmap="gray"
            )
            axes[1].set_title("Before Shadow")

            axes[2].imshow(
                initial_shadow,
                cmap="gray"
            )
            axes[2].set_title("Detected Shadow")

            axes[3].imshow(
                leaf_final,
                cmap="gray"
            )
            axes[3].set_title("Final Leaf")

            axes[4].imshow(overlay)
            axes[4].set_title(
                f"Severity={qc['severity_percent']:.2f}%"
            )

            for ax in axes:
                ax.axis("off")

            plt.tight_layout()

            combined_path = (
                split_dirs["combined"] /
                f"{image_path.stem}_combined.jpg"
            )

            plt.savefig(
                combined_path,
                dpi=140,
                bbox_inches="tight"
            )
            plt.close(fig)

            # ----------------------------------------------------
            # Record
            # ----------------------------------------------------
            record = {
                "split": split,
                "filename": image_path.name,
                "class_name": row["class_name"],
                "image_path": str(image_path),
                "lesion_mask_path": str(mask_path),
                "leaf_mask_path": str(leaf_mask_path),
                "overlay_path": str(overlay_path),
                "combined_path": str(combined_path),
                "severity_percent": qc["severity_percent"],
                "leaf_pixels": qc["leaf_pixels"],
                "lesion_pixels": qc["lesion_pixels"],
                "lesion_inside_leaf": qc["lesion_inside_leaf"],
                "lesion_outside_leaf": qc["lesion_outside_leaf"],
                "leaf_area_ratio": qc["leaf_area_ratio"],
                "shadow_pixels_detected": (
                    shadow_debug["original_shadow_pixels"]
                    if shadow_debug is not None else 0
                ),
                "shadow_pixels_after_close": (
                    shadow_debug["closed_shadow_pixels"]
                    if shadow_debug is not None else 0
                ),
                "shadow_pixels_removed": (
                    shadow_debug["removed_pixels"]
                    if shadow_debug is not None else 0
                ),
                "shadow_removed_ratio": (
                    shadow_debug["removed_ratio"]
                    if shadow_debug is not None else 0.0
                ),
                "shadow_components_total": (
                    shadow_debug["components_total"]
                    if shadow_debug is not None else 0
                ),
                "shadow_components_removed": (
                    shadow_debug["components_removed"]
                    if shadow_debug is not None else 0
                ),
                "qc_status": qc["qc_status"],
                "qc_flags": qc["qc_flags"]
            }

            results.append(record)

            print(
                f"Leaf area: {qc['leaf_area_ratio'] * 100:.2f}%"
            )
            print(
                f"Shadow detected: "
                f"{record['shadow_pixels_detected']} px"
            )
            print(
                f"Shadow removed: "
                f"{record['shadow_pixels_removed']} px"
            )
            print(
                f"Severity: {qc['severity_percent']:.2f}%"
            )
            print(f"QC: {qc['qc_status']}")

            if qc["qc_flags"]:
                print("Flags:", qc["qc_flags"])

        except Exception as e:

            print("ERROR:", repr(e))

            results.append({
                "split": split,
                "filename": image_path.name,
                "class_name": row["class_name"],
                "image_path": str(image_path),
                "lesion_mask_path": str(mask_path),
                "qc_status": "ERROR",
                "qc_flags": str(e)
            })

    return results


# ============================================================
# STEP 19 — RUN ALL SPLITS
# ============================================================

for split in SPLITS:

    images_dir = ROOT / split / "images"
    masks_dir = ROOT / split / "lesion_masks"

    if not images_dir.exists():
        print(f"Skipping {split} (folder not found): {images_dir}")
        continue

    split_results = process_split(
        split,
        images_dir,
        masks_dir
    )

    all_results.extend(split_results)


# ============================================================
# STEP 20 — RESULTS DATAFRAME + MASTER CSV
# ============================================================

results_df = pd.DataFrame(all_results)

CSV_PATH = (
    OUTPUT_DIR /
    "leaf_segmentation_v5_ALL_SPLITS.csv"
)

results_df.to_csv(
    CSV_PATH,
    index=False
)

print("\n" + "=" * 70)
print("V5 FULL DATASET PROCESSING COMPLETED")
print("=" * 70)
print("Results saved to:")
print(CSV_PATH)
print("Result shape:", results_df.shape)


# ============================================================
# STEP 21 — SUMMARY STATISTICS
# ============================================================

if not results_df.empty:

    numeric_cols = [
        "severity_percent",
        "leaf_area_ratio",
        "shadow_pixels_detected",
        "shadow_pixels_removed",
        "shadow_removed_ratio",
    ]

    available_numeric = [
        c for c in numeric_cols
        if c in results_df.columns
    ]

    print("\nOverall summary statistics:")
    display(
        results_df[available_numeric].describe()
    )

    print("\nSplit-wise summary:")
    split_summary = (
        results_df
        .groupby("split")
        .agg(
            images=("filename", "count"),
            qc_check=("qc_status", lambda s: int((s == "CHECK").sum())),
            qc_error=("qc_status", lambda s: int((s == "ERROR").sum())),
            mean_severity=("severity_percent", "mean"),
            mean_leaf_area=("leaf_area_ratio", "mean"),
            mean_shadow_removed_ratio=("shadow_removed_ratio", "mean"),
        )
        .reset_index()
    )

    display(split_summary)

    print("\nQC distribution by split:")
    print(
        results_df
        .groupby("split")["qc_status"]
        .value_counts()
    )

    print("\nCHECK-flagged images:")
    check_df = results_df[
        results_df.qc_status == "CHECK"
    ][["filename", "split", "qc_flags"]]

    display(check_df)

    print(
        f"\nCHECK count: {len(check_df)} / {len(results_df)} "
        f"({100.0 * len(check_df) / len(results_df):.2f}%)"
    )


# ============================================================
# STEP 22 — DISPLAY SAMPLE RESULTS
# ============================================================

display_cols = [
    "split",
    "filename",
    "class_name",
    "severity_percent",
    "leaf_area_ratio",
    "shadow_pixels_detected",
    "shadow_pixels_removed",
    "shadow_removed_ratio",
    "lesion_outside_leaf",
    "qc_status",
    "qc_flags"
]

display_cols = [
    c for c in display_cols
    if c in results_df.columns
]

if not results_df.empty:
    print("\nFirst results:")
    display(
        results_df[display_cols].head(20)
    )


# ============================================================
# STEP 23 — RANDOM VISUAL CHECK
# ============================================================

def show_random_results(n=6, seed=42):

    if results_df.empty:
        print("No results.")
        return

    n = min(n, len(results_df))

    sample_df = results_df.sample(
        n=n,
        random_state=seed
    )

    cols = 3
    rows = math.ceil(n / cols)

    fig, axes = plt.subplots(
        rows,
        cols,
        figsize=(18, 6 * rows)
    )

    axes = np.array(axes).reshape(-1)

    for ax, (_, row) in zip(
        axes,
        sample_df.iterrows()
    ):

        combined_path = Path(
            row.get("combined_path", "")
        )

        if combined_path.exists():
            img = cv2.imread(str(combined_path))
            img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
            ax.imshow(img)

        ax.set_title(
            f"{row['split']} | {row['filename']}"
        )
        ax.axis("off")

    for ax in axes[len(sample_df):]:
        ax.axis("off")

    plt.tight_layout()
    plt.show()


show_random_results(n=6, seed=42)


# ============================================================
# STEP 24 — SHADOW DEBUG REPORT
# ============================================================

if not results_df.empty:

    shadow_cols = [
        "split",
        "filename",
        "shadow_pixels_detected",
        "shadow_pixels_after_close",
        "shadow_pixels_removed",
        "shadow_removed_ratio",
        "shadow_components_total",
        "shadow_components_removed"
    ]

    shadow_cols = [
        c for c in shadow_cols
        if c in results_df.columns
    ]

    print("\nShadow refinement report:")
    display(
        results_df[shadow_cols]
        .sort_values(
            "shadow_removed_ratio",
            ascending=False
        )
        .head(20)
    )


# ============================================================
# STEP 25 — EXPORT PATH SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("OUTPUT DIRECTORIES")
print("=" * 70)
print("Master CSV:", CSV_PATH)

for split in SPLITS:
    if split not in SPLIT_DIRS:
        continue
    print(f"\n[{split}]")
    print("Final leaf masks :", SPLIT_DIRS[split]["mask"])
    print("Overlays         :", SPLIT_DIRS[split]["overlay"])
    print("Combined panels  :", SPLIT_DIRS[split]["combined"])
    print("Before-shadow    :", SPLIT_DIRS[split]["before"])
    print("Shadow masks     :", SPLIT_DIRS[split]["shadow"])
    print("Final debug      :", SPLIT_DIRS[split]["final"])

```

```python

# ============================================================
# V5 — SEVERITY-AWARE QC V2
# Does NOT modify the V5 segmentation masks.
# It only reclassifies the existing CSV records using:
#   1) lesion pixels outside leaf
#   2) existing V5 QC flags
#   3) optional visual-review priority
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path

# ------------------------------------------------------------
# STEP 0 — PATH
# ------------------------------------------------------------

ROOT = Path("/content/Potato_Severity_Dataset")
OUTPUT_DIR = ROOT / "leaf_segmentation_v5"

CSV_PATH = OUTPUT_DIR / "leaf_segmentation_v5_ALL_SPLITS.csv"

if not CSV_PATH.exists():
    raise FileNotFoundError(f"CSV not found:\n{CSV_PATH}")

# Proposed thresholds — easy to change later after visual review
ACCEPT_OUTSIDE_PCT_MAX = 1.0
REVIEW_OUTSIDE_PCT_MAX = 5.0

# ------------------------------------------------------------
# STEP 1 — LOAD
# ------------------------------------------------------------

df = pd.read_csv(CSV_PATH)

required_cols = [
    "filename",
    "split",
    "severity_percent",
    "leaf_pixels",
    "lesion_pixels",
    "lesion_inside_leaf",
    "lesion_outside_leaf",
    "leaf_area_ratio",
    "qc_status",
    "qc_flags",
]

missing = [c for c in required_cols if c not in df.columns]
if missing:
    raise ValueError(f"Missing required columns: {missing}")

# ------------------------------------------------------------
# STEP 2 — DERIVE LESION-OUTSIDE PERCENTAGE
# ------------------------------------------------------------

df["lesion_outside_pct"] = np.where(
    df["lesion_pixels"] > 0,
    100.0 * df["lesion_outside_leaf"] / df["lesion_pixels"],
    0.0
)

df["lesion_inside_pct"] = np.where(
    df["lesion_pixels"] > 0,
    100.0 * df["lesion_inside_leaf"] / df["lesion_pixels"],
    100.0
)

# Original V5 severity is preserved exactly.
# This is only a consistency check.
df["severity_recomputed"] = np.where(
    df["leaf_pixels"] > 0,
    100.0 * df["lesion_inside_leaf"] / df["leaf_pixels"],
    np.nan
)

df["severity_difference"] = (
    df["severity_percent"] - df["severity_recomputed"]
).abs()

# ------------------------------------------------------------
# STEP 3 — SPLIT EXISTING FLAGS
# ------------------------------------------------------------

flags = df["qc_flags"].fillna("").astype(str)

df["has_lesion_outside"] = flags.str.contains(
    "lesion_outside_leaf",
    regex=False
)

df["has_border_touch"] = flags.str.contains(
    "leaf_touches_border",
    regex=False
)

df["has_other_qc_flag"] = (
    (flags != "") &
    (~df["has_lesion_outside"]) &
    (~df["has_border_touch"])
)

# ------------------------------------------------------------
# STEP 4 — SEVERITY-AWARE QC CLASS
# ------------------------------------------------------------

def classify_row(row):
    # Hard failure / runtime failure
    if str(row["qc_status"]).upper() == "ERROR":
        return "REJECT"

    outside_pct = float(row["lesion_outside_pct"])

    # Other QC problems remain review/reject candidates.
    if bool(row["has_other_qc_flag"]):
        return "REJECT"

    # Existing border-touching cases deserve manual inspection,
    # even when lesion leakage is small.
    if bool(row["has_border_touch"]):
        if outside_pct <= ACCEPT_OUTSIDE_PCT_MAX:
            return "REVIEW_BORDER"
        elif outside_pct <= REVIEW_OUTSIDE_PCT_MAX:
            return "REVIEW"
        else:
            return "REJECT"

    # No lesion outside leaf
    if outside_pct == 0:
        return "ACCEPT"

    # Tiny annotation/contour mismatch
    if outside_pct <= ACCEPT_OUTSIDE_PCT_MAX:
        return "ACCEPT_MINOR_LEAKAGE"

    # Moderate mismatch
    if outside_pct <= REVIEW_OUTSIDE_PCT_MAX:
        return "REVIEW"

    # Large mismatch
    return "REJECT"

df["qc_v2"] = df.apply(classify_row, axis=1)

# ------------------------------------------------------------
# STEP 5 — VISUAL REVIEW PRIORITY
# ------------------------------------------------------------

# Larger lesion leakage + larger severity = higher priority.
df["review_priority_score"] = (
    df["lesion_outside_pct"].clip(lower=0, upper=100) * 2.0
    + df["severity_percent"].fillna(0).clip(lower=0, upper=100) * 0.25
)

# Force all non-ACCEPT records into review ordering.
review_df = (
    df[df["qc_v2"] != "ACCEPT"]
    .sort_values(
        ["qc_v2", "review_priority_score"],
        ascending=[True, False]
    )
    .reset_index(drop=True)
)

# ------------------------------------------------------------
# STEP 6 — OUTPUTS
# ------------------------------------------------------------

QC_DIR = OUTPUT_DIR / "qc_v2"
QC_DIR.mkdir(parents=True, exist_ok=True)

ALL_QC_CSV = QC_DIR / "v5_qc_v2_ALL.csv"
REVIEW_CSV = QC_DIR / "v5_qc_v2_REVIEW.csv"
ACCEPTED_CSV = QC_DIR / "v5_qc_v2_ACCEPTED.csv"
REJECT_CSV = QC_DIR / "v5_qc_v2_REJECT.csv"
SUMMARY_CSV = QC_DIR / "v5_qc_v2_SPLIT_SUMMARY.csv"

df.to_csv(ALL_QC_CSV, index=False)
review_df.to_csv(REVIEW_CSV, index=False)
df[df["qc_v2"].str.startswith("ACCEPT")].to_csv(ACCEPTED_CSV, index=False)
df[df["qc_v2"].isin(["REJECT"])].to_csv(REJECT_CSV, index=False)

summary = (
    df.groupby("split")
      .agg(
          images=("filename", "count"),
          accept=("qc_v2", lambda s: s.str.startswith("ACCEPT").sum()),
          review=("qc_v2", lambda s: s.str.startswith("REVIEW").sum()),
          reject=("qc_v2", lambda s: (s == "REJECT").sum()),
          mean_severity=("severity_percent", "mean"),
          mean_lesion_outside_pct=("lesion_outside_pct", "mean"),
      )
      .reset_index()
)

summary["review_plus_reject_pct"] = (
    100.0 * (summary["review"] + summary["reject"])
    / summary["images"]
)

summary.to_csv(SUMMARY_CSV, index=False)

# ------------------------------------------------------------
# STEP 7 — PRINT REPORT
# ------------------------------------------------------------

print("=" * 72)
print("V5 SEVERITY-AWARE QC V2")
print("=" * 72)

print("\nQC V2 distribution:")
print(df["qc_v2"].value_counts(dropna=False))

print("\nSplit-wise QC V2:")
print(summary.to_string(index=False))

print("\nOriginal V5 CHECK:", int((df["qc_status"] == "CHECK").sum()))
print(
    "V2 REVIEW + REJECT:",
    int((df["qc_v2"].isin(["REVIEW", "REVIEW_BORDER", "REJECT"])).sum())
)

print(
    "\nOverall REVIEW + REJECT %:",
    f"{100.0 * df['qc_v2'].isin(['REVIEW', 'REVIEW_BORDER', 'REJECT']).mean():.2f}%"
)

print("\nHighest-priority cases:")
priority_cols = [
    "filename",
    "split",
    "qc_v2",
    "lesion_outside_leaf",
    "lesion_outside_pct",
    "leaf_area_ratio",
    "severity_percent",
    "qc_flags",
]
display(
    review_df[priority_cols].head(30)
)

print("\nSaved:")
print("All QC      :", ALL_QC_CSV)
print("Review list :", REVIEW_CSV)
print("Accepted    :", ACCEPTED_CSV)
print("Reject      :", REJECT_CSV)
print("Summary     :", SUMMARY_CSV)

# ------------------------------------------------------------
# STEP 8 — OPTIONAL: CREATE A TOP-30 VISUAL REVIEW PANEL
# ------------------------------------------------------------

import cv2
import matplotlib.pyplot as plt
import math

COMBINED_ROOT = OUTPUT_DIR / "combined"
VIS_DIR = QC_DIR / "visual_review"
VIS_DIR.mkdir(parents=True, exist_ok=True)

top_n = min(30, len(review_df))

if top_n > 0:
    sample = review_df.head(top_n)

    cols = 3
    rows = math.ceil(len(sample) / cols)

    fig, axes = plt.subplots(
        rows,
        cols,
        figsize=(22, 6 * rows)
    )

    axes = np.array(axes).reshape(-1)

    for ax, (_, row) in zip(axes, sample.iterrows()):
        # Search split folder first, then fallback recursively.
        candidates = list(
            (COMBINED_ROOT / str(row["split"])).glob(
                f"{Path(row['filename']).stem}_combined.jpg"
            )
        )

        if not candidates:
            candidates = list(
                COMBINED_ROOT.rglob(
                    f"{Path(row['filename']).stem}_combined.jpg"
                )
            )

        if candidates:
            img = cv2.imread(str(candidates[0]))
            img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
            ax.imshow(img)

        ax.set_title(
            f"{row['split']} | {row['qc_v2']}\n"
            f"outside={row['lesion_outside_pct']:.2f}% | "
            f"severity={row['severity_percent']:.2f}%"
        )
        ax.axis("off")

    for ax in axes[len(sample):]:
        ax.axis("off")

    plt.tight_layout()

    panel_path = VIS_DIR / "TOP_30_QC_REVIEW.jpg"
    plt.savefig(
        panel_path,
        dpi=140,
        bbox_inches="tight"
    )
    plt.show()

    print("\nVisual review panel:", panel_path)
else:
    print("\nNo REVIEW/REJECT cases found.")
```

```python
# ============================================================
# STATIC MANUAL VISUAL QC — ROBUST VERSION
# Does NOT use combined images
# Does NOT use input()
# Does NOT interact
#
# Shows:
#   1. Original image
#   2. Lesion mask
#   3. Final leaf mask
#   4. Final overlay
#
# Only REVIEW / REVIEW_BORDER / REJECT cases
# ============================================================

import pandas as pd
import numpy as np
import cv2
import matplotlib.pyplot as plt
from pathlib import Path
import math
from IPython.display import display

# ------------------------------------------------------------
# PATHS
# ------------------------------------------------------------

ROOT = Path("/content/Potato_Severity_Dataset")
OUTPUT_DIR = ROOT / "leaf_segmentation_v5"

CSV_PATH = (
    OUTPUT_DIR
    / "qc_v2"
    / "v5_qc_v2_ALL.csv"
)

# ------------------------------------------------------------
# LOAD QC CSV
# ------------------------------------------------------------

df = pd.read_csv(CSV_PATH)

check_df = df[
    df["qc_v2"].isin([
        "REVIEW",
        "REVIEW_BORDER",
        "REJECT"
    ])
].copy()

# Priority
priority_map = {
    "REJECT": 0,
    "REVIEW_BORDER": 1,
    "REVIEW": 2
}

check_df["priority"] = (
    check_df["qc_v2"].map(priority_map)
)

check_df = (
    check_df
    .sort_values(
        ["priority", "lesion_outside_pct"],
        ascending=[True, False]
    )
    .reset_index(drop=True)
)

print("=" * 80)
print("STATIC MANUAL VISUAL QC")
print("=" * 80)
print(
    f"Cases to inspect: {len(check_df)}"
)

# ------------------------------------------------------------
# IMAGE SEARCH
# ------------------------------------------------------------

IMAGE_EXTS = [
    ".jpg",
    ".jpeg",
    ".png",
    ".bmp",
    ".webp"
]


def find_original_image(filename, split):

    stem = Path(filename).stem

    # Expected location
    split_dir = (
        ROOT
        / str(split)
        / "images"
    )

    # Exact filename first
    exact = (
        split_dir
        / filename
    )

    if exact.exists():
        return exact

    # Search by stem
    for ext in IMAGE_EXTS:

        p = (
            split_dir
            / f"{stem}{ext}"
        )

        if p.exists():
            return p

    # Recursive fallback
    if split_dir.exists():

        matches = list(
            split_dir.rglob(
                f"{stem}.*"
            )
        )

        for p in matches:

            if (
                p.is_file()
                and p.suffix.lower()
                in IMAGE_EXTS
            ):
                return p

    return None


# ------------------------------------------------------------
# LESION MASK SEARCH
# ------------------------------------------------------------

def find_lesion_mask(filename, split):

    stem = Path(filename).stem

    mask_dir = (
        ROOT
        / str(split)
        / "lesion_masks"
    )

    if not mask_dir.exists():
        return None

    possible = [
        mask_dir / f"{stem}.png",
        mask_dir / f"{stem}.jpg",
        mask_dir / f"{stem}.jpeg",
        mask_dir / f"{stem}.bmp",
    ]

    for p in possible:

        if p.exists():
            return p

    matches = list(
        mask_dir.rglob(
            f"{stem}.*"
        )
    )

    for p in matches:

        if p.is_file():
            return p

    return None


# ------------------------------------------------------------
# LEAF MASK SEARCH
# ------------------------------------------------------------

def find_leaf_mask(row):

    # First use the exact path stored in CSV
    if "leaf_mask_path" in row.index:

        stored = row["leaf_mask_path"]

        if isinstance(stored, str):

            p = Path(stored)

            if p.exists():
                return p

    # Otherwise search generated leaf_masks
    stem = Path(
        row["filename"]
    ).stem

    mask_dir = (
        OUTPUT_DIR
        / "leaf_masks"
    )

    if mask_dir.exists():

        matches = list(
            mask_dir.rglob(
                f"{stem}_leaf.png"
            )
        )

        if matches:
            return matches[0]

    return None


# ------------------------------------------------------------
# TEST PATH DISCOVERY
# ------------------------------------------------------------

print("\nSearching image/mask files...")

records = []

for _, row in check_df.iterrows():

    image_path = find_original_image(
        row["filename"],
        row["split"]
    )

    lesion_path = find_lesion_mask(
        row["filename"],
        row["split"]
    )

    leaf_path = find_leaf_mask(
        row
    )

    records.append({
        "row": row,
        "image": image_path,
        "lesion": lesion_path,
        "leaf": leaf_path
    })


found_original = sum(
    x["image"] is not None
    for x in records
)

found_lesion = sum(
    x["lesion"] is not None
    for x in records
)

found_leaf = sum(
    x["leaf"] is not None
    for x in records
)

print(
    f"Original images found : "
    f"{found_original}/{len(records)}"
)

print(
    f"Lesion masks found    : "
    f"{found_lesion}/{len(records)}"
)

print(
    f"Leaf masks found      : "
    f"{found_leaf}/{len(records)}"
)

# ------------------------------------------------------------
# DISPLAY SETTINGS
# ------------------------------------------------------------

IMAGES_PER_PAGE = 6
COLS = 2

# ------------------------------------------------------------
# CREATE STATIC PAGES
# ------------------------------------------------------------

page_paths = []

for page_start in range(
    0,
    len(records),
    IMAGES_PER_PAGE
):

    page_records = records[
        page_start:
        page_start + IMAGES_PER_PAGE
    ]

    n = len(page_records)

    rows = math.ceil(
        n / COLS
    )

    fig, axes = plt.subplots(
        rows,
        COLS,
        figsize=(
            22,
            9 * rows
        )
    )

    axes = np.array(
        axes
    ).reshape(-1)

    for ax, item in zip(
        axes,
        page_records
    ):

        row = item["row"]

        image_path = item["image"]
        lesion_path = item["lesion"]
        leaf_path = item["leaf"]

        # ----------------------------------------------------
        # Read original
        # ----------------------------------------------------

        if image_path is None:

            ax.text(
                0.5,
                0.5,
                "ORIGINAL IMAGE NOT FOUND",
                ha="center",
                va="center",
                fontsize=14
            )

            ax.axis("off")
            continue

        img = cv2.imread(
            str(image_path)
        )

        if img is None:

            ax.text(
                0.5,
                0.5,
                "IMAGE READ ERROR",
                ha="center",
                va="center",
                fontsize=14
            )

            ax.axis("off")
            continue

        rgb = cv2.cvtColor(
            img,
            cv2.COLOR_BGR2RGB
        )

        # ----------------------------------------------------
        # Leaf mask
        # ----------------------------------------------------

        leaf = None

        if leaf_path is not None:

            leaf = cv2.imread(
                str(leaf_path),
                cv2.IMREAD_GRAYSCALE
            )

        # ----------------------------------------------------
        # Lesion mask
        # ----------------------------------------------------

        lesion = None

        if lesion_path is not None:

            lesion = cv2.imread(
                str(lesion_path),
                cv2.IMREAD_GRAYSCALE
            )

        # ----------------------------------------------------
        # Create overlay
        # ----------------------------------------------------

        overlay = rgb.copy()

        if leaf is not None:

            leaf_bool = leaf > 127

            boundary = cv2.morphologyEx(
                leaf.astype(np.uint8),
                cv2.MORPH_GRADIENT,
                np.ones(
                    (3, 3),
                    np.uint8
                )
            ) > 0

            # Green leaf boundary
            overlay[
                boundary
            ] = [0, 255, 0]

            # Red lesion
            if lesion is not None:

                lesion_bool = (
                    lesion > 127
                )

                lesion_layer = (
                    overlay.copy()
                )

                lesion_layer[
                    lesion_bool
                ] = [255, 0, 0]

                overlay = cv2.addWeighted(
                    overlay,
                    0.70,
                    lesion_layer,
                    0.30,
                    0
                )

                # redraw green boundary
                overlay[
                    boundary
                ] = [0, 255, 0]

        # ----------------------------------------------------
        # Create 4-panel image
        # ----------------------------------------------------

        panel = np.concatenate(
            [
                rgb,
                cv2.cvtColor(
                    lesion
                    if lesion is not None
                    else np.zeros(
                        rgb.shape[:2],
                        np.uint8
                    ),
                    cv2.COLOR_GRAY2RGB
                ),
                cv2.cvtColor(
                    leaf
                    if leaf is not None
                    else np.zeros(
                        rgb.shape[:2],
                        np.uint8
                    ),
                    cv2.COLOR_GRAY2RGB
                ),
                overlay
            ],
            axis=1
        )

        ax.imshow(
            panel
        )

        severity = float(
            row["severity_percent"]
        )

        outside = float(
            row["lesion_outside_pct"]
        )

        leaf_area = float(
            row["leaf_area_ratio"]
        ) * 100

        ax.set_title(
            f"{row['filename']}\n"
            f"{row['split']} | "
            f"QC={row['qc_v2']} | "
            f"Severity={severity:.2f}% | "
            f"Outside={outside:.3f}% | "
            f"Leaf={leaf_area:.2f}%",
            fontsize=10
        )

        ax.axis("off")

    # Hide unused axes
    for ax in axes[n:]:
        ax.axis("off")

    page_number = (
        page_start
        // IMAGES_PER_PAGE
    ) + 1

    plt.suptitle(
        f"V5 Manual QC — Page {page_number}\n"
        f"Left→Right inside each case: "
        f"Original | Lesion Mask | Leaf Mask | Overlay",
        fontsize=16
    )

    plt.tight_layout(
        rect=[
            0,
            0,
            1,
            0.96
        ]
    )

    page_path = (
        OUTPUT_DIR
        / "qc_v2"
        / "visual_review_static"
        / f"MANUAL_QC_PAGE_{page_number:02d}.jpg"
    )

    page_path.parent.mkdir(
        parents=True,
        exist_ok=True
    )

    plt.savefig(
        page_path,
        dpi=130,
        bbox_inches="tight"
    )

    plt.show()

    plt.close(fig)

    page_paths.append(
        page_path
    )


# ------------------------------------------------------------
# FINAL REPORT
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("MANUAL VISUAL QC PAGES CREATED")
print("=" * 80)

print(
    f"Pages: {len(page_paths)}"
)

for p in page_paths:
    print(p)
```

```python
from pathlib import Path
import pandas as pd
import shutil

SOURCE = Path("/content/Potato_Severity_Dataset")
V5 = SOURCE / "leaf_segmentation_v5"
FINAL = Path("/content/Potato_Severity_Dataset_FINAL")

CSV = V5 / "potato_severity_FINAL.csv"

df = pd.read_csv(CSV)

print("Final CSV rows:", len(df))

# ------------------------------------------------------------
# Resolve lesion mask using the ORIGINAL CSV path first
# ------------------------------------------------------------
def resolve_lesion_mask(row):
    p = row["lesion_mask_path"]

    if pd.isna(p):
        return None

    p = Path(str(p))

    # 1. Exact path stored in CSV
    if p.exists():
        return p

    # 2. Same basename inside the original split lesion_masks folder
    split = str(row["split"])
    candidate = SOURCE / split / "lesion_masks" / p.name

    if candidate.exists():
        return candidate

    # 3. Search by filename stem, allowing extension differences
    lesion_dir = SOURCE / split / "lesion_masks"

    if lesion_dir.exists():
        matches = list(lesion_dir.glob(p.stem + ".*"))
        if len(matches) == 1:
            return matches[0]

    return None


missing = []
copied = 0

for _, row in df.iterrows():

    split = str(row["split"])
    filename = str(row["filename"])

    src = resolve_lesion_mask(row)

    if src is None:
        missing.append(filename)
        continue

    # Preserve original lesion-mask filename
    dst = (
        FINAL
        / "segmentation"
        / split
        / "lesion_masks"
        / src.name
    )

    dst.parent.mkdir(parents=True, exist_ok=True)
    shutil.copy2(src, dst)

    copied += 1


print("\n" + "=" * 70)
print("LESION MASK COPY RESULT")
print("=" * 70)

print("Expected:", len(df))
print("Copied  :", copied)
print("Missing :", len(missing))

if missing:
    print("\nFirst missing records:")
    for x in missing[:20]:
        print(" -", x)

# ------------------------------------------------------------
# Integrity check
# ------------------------------------------------------------
assert copied == len(df), (
    f"Expected {len(df)} lesion masks, copied {copied}"
)

assert len(missing) == 0

print("\n✅ ALL 1153 LESION MASKS COPIED SUCCESSFULLY")
```

```python
import pandas as pd
from pathlib import Path
import shutil

# ============================================================
# FINAL GOOGLE DRIVE MASTER COPY
# Source:
#   images + lesion masks -> original dataset
#   leaf masks            -> V5 output
#   CSV                   -> final 1153-row CSV
# ============================================================

# ------------------------------------------------------------
# SOURCE
# ------------------------------------------------------------
ROOT = Path("/content/Potato_Severity_Dataset")
V5 = ROOT / "leaf_segmentation_v5"

FINAL_CSV = V5 / "potato_severity_FINAL.csv"

# ------------------------------------------------------------
# DRIVE DESTINATION
# ------------------------------------------------------------
DEST = Path(
    "/content/drive/MyDrive/"
    "Research/Potato_Late_Blight_Project/"
    "Potato_Severity_Dataset_FINAL"
)

# ------------------------------------------------------------
# LOAD FINAL CSV
# ------------------------------------------------------------
assert FINAL_CSV.exists(), f"Final CSV not found: {FINAL_CSV}"

df = pd.read_csv(FINAL_CSV)

assert len(df) == 1153, (
    f"Expected 1153 rows, got {len(df)}"
)

print("=" * 70)
print("FINAL GOOGLE DRIVE MASTER COPY")
print("=" * 70)

print("CSV rows:", len(df))

# ------------------------------------------------------------
# REMOVE OLD INCOMPLETE DESTINATION
# ------------------------------------------------------------
if DEST.exists():
    print("\nRemoving previous incomplete Drive package:")
    print(DEST)
    shutil.rmtree(DEST)

# ------------------------------------------------------------
# CREATE DIRECTORY STRUCTURE
# ------------------------------------------------------------
for split in ["train", "valid", "test"]:

    for kind in [
        "images",
        "lesion_masks",
        "leaf_masks",
    ]:
        (
            DEST
            / "segmentation"
            / split
            / kind
        ).mkdir(
            parents=True,
            exist_ok=True
        )

(DEST / "severity_dataset").mkdir(
    parents=True,
    exist_ok=True
)

(DEST / "qc").mkdir(
    parents=True,
    exist_ok=True
)

# ------------------------------------------------------------
# COPY FINAL CSV
# ------------------------------------------------------------
shutil.copy2(
    FINAL_CSV,
    DEST / "severity_dataset" / "potato_severity_FINAL.csv"
)

# ------------------------------------------------------------
# COPY QC FILES
# ------------------------------------------------------------
for filename in [
    "v5_FINAL_KEEP.csv",
    "v5_FINAL_ALL_WITH_DECISION.csv",
    "v5_FINAL_REJECTED.csv",
]:

    src = V5 / filename

    if src.exists():
        shutil.copy2(
            src,
            DEST / "qc" / filename
        )

# ------------------------------------------------------------
# COPY RECORD FILES
# ------------------------------------------------------------

missing_images = []
missing_lesions = []
missing_leaves = []

for i, row in df.iterrows():

    split = str(row["split"])
    filename = str(row["filename"])

    assert split in {
        "train",
        "valid",
        "test"
    }

    # ========================================================
    # IMAGE
    # ========================================================

    src_image = (
        ROOT
        / split
        / "images"
        / filename
    )

    if not src_image.exists():
        missing_images.append(filename)
    else:
        shutil.copy2(
            src_image,
            DEST
            / "segmentation"
            / split
            / "images"
            / filename
        )

    # ========================================================
    # LESION MASK
    # ========================================================

    # Use the actual lesion_mask_path from the FINAL CSV
    lesion_path = row["lesion_mask_path"]

    if pd.isna(lesion_path):
        missing_lesions.append(filename)
    else:

        lesion_name = Path(
            str(lesion_path)
        ).name

        src_lesion = (
            ROOT
            / split
            / "lesion_masks"
            / lesion_name
        )

        if not src_lesion.exists():

            # fallback search by stem
            candidates = list(
                (
                    ROOT
                    / split
                    / "lesion_masks"
                ).glob(
                    Path(lesion_name).stem + ".*"
                )
            )

            if len(candidates) == 1:
                src_lesion = candidates[0]

        if not src_lesion.exists():
            missing_lesions.append(filename)

        else:
            shutil.copy2(
                src_lesion,
                DEST
                / "segmentation"
                / split
                / "lesion_masks"
                / src_lesion.name
            )

    # ========================================================
    # LEAF MASK
    # ========================================================

    leaf_path_str = row["leaf_mask_path"]

    if pd.isna(leaf_path_str):

        missing_leaves.append(filename)

    else:

        src_leaf = Path(leaf_path_str)

        if not src_leaf.exists():

            missing_leaves.append(filename)

        else:

            shutil.copy2(
                src_leaf,
                DEST
                / "segmentation"
                / split
                / "leaf_masks"
                / src_leaf.name
            )

# ------------------------------------------------------------
# REPORT COPY RESULT
# ------------------------------------------------------------
print("\n" + "=" * 70)
print("COPY RESULT")
print("=" * 70)

print("Missing images     :", len(missing_images))
print("Missing lesion     :", len(missing_lesions))
print("Missing leaf masks :", len(missing_leaves))

if missing_images:
    print("\nMissing images:")
    for x in missing_images[:10]:
        print(" -", x)

if missing_lesions:
    print("\nMissing lesion masks:")
    for x in missing_lesions[:10]:
        print(" -", x)

if missing_leaves:
    print("\nMissing leaf masks:")
    for x in missing_leaves[:10]:
        print(" -", x)

# ------------------------------------------------------------
# HARD ASSERTIONS
# ------------------------------------------------------------
assert len(missing_images) == 0
assert len(missing_lesions) == 0
assert len(missing_leaves) == 0

# ------------------------------------------------------------
# VERIFY EACH SPLIT
# ------------------------------------------------------------
print("\n" + "=" * 70)
print("DRIVE MASTER VERIFICATION")
print("=" * 70)

total_images = 0
total_lesions = 0
total_leaves = 0

for split in [
    "train",
    "valid",
    "test"
]:

    expected = len(
        df[df["split"] == split]
    )

    n_images = len([
        x for x in
        (
            DEST
            / "segmentation"
            / split
            / "images"
        ).iterdir()
        if x.is_file()
    ])

    n_lesions = len([
        x for x in
        (
            DEST
            / "segmentation"
            / split
            / "lesion_masks"
        ).iterdir()
        if x.is_file()
    ])

    n_leaves = len([
        x for x in
        (
            DEST
            / "segmentation"
            / split
            / "leaf_masks"
        ).iterdir()
        if x.is_file()
    ])

    print(f"\n{split.upper()}")
    print("CSV records :", expected)
    print("Images      :", n_images)
    print("Lesion masks:", n_lesions)
    print("Leaf masks  :", n_leaves)

    assert n_images == expected
    assert n_lesions == expected
    assert n_leaves == expected

    total_images += n_images
    total_lesions += n_lesions
    total_leaves += n_leaves

# ------------------------------------------------------------
# FINAL CHECK
# ------------------------------------------------------------
assert len(df) == 1153
assert total_images == 1153
assert total_lesions == 1153
assert total_leaves == 1153

print("\n" + "=" * 70)
print("✅ GOOGLE DRIVE MASTER COPY VERIFIED")
print("=" * 70)

print("CSV rows     :", len(df))
print("Images       :", total_images)
print("Lesion masks :", total_lesions)
print("Leaf masks   :", total_leaves)
print("MASTER PATH  :", DEST)
```
