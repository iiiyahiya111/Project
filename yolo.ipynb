```python
from ultralytics import YOLO
import matplotlib.pyplot as plt
import cv2
import glob

# Load the best trained model
model = YOLO('/content/drive/MyDrive/yolo_runs/erwiam_gpu-2/weights/best.pt')

# Get a list of image files from the test set
test_images_path = '/content/yolo_erwiam/images/test/'
image_files = glob.glob(test_images_path + '*.jpg') + glob.glob(test_images_path + '*.png')

# Take the first few images for demonstration
num_images_to_show = min(5, len(image_files))
selected_images = image_files[:num_images_to_show]

for img_path in selected_images:
    print(f'Predicting on: {img_path}')
    # Run inference
    results = model.predict(source=img_path, conf=0.25) # Adjust confidence threshold as needed

    # Plot results if available
    if results:
        # The `plot()` method returns an image with detections drawn on it
        # It's usually a numpy array in BGR format
        im_bgr = results[0].plot()
        im_rgb = cv2.cvtColor(im_bgr, cv2.COLOR_BGR2RGB)

        plt.figure(figsize=(10, 10))
        plt.imshow(im_rgb)
        plt.axis('off')
        plt.title(f'Prediction for {img_path.split('/')[-1]}')
        plt.show()
    else:
        print(f"No detections for {img_path}")
```

```python
from ultralytics import YOLO
import matplotlib.pyplot as plt
import cv2
import glob
import os

# Load the best trained model
model = YOLO('/content/drive/MyDrive/yolo_runs/erwiam_gpu-2/weights/best.pt')

# Get a list of image files from the test set
test_images_path = '/content/yolo_erwiam/images/test/'

# Use os.listdir and filter for image files explicitly
image_filenames = [f for f in os.listdir(test_images_path) if f.lower().endswith(('.png', '.jpg', '.jpeg', '.gif', '.bmp'))]
image_files = [os.path.join(test_images_path, f) for f in image_filenames]

# Take the first few images for demonstration
num_images_to_show = min(5, len(image_files))
selected_images = image_files[:num_images_to_show]

if not selected_images:
    print(f"No image files found in {test_images_path} with common image extensions.")
else:
    for img_path in selected_images:
        print(f'Predicting on: {img_path}')
        # Run inference
        results = model.predict(source=img_path, conf=0.25) # Adjust confidence threshold as needed

        # Plot results if available
        if results and results[0].boxes.xywhn.numel() > 0: # Check if any detections are made
            # The `plot()` method returns an image with detections drawn on it
            # It's usually a numpy array in BGR format
            im_bgr = results[0].plot()
            im_rgb = cv2.cvtColor(im_bgr, cv2.COLOR_BGR2RGB)

            plt.figure(figsize=(10, 10))
            plt.imshow(im_rgb)
            plt.axis('off')
            plt.title(f'Prediction for {img_path.split('/')[-1]}')
            plt.show()
        else:
            print(f"No detections for {img_path} at confidence 0.25, or no results object.")
```
